### 2.2 Uneven Subplot Layouts

Use `GridSpec` for complex layouts.

```python
from matplotlib.gridspec import GridSpec

# Create figure with GridSpec
fig = plt.figure(figsize=(12, 8))
gs = GridSpec(2, 3, figure=fig)

# Large plot spanning 2 columns
ax1 = fig.add_subplot(gs[0, :2])  # Row 0, columns 0-1
ax1.plot(np.random.randn(100).cumsum())
ax1.set_title('Main Plot (spans 2 columns)')

# Small plot in top-right
ax2 = fig.add_subplot(gs[0, 2])  # Row 0, column 2
ax2.bar(['A', 'B', 'C'], [10, 20, 15])
ax2.set_title('Bar Chart')

# Three plots on bottom row
ax3 = fig.add_subplot(gs[1, 0])
ax3.hist(np.random.randn(100), bins=15)
ax3.set_title('Histogram')

ax4 = fig.add_subplot(gs[1, 1])
ax4.scatter(np.random.rand(30), np.random.rand(30))
ax4.set_title('Scatter')

ax5 = fig.add_subplot(gs[1, 2])
ax5.pie([30, 40, 30], labels=['X', 'Y', 'Z'])
ax5.set_title('Pie Chart')

plt.tight_layout()
plt.show()
```

### Best Practices for Twin Axes

| Do | Don't |
|-----|-------|
| Use different colors for each axis | Use similar colors |
| Label axes clearly | Leave axes unlabeled |
| Use when scales are very different | Use for similar scales |
| Consider if separate plots would be clearer | Overcrowd with too many lines |

## 6. 3D Visualization

**When to use 3D:**
- You have 3 continuous variables
- Spatial data
- Interactive exploration

**Warning:** 3D is often harder to read than 2D. Use sparingly!

### 6.1 3D Scatter Plot

```python
# 3D scatter with Plotly
iris = px.data.iris()

fig = px.scatter_3d(iris,
                    x='sepal_length',
                    y='sepal_width',
                    z='petal_length',
                    color='species',
                    title='Iris Dataset in 3D')

fig.show()
```

**Try it:** Click and drag to rotate the view!

### 6.2 3D Surface Plot

```python
# 3D surface
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y, colorscale='Viridis')])

fig.update_layout(
    title='3D Surface Plot',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    )
)

fig.show()
```

### When NOT to Use 3D

| Avoid 3D When | Better Alternative |
|---------------|--------------------|
| You only have 2 variables | 2D scatter |
| Comparison is important | Faceted 2D plots |
| Exact values matter | Tables or bar charts |
| Static presentation | 2D with color for 3rd dimension |

## 8. When to Use Interactive vs Static

| Use Interactive (Plotly) | Use Static (Matplotlib/Seaborn) |
|--------------------------|----------------------------------|
| Web dashboards | Academic papers |
| Exploratory analysis | Printed reports |
| Presentations | Journal submissions |
| Large datasets | Small, focused datasets |
| When hover info helps | When every value is labeled |
