## 1. Introduction to Outliers

### What is an Outlier?

An **outlier** is a data point that differs significantly from other observations. Outliers can be:

| Type | Description | Example |
|------|-------------|--------|
| **Valid extreme values** | Legitimate but rare | A billionaire in income data |
| **Measurement errors** | Mistakes in data collection | Typo: age = 500 |
| **Data entry errors** | Wrong values entered | Height in feet instead of cm |
| **Sampling issues** | Different population | Including commercial flights in personal travel data |

### Why Do Outliers Matter?

1. **Statistical Impact**: Mean is heavily influenced by outliers
2. **Model Performance**: Many ML models are sensitive to outliers
3. **Visualization**: Can compress the scale and hide patterns
4. **Business Decisions**: May lead to wrong conclusions

**Formula:**
- IQR = Q3 - Q1
- Lower bound = Q1 - 1.5 × IQR
- Upper bound = Q3 + 1.5 × IQR

### 4.2 Z-Score Method

The Z-score measures how many standard deviations a value is from the mean.

**Formula:** Z = (X - μ) / σ

**Common thresholds:**
- |Z| > 3: Outlier (common choice)
- |Z| > 2: Potential outlier (more conservative)

```python
# ═══ CHECKPOINT — "Comparison: IQR vs Z-Score" ═══════════════════════════════════════════════
#
# Run this ONLY if you are behind, or a cell above threw an error and the rest
# stopped working. It rebuilds everything this section needs, so you can rejoin
# here instead of being stuck for the rest of the class.
#
# Running it when you are NOT stuck is harmless. If it leaves you stuck anyway,
# that is a bug in the template, not in your code — say so.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
df = pd.read_csv('../../00-data/Divorce Data Analysis.csv')
numerical_cols = ['income_man', 'income_woman', 'marriage_duration', 'num_kids']
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
for i, col in enumerate(numerical_cols):
    ax = axes[i // 2, i % 2]
    data = df[col].dropna()
    ax.hist(data, bins=50, edgecolor='black', alpha=0.7)
    ax.axvline(data.mean(), color='red', linestyle='--', label=f'Mean: {data.mean():.1f}')
    ax.axvline(data.median(), color='green', linestyle='--', label=f'Median: {data.median():.1f}')
    ax.set_title(f'{col}\nSkewness: {data.skew():.2f}')
    ax.legend()
max_val = max(df['income_man'].max(), df['income_woman'].max())
def detect_outliers_iqr(data, column):
    """
    Detect outliers using the IQR method.
    Returns the outlier bounds and count.
    """
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return {'column': column, 'Q1': Q1, 'Q3': Q3, 'IQR': IQR, 'lower_bound': lower_bound, 'upper_bound': upper_bound, 'outlier_count': len(outliers), 'outlier_pct': len(outliers) / len(data) * 100}
for col in numerical_cols:
    result = detect_outliers_iqr(df, col)
    print(f'\n{col}:')
    print(f"  Bounds: [{result['lower_bound']:.2f}, {result['upper_bound']:.2f}]")
    print(f"  Outliers: {result['outlier_count']} ({result['outlier_pct']:.2f}%)")
def detect_outliers_zscore(data, column, threshold=3):
    """
    Detect outliers using the Z-score method.
    Default threshold is 3 standard deviations.
    """
    z_scores = np.abs(stats.zscore(data[column].dropna()))
    outlier_count = np.sum(z_scores > threshold)
    outlier_pct = outlier_count / len(z_scores) * 100
    return {'column': column, 'threshold': threshold, 'outlier_count': outlier_count, 'outlier_pct': outlier_pct}
for col in numerical_cols:
    result = detect_outliers_zscore(df, col)
    print(f"{col}: {result['outlier_count']} outliers ({result['outlier_pct']:.2f}%)")

print("Checkpoint restored. Ready for: Comparison: IQR vs Z-Score")
```

### 4.3 Comparison: IQR vs Z-Score

| Method | Best For | Limitations |
|--------|----------|-------------|
| **IQR** | Skewed data, robust | May flag too many in normal data |
| **Z-Score** | Normal distributions | Sensitive to outliers themselves |

**Recommendation:** Use IQR for initial exploration, Z-score when you know data is approximately normal.

## 6. Data Transformations

Transformations can reduce skewness and make data more suitable for modeling.

### 6.1 Log Transformation

Best for **right-skewed** data (long tail to the right).

**When to use:**
- Data spans multiple orders of magnitude
- Positive skewness
- Multiplicative relationships

```python
# Log transformation of income_man
# Add 1 to handle zeros: log(x + 1)
df['income_man_log'] = np.log1p(df['income_man'])

# Compare distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Original
axes[0].hist(df['income_man'].dropna(), bins=50, edgecolor='black', alpha=0.7)
axes[0].set_title(f'Original Income\nSkewness: {df["income_man"].skew():.2f}')
axes[0].set_xlabel('Income')

# Log transformed
axes[1].hist(df['income_man_log'].dropna(), bins=50, edgecolor='black', alpha=0.7, color='green')
axes[1].set_title(f'Log-Transformed Income\nSkewness: {df["income_man_log"].skew():.2f}')
axes[1].set_xlabel('log(Income + 1)')

plt.suptitle('Effect of Log Transformation', fontsize=14)
plt.tight_layout()
plt.show()

print(f"Skewness reduced from {df['income_man'].skew():.2f} to {df['income_man_log'].skew():.2f}")
```

### 6.2 Square Root Transformation

### 6.3 Box-Cox Transformation

Automatically finds the optimal power transformation to make data more normal.

**Requirements:** Data must be strictly positive (> 0)

```python
from scipy.stats import boxcox

# Box-Cox requires positive values
income_positive = df['income_man'].dropna()
income_positive = income_positive[income_positive > 0]

# Apply Box-Cox
income_boxcox, lambda_param = boxcox(income_positive)

print(f"Optimal lambda: {lambda_param:.4f}")
print(f"\nInterpretation:")
print(f"  lambda = 0: log transform")
print(f"  lambda = 0.5: sqrt transform")
print(f"  lambda = 1: no transform")

# Compare
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(income_positive, bins=50, edgecolor='black', alpha=0.7)
axes[0].set_title(f'Original\nSkewness: {income_positive.skew():.2f}')

axes[1].hist(income_boxcox, bins=50, edgecolor='black', alpha=0.7, color='purple')
axes[1].set_title(f'Box-Cox (λ={lambda_param:.2f})\nSkewness: {pd.Series(income_boxcox).skew():.2f}')

plt.tight_layout()
plt.show()
```

### 6.4 When to Use Which Transformation

| Skewness | Recommended Transform |
|----------|----------------------|
| Moderate (0.5 - 1) | Square root |
| High (1 - 2) | Log |
| Very high (> 2) | Box-Cox or inverse |
| Left-skewed | Square or exponential |

## 8. Summary

### Key Takeaways

1. **Detection Methods:**
   - Visual: Box plots, histograms, scatter plots
   - Statistical: IQR method, Z-score method

2. **Handling Strategies:**
   - Remove: For errors or when sample is large
   - Cap/Winsorize: Preserves sample size
   - Transform: Reduces impact without removal
   - Flag: Keep for separate analysis

3. **Transformations:**
   - Log: For right-skewed data
   - Square root: For moderate skewness
   - Box-Cox: Automatic optimal transformation

4. **Best Practices:**
   - Always visualize before and after
   - Document your decisions
   - Consider the impact on interpretability
   - Don't automatically remove all outliers
