# Module 03a: Regular Expressions (RegEx) Fundamentals

## What is RegEx?

**Regular Expressions (RegEx)** are patterns used to match, search, and manipulate text. Think of them as a powerful "Find & Replace" tool on steroids.

### Why Learn RegEx for Data Science?

| Use Case | Example |
|----------|--------|
| **Data Cleaning** | Remove special characters, standardize formats |
| **Data Extraction** | Pull emails, phone numbers, dates from text |
| **Data Validation** | Check if values match expected patterns |
| **Text Preprocessing** | Clean text for NLP tasks |

### Real-World Examples

- Extract all email addresses from customer feedback
- Validate phone numbers in a dataset
- Clean messy date formats ("Jan 1, 2024" → "2024-01-01")
- Remove HTML tags from scraped web data

## 1. Basic Matching - Literal Text

The simplest regex is just literal text - it matches exactly what you type.

### `re.findall()` - Find All Matches

Returns a list of all matches found in the text.

````python
text = "I love Python. Python is great for data science. python rules!"

# Find all occurrences of "Python" (exact match)
matches = re.findall("Python", text)
print("Matches:", matches)
print("Count:", len(matches))
````

**Notice:** It only found 2 matches, not 3. Why? Because "python" (lowercase) didn't match!

### Case-Insensitive Matching with `re.IGNORECASE`

````python
# Add the IGNORECASE flag to match regardless of case
matches = re.findall("python", text, re.IGNORECASE)
print("Case-insensitive matches:", matches)
````

### Raw Strings: Always Use `r"pattern"`

In Python, backslashes have special meaning. Use raw strings to avoid confusion.

````python
# Without raw string - \n means newline
print("Without raw string:", "\n")

# With raw string - \n is literal backslash-n
print("With raw string:", r"\n")

# Always use raw strings for regex patterns!
pattern = r"\d+"  # This is the correct way
````

## 2. Character Classes - Match Types of Characters

Instead of matching specific characters, match *categories* of characters.

### Common Character Classes

| Pattern | Meaning | Example Match |
|---------|---------|---------------|
| `\d` | Any digit (0-9) | "5", "0" |
| `\D` | Any NON-digit | "a", "@" |
| `\w` | Word character (a-z, A-Z, 0-9, _) | "a", "Z", "5" |
| `\W` | NON-word character | "@", " ", "!" |
| `\s` | Whitespace (space, tab, newline) | " ", "\t" |
| `\S` | NON-whitespace | "a", "5" |
| `.` | Any character (except newline) | "a", "5", "@" |

````python
sample = "Order #12345 on 2024-03-15. Contact: john@email.com"

# Find all digits
digits = re.findall(r"\d", sample)
print("All digits:", digits)

# Find all word characters
words = re.findall(r"\w", sample)
print("Word characters (first 10):", words[:10])
````

### Custom Character Classes with `[ ]`

Create your own set of characters to match.

````python
text = "The quick brown fox jumps over the lazy dog"

# Match only vowels
vowels = re.findall(r"[aeiou]", text)
print("Vowels:", vowels)

# Match only uppercase letters
uppercase = re.findall(r"[A-Z]", text)
print("Uppercase:", uppercase)

# Match digits and letters only
alphanumeric = re.findall(r"[a-zA-Z0-9]", "Hello123!@#")
print("Alphanumeric:", alphanumeric)
````

### Negated Character Classes with `[^ ]`

Match anything EXCEPT the specified characters.

````python
# Match anything that is NOT a digit
non_digits = re.findall(r"[^0-9]", "ABC123XYZ")
print("Non-digits:", non_digits)

# Match anything that is NOT a vowel
non_vowels = re.findall(r"[^aeiouAEIOU]", "Hello")
print("Non-vowels:", non_vowels)
````

## 3. Quantifiers - How Many Times to Match

Quantifiers specify how many times a character/pattern should repeat.

| Quantifier | Meaning | Example |
|------------|---------|--------|
| `*` | 0 or more | `a*` matches "", "a", "aaa" |
| `+` | 1 or more | `a+` matches "a", "aaa" (not "") |
| `?` | 0 or 1 (optional) | `a?` matches "", "a" |
| `{n}` | Exactly n times | `a{3}` matches "aaa" |
| `{n,}` | n or more times | `a{2,}` matches "aa", "aaa" |
| `{n,m}` | Between n and m times | `a{2,4}` matches "aa", "aaa", "aaaa" |

````python
sample = "Order #12345 on 2024-03-15. Call 555-1234 or 555-12345678"

# \d+ means "one or more digits" - much more useful!
numbers = re.findall(r"\d+", sample)
print("All numbers:", numbers)

# Exactly 4 digits
four_digits = re.findall(r"\d{4}", sample)
print("Exactly 4 digits:", four_digits)

# 2 to 4 digits
two_to_four = re.findall(r"\d{2,4}", sample)
print("2-4 digits:", two_to_four)
````

### Combining Character Classes with Quantifiers

````python
text = "Contact: john@email.com, jane_doe@company.co.uk, support@help.org"

# \w+ means "one or more word characters" - matches words
words = re.findall(r"\w+", text)
print("Words:", words)

# Find patterns that look like email parts
# \w+ @ \w+ . \w+
emails = re.findall(r"\w+@\w+\.\w+", text)
print("Simple email pattern:", emails)
````

## 4. Anchors - Match Positions

Anchors don't match characters - they match *positions* in the text.

| Anchor | Meaning |
|--------|--------|
| `^` | Start of string/line |
| `$` | End of string/line |
| `\b` | Word boundary |

````python
text = "Python is great. I love Python."

# Match "Python" only at the START
at_start = re.findall(r"^Python", text)
print("At start:", at_start)

# Match "Python" only at the END (with period)
at_end = re.findall(r"Python\.$", text)
print("At end:", at_end)
````

### Word Boundaries `\b`

Match the position between a word character and a non-word character.

````python
text = "The cat scattered the category catalog"

# Without word boundary - matches "cat" inside other words too
no_boundary = re.findall(r"cat", text)
print("Without boundary:", no_boundary)

# With word boundary - matches only the standalone word "cat"
with_boundary = re.findall(r"\bcat\b", text)
print("With boundary:", with_boundary)
````

## 5. Groups - Capture Parts of a Match

Use parentheses `( )` to capture specific parts of your match.

### Basic Capturing Groups

````python
text = "Date: 2024-03-15, Time: 14:30"

# Without groups - returns whole match
dates = re.findall(r"\d{4}-\d{2}-\d{2}", text)
print("Full date:", dates)

# With groups - returns captured parts as tuples
date_parts = re.findall(r"(\d{4})-(\d{2})-(\d{2})", text)
print("Date parts (year, month, day):", date_parts)
````

````python
# Extract username and domain from emails
emails = "Contact: john@gmail.com, jane@company.org"

# Capture username and domain separately
pattern = r"(\w+)@([\w.]+)"
matches = re.findall(pattern, emails)

print("Email parts:")
for username, domain in matches:
    print(f"  Username: {username}, Domain: {domain}")
````

### Named Groups `(?P<name>...)`

Give your groups meaningful names for better readability.

````python
text = "Order #12345 placed on 2024-03-15"

# Named groups for clarity
pattern = r"Order #(?P<order_id>\d+) placed on (?P<date>\d{4}-\d{2}-\d{2})"

match = re.search(pattern, text)
if match:
    print("Order ID:", match.group('order_id'))
    print("Date:", match.group('date'))
    print("All groups:", match.groupdict())
````

## 6. Core `re` Functions

### Function Comparison

| Function | Purpose | Returns |
|----------|---------|--------|
| `re.search()` | Find first match anywhere | Match object or None |
| `re.match()` | Find match at START only | Match object or None |
| `re.findall()` | Find all matches | List of strings/tuples |
| `re.finditer()` | Find all matches | Iterator of match objects |
| `re.sub()` | Replace matches | New string |
| `re.split()` | Split on pattern | List of strings |

### `re.search()` vs `re.match()`

````python
text = "Hello Python World"

# search() finds pattern ANYWHERE in the string
search_result = re.search(r"Python", text)
print("search() found:", search_result.group() if search_result else "Not found")

# match() only finds pattern at the START
match_result = re.match(r"Python", text)
print("match() found:", match_result.group() if match_result else "Not found")

# match() works if pattern is at start
match_hello = re.match(r"Hello", text)
print("match('Hello'):", match_hello.group() if match_hello else "Not found")
````

### `re.sub()` - Search and Replace

This is extremely useful for data cleaning!

````python
# Basic replacement
text = "The price is $100 and $200"
cleaned = re.sub(r"\$", "USD ", text)
print("Replace $:", cleaned)

# Remove all digits
text = "Order123 has 5 items"
no_digits = re.sub(r"\d+", "", text)
print("Remove digits:", no_digits)

# Clean extra whitespace (common data cleaning task!)
messy = "Too    many     spaces    here"
clean = re.sub(r"\s+", " ", messy)
print("Clean spaces:", clean)
````

### Using Groups in Replacement

````python
# Reformat dates: DD/MM/YYYY → YYYY-MM-DD
dates = "15/03/2024 and 20/12/2023"

# Capture day, month, year and rearrange
reformatted = re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\3-\2-\1", dates)
print("Reformatted dates:", reformatted)

# Hide email addresses
emails = "Contact john@gmail.com or jane@company.org"
hidden = re.sub(r"(\w+)@([\w.]+)", r"***@\2", emails)
print("Hidden emails:", hidden)
````

### `re.split()` - Split on Pattern

Split text on complex patterns (more powerful than `str.split()`).

````python
# Split on multiple delimiters
text = "apple,banana;cherry orange|grape"

# Regular split only works on one delimiter
print("str.split(','):", text.split(','))

# Regex split on multiple delimiters
parts = re.split(r"[,;|\s]+", text)
print("re.split():", parts)
````

````python
# Split on one or more whitespace characters
text = "Name:   John    Age:  30   City: NYC"
parts = re.split(r"\s+", text)
print("Split on whitespace:", parts)
````

## Quick Reference: Common Patterns for Data Cleaning

| Task | Pattern | Example |
|------|---------|--------|
| All digits | `\d+` | "123", "45" |
| All words | `\w+` | "Hello", "World" |
| Email | `[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}` | "test@email.com" |
| Phone (US) | `\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}` | "(555) 123-4567" |
| Date YYYY-MM-DD | `\d{4}-\d{2}-\d{2}` | "2024-03-15" |
| URL | `https?://[\w./%-]+` | "https://example.com" |
| Extra whitespace | `\s+` | Multiple spaces |
| Non-alphanumeric | `[^a-zA-Z0-9\s]` | Special characters |

````python
# Cheat sheet in code form
patterns = {
    "digits": r"\d+",
    "words": r"\w+",
    "email": r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}",
    "date_ymd": r"\d{4}-\d{2}-\d{2}",
    "url": r"https?://[\w./%-]+",
}

sample = "Visit https://example.com on 2024-03-15. Contact: test@email.com, Order #12345"

print("Sample text:", sample)
print("\nMatches:")
for name, pattern in patterns.items():
    matches = re.findall(pattern, sample)
    print(f"  {name}: {matches}")
````
