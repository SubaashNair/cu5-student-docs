## Quick Reference

### Character Classes

| Pattern | Matches |
|---------|---------|
| `\d` | Any digit (0-9) |
| `\D` | Any NON-digit |
| `\w` | Word character (a-z, A-Z, 0-9, _) |
| `\W` | NON-word character |
| `\s` | Whitespace |
| `\S` | NON-whitespace |
| `.` | Any character (except newline) |
| `[abc]` | Any of a, b, c |
| `[^abc]` | NOT a, b, or c |

### Quantifiers

| Pattern | Meaning |
|---------|---------|
| `*` | 0 or more |
| `+` | 1 or more |
| `?` | 0 or 1 (optional) |
| `{n}` | Exactly n |
| `{n,m}` | Between n and m |

### Anchors & Groups

| Pattern | Meaning |
|---------|---------|
| `^` | Start of string |
| `$` | End of string |
| `\b` | Word boundary |
| `(...)` | Capturing group |
| `(?P<name>...)` | Named group |

### Key Functions

| Function | Use For |
|----------|---------|
| `re.findall()` | Extract all matches |
| `re.search()` | Find first match |
| `re.match()` | Match at start only |
| `re.sub()` | Replace matches |
| `re.split()` | Split by pattern |
