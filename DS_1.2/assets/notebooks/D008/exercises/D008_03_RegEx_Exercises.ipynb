- Extension: 2-4 letters

```python
# Q5: YOUR CODE HERE
pattern_email = r''  # Valid email pattern
```

```python
# Test Q5
emails = [
    'user@example.com',        # Valid
    'john.doe@company.org',    # Valid
    'test_123@mail.co.uk',     # Valid
    'invalid-email',           # Invalid - no @
    '@nodomain.com',           # Invalid - no username
    'user@.com',               # Invalid - no domain
    'user@domain'              # Invalid - no extension
]

print("Email validation:")
for email in emails:
    is_valid = bool(re.match(pattern_email, email))
    status = "Valid" if is_valid else "Invalid"
    print(f"  {email}: {status}")

# Expected: First 3 valid, last 4 invalid
```
