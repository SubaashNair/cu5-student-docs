# Module 03a: Regular Expressions (RegEx) Fundamentals

## What is RegEx?

**Regular Expressions (RegEx)** are patterns used to match, search, and manipulate text. Think of them as a powerful "Find & Replace" tool on steroids.

### Why Learn RegEx for Data Science?

| Use Case | Example |
|----------|--------|
| **Data Cleaning** | Remove special characters, standardize formats |
| **Data Extraction** | Pull emails, phone numbers, dates from text |
| **Data Validation** | Check if values match expected patterns |
| **Text Preprocessing** | Clean text for NLP tasks |

### Real-World Examples

- Extract all email addresses from customer feedback
- Validate phone numbers in a dataset
- Clean messy date formats ("Jan 1, 2024" → "2024-01-01")
- Remove HTML tags from scraped web data

## 3. Quantifiers - How Many Times to Match

Quantifiers specify how many times a character/pattern should repeat.

| Quantifier | Meaning | Example |
|------------|---------|--------|
| `*` | 0 or more | `a*` matches "", "a", "aaa" |
| `+` | 1 or more | `a+` matches "a", "aaa" (not "") |
| `?` | 0 or 1 (optional) | `a?` matches "", "a" |
| `{n}` | Exactly n times | `a{3}` matches "aaa" |
| `{n,}` | n or more times | `a{2,}` matches "aa", "aaa" |
| `{n,m}` | Between n and m times | `a{2,4}` matches "aa", "aaa", "aaaa" |

### Word Boundaries `\b`

Match the position between a word character and a non-word character.

````python
text = "The cat scattered the category catalog"

# Without word boundary - matches "cat" inside other words too
no_boundary = re.findall(r"cat", text)
print("Without boundary:", no_boundary)

# With word boundary - matches only the standalone word "cat"
with_boundary = re.findall(r"\bcat\b", text)
print("With boundary:", with_boundary)
````

## 5. Groups - Capture Parts of a Match

Use parentheses `( )` to capture specific parts of your match.

### Basic Capturing Groups

````python
text = "Date: 2024-03-15, Time: 14:30"

# Without groups - returns whole match
dates = re.findall(r"\d{4}-\d{2}-\d{2}", text)
print("Full date:", dates)

# With groups - returns captured parts as tuples
date_parts = re.findall(r"(\d{4})-(\d{2})-(\d{2})", text)
print("Date parts (year, month, day):", date_parts)
````

## 6. Core `re` Functions

### Function Comparison

| Function | Purpose | Returns |
|----------|---------|--------|
| `re.search()` | Find first match anywhere | Match object or None |
| `re.match()` | Find match at START only | Match object or None |
| `re.findall()` | Find all matches | List of strings/tuples |
| `re.finditer()` | Find all matches | Iterator of match objects |
| `re.sub()` | Replace matches | New string |
| `re.split()` | Split on pattern | List of strings |

### `re.search()` vs `re.match()`

````python
text = "Hello Python World"

# search() finds pattern ANYWHERE in the string
search_result = re.search(r"Python", text)
print("search() found:", search_result.group() if search_result else "Not found")

# match() only finds pattern at the START
match_result = re.match(r"Python", text)
print("match() found:", match_result.group() if match_result else "Not found")

# match() works if pattern is at start
match_hello = re.match(r"Hello", text)
print("match('Hello'):", match_hello.group() if match_hello else "Not found")
````

### `re.sub()` - Search and Replace

This is extremely useful for data cleaning!

````python
# Basic replacement
text = "The price is $100 and $200"
cleaned = re.sub(r"\$", "USD ", text)
print("Replace $:", cleaned)

# Remove all digits
text = "Order123 has 5 items"
no_digits = re.sub(r"\d+", "", text)
print("Remove digits:", no_digits)

# Clean extra whitespace (common data cleaning task!)
messy = "Too    many     spaces    here"
clean = re.sub(r"\s+", " ", messy)
print("Clean spaces:", clean)
````

### Using Groups in Replacement

````python
# Reformat dates: DD/MM/YYYY → YYYY-MM-DD
dates = "15/03/2024 and 20/12/2023"

# Capture day, month, year and rearrange
reformatted = re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\3-\2-\1", dates)
print("Reformatted dates:", reformatted)

# Hide email addresses
emails = "Contact john@gmail.com or jane@company.org"
hidden = re.sub(r"(\w+)@([\w.]+)", r"***@\2", emails)
print("Hidden emails:", hidden)
````

### `re.split()` - Split on Pattern

Split text on complex patterns (more powerful than `str.split()`).

````python
# Split on multiple delimiters
text = "apple,banana;cherry orange|grape"

# Regular split only works on one delimiter
print("str.split(','):", text.split(','))

# Regex split on multiple delimiters
parts = re.split(r"[,;|\s]+", text)
print("re.split():", parts)
````

````python
# Split on one or more whitespace characters
text = "Name:   John    Age:  30   City: NYC"
parts = re.split(r"\s+", text)
print("Split on whitespace:", parts)
````
