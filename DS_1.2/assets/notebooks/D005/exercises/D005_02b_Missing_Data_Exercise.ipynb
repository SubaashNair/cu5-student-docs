## Bonus Exercise: Validate Imputation Quality

Using the validation function from the lesson, compare the distribution of `num_kids` before and after KNN imputation.

Create side-by-side:
1. Histogram comparison
2. Box plot comparison
3. Summary statistics table

```python
# Validation function
def validate_imputation(original, imputed, column):
    """Compare distributions before and after imputation."""
    
    print(f"=== Validation for '{column}' ===")
    
    # Statistics comparison
    stats = pd.DataFrame({
        'Metric': ['Mean', 'Median', 'Std Dev', 'Min', 'Max'],
        'Original': [
            original[column].mean(),
            original[column].median(),
            original[column].std(),
            original[column].min(),
            original[column].max()
        ],
        'After Imputation': [
            imputed[column].mean(),
            imputed[column].median(),
            imputed[column].std(),
            imputed[column].min(),
            imputed[column].max()
        ]
    })
    print(stats.to_string(index=False))
    
    return stats

# Your code to apply and validate KNN imputation
```
