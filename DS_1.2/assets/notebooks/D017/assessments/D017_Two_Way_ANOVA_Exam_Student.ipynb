**H1 (Main effect — Machine_Type):**  
- H₀: There is no significant difference in mean Defect_Rate between Automatic and Manual machines.  
- H₁: There is a significant difference in mean Defect_Rate between Automatic and Manual machines.

**H2 (Main effect — Shift):**  
- H₀: There is no significant difference in mean Defect_Rate across Morning, Afternoon, and Night shifts.  
- H₁: At least one shift has a significantly different mean Defect_Rate.

**H3 (Interaction — Machine_Type × Shift):**  
- H₀: The effect of Machine_Type on Defect_Rate does not depend on Shift.  
- H₁: The effect of Machine_Type on Defect_Rate depends on Shift (interaction exists).
