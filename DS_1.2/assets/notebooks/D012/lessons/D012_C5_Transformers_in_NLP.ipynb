While methods like Bag-of-Words and spaCy are powerful, the modern state-of-the-art in NLP is driven by **Deep Learning**, specifically the **Transformer** architecture (introduced by Google in 2017).

## 1. The Concept: From RNNs to Transformers

Before Transformers, we used **Recurrent Neural Networks (RNNs)**. RNNs processed words one by one, sequentially. This made them slow and bad at remembering long sentences.

**Transformers** process the *entire sentence at once* (Parallelism). They use a mechanism called **Self-Attention** to understand context.

### Diagram: The Self-Attention Mechanism

Self-Attention allows a word to "look at" other words in the sentence to figure out what it means.

For example, in "**The animal didn't cross the street because it was too tired**":
*   What does "it" refer to?
*   Self-Attention helps "it" look strongly at "animal" and weakly at "street".

```python
from transformers import pipeline
import logging

# Suppress verbose warnings for cleaner output
logging.getLogger("transformers").setLevel(logging.ERROR)
```

## 3. Task 1: Sentiment Analysis

We will compare two models to see why picking the right model matters.

**Model A**: `distilbert-base-uncased-finetuned-sst-2-english` (Default). Good for formal English.
**Model B**: `cardiffnlp/twitter-roberta-base-sentiment`. Good for social media (slang, emojis).

```python
# Load pipelines
classifier_default = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
classifier_twitter = pipeline("sentiment-analysis", model="cardiffnlp/twitter-roberta-base-sentiment", tokenizer="cardiffnlp/twitter-roberta-base-sentiment")

# Test sentence with slang and emoji
text = "The food was meh... :/ but the service was ok"

print(f"Text: '{text}'\n")

result_a = classifier_default(text)
print(f"DistilBERT says: {result_a}")

result_b = classifier_twitter(text)
print(f"RoBERTa (Twitter) says: {result_b}")
```

*Note: Evaluating `LABEL_0`, `LABEL_1` etc. requires checking the model documentation. For Twitter-RoBERTa, 0=Negative, 1=Neutral, 2=Positive.*

## 4. Task 2: Named Entity Recognition (NER)

We used spaCy for this before. Let's look at how **BERT** does it.

Model: `dbmdz/bert-large-cased-finetuned-conll03-english`

```python
ner_bert = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")

text = "Apple was founded by Steve Jobs in California."

entities = ner_bert(text)

print(f"Text: '{text}'")
for entity in entities:
    print(f" - {entity['word']} ({entity['entity_group']}) score: {entity['score']:.4f}")
```

## 5. Task 3: Text Generation

Transformers can write text! This is called **Causal Language Modeling**.

Model: `gpt2` (The ancestor of ChatGPT). It predicts the next word in a sequence.

```python
generator = pipeline("text-generation", model="gpt2")

prompt = "Artificial Intelligence is going to"

# Generate text
output = generator(prompt, max_length=50, num_return_sequences=1, truncation=True)

print(f"Prompt: '{prompt}'...\n")
print(output[0]['generated_text'])
```

## 6. Task 4: Zero-Shot Classification

This is one of the most powerful features of Transformers. It allows you to classify text into **categories that the model has never seen before**.

You just give it the text and a list of `candidate_labels`, and it figures it out!

Model: `facebook/bart-large-mnli` (trained on Multi-Genre Natural Language Inference).

```python
classifier_zero = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

text = "My internet is down and I cannot work."
labels = ["urgent", "not urgent", "tablet", "computer", "internet"]

result = classifier_zero(text, candidate_labels=labels)

print(f"Text: '{text}'\n")
print("Scores:")
for label, score in zip(result['labels'], result['scores']):
    print(f" - {label}: {score:.4f}")
```
