# C5: Deep Learning in NLP (Transformers)

While methods like Bag-of-Words and spaCy are powerful, the modern state-of-the-art in NLP is driven by **Deep Learning**, specifically the **Transformer** architecture (introduced by Google in 2017).

## 1. The Concept: From RNNs to Transformers

Before Transformers, we used **Recurrent Neural Networks (RNNs)**. RNNs processed words one by one, sequentially. This made them slow and bad at remembering long sentences.

**Transformers** process the *entire sentence at once* (Parallelism). They use a mechanism called **Self-Attention** to understand context.

### Diagram: The Self-Attention Mechanism

Self-Attention allows a word to "look at" other words in the sentence to figure out what it means.

For example, in "**The animal didn't cross the street because it was too tired**":
*   What does "it" refer to?
*   Self-Attention helps "it" look strongly at "animal" and weakly at "street".
