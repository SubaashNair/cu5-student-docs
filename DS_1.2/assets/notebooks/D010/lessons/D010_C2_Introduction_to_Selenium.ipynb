### Setting Up Chrome WebDriver

Let's create a function to configure and initialize our Chrome WebDriver with optimal settings.

```python
def setup_driver(headless=False):
    """
    Configure and initialize Chrome WebDriver
    
    Args:
        headless (bool): Run browser in headless mode (no GUI)
    
    Returns:
        webdriver.Chrome: Configured Chrome WebDriver instance
    """
    chrome_options = Options()
    
    # Browser configuration options
    if headless:
        chrome_options.add_argument("--headless=new")  # Run without GUI
    
    chrome_options.add_argument("--disable-gpu")        # Disable GPU acceleration
    chrome_options.add_argument("--no-sandbox")         # Bypass OS security model
    chrome_options.add_argument("--window-size=1920,1080")  # Set window size
    chrome_options.add_argument("--disable-dev-shm-usage")  # Overcome limited resource problems
    
    # Automatically download and setup ChromeDriver
    service = Service(ChromeDriverManager().install())
    
    # Create driver instance
    driver = webdriver.Chrome(service=service, options=chrome_options)
    driver.set_page_load_timeout(30)  # Set page load timeout
    
    return driver

# Test the setup
print("WebDriver setup function created successfully!")
print("Options explained:")
print("- headless: Run browser without GUI (faster, less resource-intensive)")
print("- disable-gpu: Disable GPU acceleration for stability")
print("- no-sandbox: Bypass OS security (needed in some environments)")
print("- window-size: Set browser window dimensions")
```

## 3. Finding Elements - The Foundation of Web Scraping

Finding elements is the core skill in Selenium. You need to locate elements before you can interact with them or extract data.

### Common Element Locators

Selenium provides several ways to find elements:

```python
# Let's start with a simple example
driver = setup_driver(headless=None)  # Set to True for headless mode

try:
    # Navigate to a test page
    driver.get("https://sandbox.oxylabs.io/products")
    print("Successfully navigated to Oxylabs sandbox!")
    
    # Wait a moment for the page to load
    time.sleep(2)
    
    print(f"Page title: {driver.title}")
    print(f"Current URL: {driver.current_url}")
    
except Exception as e:
    print(f"Error: {e}")
finally:
    # Don't close the driver yet - we'll use it in the next examples
    pass
```

### 3.1 Finding Elements by Different Selectors

Let's explore different ways to find elements using the Oxylabs sandbox page:

```python
# Method 1: By CSS Selector (Most flexible and commonly used)
print("=== Finding Elements by CSS Selector ===")

try:
    # Find product cards using CSS selector
    product_cards = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
    print(f"Found {len(product_cards)} product cards using CSS selector '.css-e8at8d'")
    
    # Find product titles
    titles = driver.find_elements(By.CSS_SELECTOR, "h4.css-7u5e79")
    print(f"Found {len(titles)} product titles using CSS selector 'h4.css-7u5e79'")
    
    # Display first few titles
    print("\nFirst 3 product titles:")
    for i, title in enumerate(titles[:3]):
        print(f"{i+1}. {title.text}")
        
except Exception as e:
    print(f"Error finding elements: {e}")
```

```python
# Method 2: By Tag Name
print("\n=== Finding Elements by Tag Name ===")

try:
    # Find all images on the page
    images = driver.find_elements(By.TAG_NAME, "img")
    print(f"Found {len(images)} images on the page")
    
    # Find all links
    links = driver.find_elements(By.TAG_NAME, "a")
    print(f"Found {len(links)} links on the page")
    
    # Display first few image sources
    print("\nFirst 3 image sources:")
    for i, img in enumerate(images[:3]):
        src = img.get_attribute("src")
        alt = img.get_attribute("alt")
        print(f"{i+1}. {src[:50]}... (alt: {alt})")
        
except Exception as e:
    print(f"Error finding elements: {e}")
```

```python
# Method 3: By Class Name
print("\n=== Finding Elements by Class Name ===")

try:
    # Note: Class names in this site use CSS-in-JS, so they might be dynamic
    # Let's find elements with specific classes
    
    # Find elements by partial class name using CSS selector
    price_elements = driver.find_elements(By.CSS_SELECTOR, "[class*='css-li4v8k']")
    print(f"Found {len(price_elements)} price elements")
    
    # Display first few prices
    print("\nFirst 3 prices:")
    for i, price in enumerate(price_elements[:3]):
        print(f"{i+1}. {price.text}")
        
except Exception as e:
    print(f"Error finding elements: {e}")
```

```python
# Method 4: By XPath (Very powerful but can be complex)
print("\n=== Finding Elements by XPath ===")

try:
    # Find product titles using XPath
    xpath_titles = driver.find_elements(By.XPATH, "//h4[contains(@class, 'css-7u5e79')]")
    print(f"Found {len(xpath_titles)} titles using XPath")
    
    # Find buttons using XPath
    buttons = driver.find_elements(By.XPATH, "//button[contains(@class, 'css-13df51w')]")
    print(f"Found {len(buttons)} buttons using XPath")
    
    # Display button texts
    print("\nFirst 3 button texts:")
    for i, button in enumerate(buttons[:3]):
        print(f"{i+1}. {button.text}")
        
except Exception as e:
    print(f"Error finding elements: {e}")
```

### 3.2 Single vs Multiple Elements

Understanding the difference between `find_element()` and `find_elements()` is crucial:

```python
print("=== Single vs Multiple Elements ===")

try:
    # find_element() - Returns the FIRST matching element
    first_title = driver.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
    print(f"First title (single): {first_title.text}")
    print(f"Type: {type(first_title)}")
    
    # find_elements() - Returns a LIST of ALL matching elements
    all_titles = driver.find_elements(By.CSS_SELECTOR, "h4.css-7u5e79")
    print(f"\nAll titles (multiple): Found {len(all_titles)} titles")
    print(f"Type: {type(all_titles)}")
    
    # Important difference in behavior when no elements found:
    try:
        # This will raise NoSuchElementException if not found
        non_existent = driver.find_element(By.CSS_SELECTOR, ".non-existent-class")
    except Exception as e:
        print(f"\nfind_element() with non-existent selector: {type(e).__name__}")
    
    # This will return an empty list if not found
    non_existent_list = driver.find_elements(By.CSS_SELECTOR, ".non-existent-class")
    print(f"find_elements() with non-existent selector: {non_existent_list} (length: {len(non_existent_list)})")
    
except Exception as e:
    print(f"Error: {e}")
```

### 3.3 Extracting Information from Elements

Once you find elements, you need to extract information from them:

```python
print("=== Extracting Information from Elements ===")

try:
    # Get the first product card for detailed extraction
    first_card = driver.find_element(By.CSS_SELECTOR, ".css-e8at8d")
    
    # Extract text content
    title_element = first_card.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
    title_text = title_element.text
    print(f"Title text: {title_text}")
    
    # Extract attributes
    link_element = first_card.find_element(By.CSS_SELECTOR, "a.css-o171kl")
    href = link_element.get_attribute("href")
    print(f"Product link: {href}")
    
    # Extract image source
    img_element = first_card.find_element(By.CSS_SELECTOR, "img")
    img_src = img_element.get_attribute("src")
    img_alt = img_element.get_attribute("alt")
    print(f"Image source: {img_src}")
    print(f"Image alt text: {img_alt}")
    
    # Extract price
    try:
        price_element = first_card.find_element(By.CSS_SELECTOR, ".css-li4v8k")
        price = price_element.text
        print(f"Price: {price}")
    except:
        print("Price not found for this product")
    
    # Extract other attributes
    print(f"\nElement tag name: {title_element.tag_name}")
    print(f"Element is displayed: {title_element.is_displayed()}")
    print(f"Element is enabled: {title_element.is_enabled()}")
    
except Exception as e:
    print(f"Error extracting information: {e}")
```

## 4. Waiting Strategies - Handling Dynamic Content

Modern websites load content dynamically with JavaScript. Simply using `time.sleep()` is unreliable and slow. Selenium provides better waiting strategies.

### Types of Waits:
1. **Implicit Wait**: Global timeout for all element finding operations
2. **Explicit Wait**: Wait for specific conditions to be met
3. **Fluent Wait**: More advanced explicit wait with custom polling

### 4.1 Implicit Waits

```python
print("=== Implicit Waits ===")

# Set implicit wait - applies to all find_element operations
driver.implicitly_wait(10)  # Wait up to 10 seconds for elements to appear

print("Implicit wait set to 10 seconds")
print("Now all find_element operations will wait up to 10 seconds before failing")

# Test implicit wait
start_time = time.time()
try:
    # This will wait up to 10 seconds for the element to appear
    element = driver.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
    end_time = time.time()
    print(f"Element found in {end_time - start_time:.2f} seconds")
except Exception as e:
    end_time = time.time()
    print(f"Element not found after {end_time - start_time:.2f} seconds: {e}")
```

### 4.2 Explicit Waits (Recommended)

Explicit waits are more flexible and reliable. They wait for specific conditions.

```python
print("=== Explicit Waits ===")

# Create a WebDriverWait object
wait = WebDriverWait(driver, 10)  # Wait up to 10 seconds

try:
    # Wait for element to be present
    print("Waiting for product grid to be present...")
    product_grid = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty"))
    )
    print("Product grid found!")
    
    # Wait for element to be visible
    print("\nWaiting for first product title to be visible...")
    first_title = wait.until(
        EC.visibility_of_element_located((By.CSS_SELECTOR, "h4.css-7u5e79"))
    )
    print(f"First title visible: {first_title.text}")
    
    # Wait for multiple elements to be present
    print("\nWaiting for all product cards to be present...")
    wait.until(
        EC.presence_of_all_elements_located((By.CSS_SELECTOR, ".css-e8at8d"))
    )
    product_cards = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
    print(f"Found {len(product_cards)} product cards")
    
except Exception as e:
    print(f"Wait condition not met: {e}")
```

### 4.3 Common Wait Conditions

Here are the most useful wait conditions:

```python
print("=== Common Wait Conditions ===")

# Create wait object
wait = WebDriverWait(driver, 10)

# 1. presence_of_element_located - Element is in DOM (may not be visible)
print("1. Testing presence_of_element_located...")
try:
    element = wait.until(EC.presence_of_element_located((By.TAG_NAME, "body")))
    print("   ✓ Body element is present in DOM")
except:
    print("   ✗ Body element not found")

# 2. visibility_of_element_located - Element is visible (has height and width > 0)
print("\n2. Testing visibility_of_element_located...")
try:
    element = wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, "h4.css-7u5e79")))
    print("   ✓ First product title is visible")
except:
    print("   ✗ First product title not visible")

# 3. element_to_be_clickable - Element is visible and enabled
print("\n3. Testing element_to_be_clickable...")
try:
    button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "button.css-13df51w")))
    print("   ✓ First button is clickable")
except:
    print("   ✗ No clickable button found")

# 4. text_to_be_present_in_element - Specific text is present in element
print("\n4. Testing text_to_be_present_in_element...")
try:
    # This might not work if we don't know the exact text
    wait_short = WebDriverWait(driver, 2)  # Shorter wait for this test
    wait_short.until(EC.text_to_be_present_in_element((By.TAG_NAME, "title"), "Oxylabs"))
    print("   ✓ 'Oxylabs' text found in title")
except:
    print("   ✗ 'Oxylabs' text not found in title (this is expected)")

print("\nWait conditions demonstration complete!")
```

### 4.4 Custom Wait Conditions

You can create custom wait conditions for specific scenarios:

```python
print("=== Custom Wait Conditions ===")

# Custom condition: Wait for at least N products to be loaded
def at_least_n_products_loaded(n):
    """Custom condition to wait for at least n products to be loaded"""
    def _predicate(driver):
        products = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
        return len(products) >= n
    return _predicate

# Custom condition: Wait for page to be fully loaded (no loading indicators)
def page_fully_loaded():
    """Custom condition to wait for page to be fully loaded"""
    def _predicate(driver):
        # Check if page is ready and no loading indicators are present
        ready_state = driver.execute_script("return document.readyState")
        loading_indicators = driver.find_elements(By.CSS_SELECTOR, ".loading, .spinner, [class*='loading']")
        return ready_state == "complete" and len(loading_indicators) == 0
    return _predicate

try:
    # Use custom condition to wait for at least 5 products
    print("Waiting for at least 5 products to load...")
    wait.until(at_least_n_products_loaded(5))
    products = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
    print(f"✓ Found {len(products)} products (at least 5 as required)")
    
    # Use custom condition to wait for page to be fully loaded
    print("\nWaiting for page to be fully loaded...")
    wait.until(page_fully_loaded())
    print("✓ Page is fully loaded")
    
except Exception as e:
    print(f"Custom wait condition failed: {e}")

print("\nCustom wait conditions demonstration complete!")
```

## 5. Guided Practical Example: Scraping Oxylabs E-commerce Sandbox

Now let's put everything together in a comprehensive example. We'll scrape product information from the Oxylabs e-commerce sandbox, which is designed for testing web scraping techniques.

### 5.1 Complete Scraping Function

Let's create a complete function that demonstrates all the concepts we've learned:

```python
def scrape_oxylabs_products(max_products=20):
    """
    Comprehensive function to scrape product data from Oxylabs sandbox
    
    Args:
        max_products (int): Maximum number of products to scrape
    
    Returns:
        list: List of dictionaries containing product information
    """
    print(f"Starting to scrape up to {max_products} products from Oxylabs sandbox...")
    
    # Setup driver with optimized options
    driver = setup_driver()
    
    # Create wait object for explicit waits
    wait = WebDriverWait(driver, 15)
    
    products_data = []
    
    try:
        # Navigate to the sandbox
        print("Navigating to Oxylabs e-commerce sandbox...")
        driver.get("https://sandbox.oxylabs.io/products")
        
        # Wait for the page to load completely
        print("Waiting for page to load...")
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty")))
        
        # Wait for products to be visible
        print("Waiting for products to be visible...")
        wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, ".css-e8at8d")))
        
        # Find all product cards
        print("Finding all product cards...")
        product_cards = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
        print(f"Found {len(product_cards)} product cards")
        
        # Limit the number of products to scrape
        products_to_scrape = min(len(product_cards), max_products)
        print(f"Will scrape {products_to_scrape} products")
        
        # Extract data from each product card
        for i, card in enumerate(product_cards[:products_to_scrape]):
            try:
                print(f"\nScraping product {i+1}/{products_to_scrape}...")
                
                # Scroll the card into view
                driver.execute_script("arguments[0].scrollIntoView(true);", card)
                time.sleep(0.5)  # Small delay for smooth scrolling
                
                # Extract product information
                product_data = {}
                
                # Product title
                try:
                    title_element = card.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
                    product_data['title'] = title_element.text.strip()
                    print(f"  Title: {product_data['title']}")
                except Exception as e:
                    product_data['title'] = "N/A"
                    print(f"  Title: Could not extract - {e}")
                
                # Product price
                try:
                    price_element = card.find_element(By.CSS_SELECTOR, ".css-1xqgvqy")
                    product_data['price'] = price_element.text.strip()
                    print(f"  Price: {product_data['price']}")
                except Exception as e:
                    product_data['price'] = "N/A"
                    print(f"  Price: Could not extract - {e}")
                
                # Product image URL
                try:
                    img_element = card.find_element(By.CSS_SELECTOR, "img")
                    product_data['image_url'] = img_element.get_attribute('src')
                    print(f"  Image URL: {product_data['image_url'][:50]}...")
                except Exception as e:
                    product_data['image_url'] = "N/A"
                    print(f"  Image URL: Could not extract - {e}")
                
                # Product rating (if available)
                try:
                    rating_element = card.find_element(By.CSS_SELECTOR, "[data-testid='rating']")
                    product_data['rating'] = rating_element.get_attribute('aria-label')
                    print(f"  Rating: {product_data['rating']}")
                except Exception as e:
                    product_data['rating'] = "N/A"
                    print(f"  Rating: Could not extract - {e}")
                
                # Add to cart button availability
                try:
                    button = card.find_element(By.CSS_SELECTOR, "button.css-13df51w")
                    product_data['add_to_cart_available'] = button.is_enabled()
                    product_data['button_text'] = button.text.strip()
                    print(f"  Add to cart: {product_data['add_to_cart_available']} - '{product_data['button_text']}'")
                except Exception as e:
                    product_data['add_to_cart_available'] = False
                    product_data['button_text'] = "N/A"
                    print(f"  Add to cart: Could not extract - {e}")
                
                # Add timestamp
                product_data['scraped_at'] = time.strftime('%Y-%m-%d %H:%M:%S')
                
                # Add to results
                products_data.append(product_data)
                
            except Exception as e:
                print(f"  Error scraping product {i+1}: {e}")
                continue
        
        print(f"\nSuccessfully scraped {len(products_data)} products!")
        
    except Exception as e:
        print(f"Error during scraping: {e}")
    
    finally:
        # Always close the driver
        print("Closing browser...")
        driver.quit()
    
    return products_data

# Run the scraping function
print("=== Starting Oxylabs Product Scraping ===")
scraped_products = scrape_oxylabs_products(max_products=10)

# Display results summary
print(f"\n=== Scraping Results Summary ===")
print(f"Total products scraped: {len(scraped_products)}")

if scraped_products:
    print("\nFirst product details:")
    for key, value in scraped_products[0].items():
        print(f"  {key}: {value}")
    
    print("\nAll product titles:")
    for i, product in enumerate(scraped_products, 1):
        print(f"  {i}. {product['title']} - {product['price']}")
```

### 5.2 Advanced Interaction Example

Let's also demonstrate more advanced interactions like clicking buttons and handling dynamic content:

```python
def demonstrate_advanced_interactions():
    """
    Demonstrate advanced Selenium interactions like clicking, form filling, etc.
    """
    print("=== Advanced Selenium Interactions Demo ===")
    
    driver = setup_driver()
    wait = WebDriverWait(driver, 15)
    
    try:
        # Navigate to the sandbox
        print("Navigating to Oxylabs sandbox...")
        driver.get("https://sandbox.oxylabs.io/products")
        
        # Wait for page to load
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty")))
        
        # 1. Demonstrate clicking on a product
        print("\n1. Clicking on first product...")
        try:
            first_product = wait.until(
                EC.element_to_be_clickable((By.CSS_SELECTOR, ".css-e8at8d:first-child"))
            )
            
            # Get product title before clicking
            title_element = first_product.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
            product_title = title_element.text
            print(f"   Clicking on: {product_title}")
            
            # Scroll into view and click
            driver.execute_script("arguments[0].scrollIntoView(true);", first_product)
            time.sleep(1)
            first_product.click()
            
            # Wait for navigation or modal to appear
            time.sleep(2)
            print(f"   Current URL after click: {driver.current_url}")
            
            # Go back to products page
            driver.back()
            wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty")))
            
        except Exception as e:
            print(f"   Error clicking product: {e}")
        
        # 2. Demonstrate Add to Cart interaction
        print("\n2. Interacting with Add to Cart buttons...")
        try:
            # Find all add to cart buttons
            cart_buttons = driver.find_elements(By.CSS_SELECTOR, "button.css-13df51w")
            print(f"   Found {len(cart_buttons)} add to cart buttons")
            
            if cart_buttons:
                # Click first available button
                first_button = cart_buttons[0]
                
                # Scroll into view
                driver.execute_script("arguments[0].scrollIntoView(true);", first_button)
                time.sleep(1)
                
                # Check if button is enabled
                if first_button.is_enabled():
                    button_text = first_button.text
                    print(f"   Clicking button: '{button_text}'")
                    first_button.click()
                    time.sleep(2)
                    print("   Button clicked successfully")
                else:
                    print("   First button is disabled")
            
        except Exception as e:
            print(f"   Error with cart buttons: {e}")
        
        # 3. Demonstrate scrolling and lazy loading
        print("\n3. Demonstrating scrolling behavior...")
        try:
            # Get initial number of products
            initial_products = len(driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d"))
            print(f"   Initial products visible: {initial_products}")
            
            # Scroll to bottom of page
            print("   Scrolling to bottom of page...")
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(3)
            
            # Check if more products loaded
            final_products = len(driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d"))
            print(f"   Products after scrolling: {final_products}")
            
            if final_products > initial_products:
                print(f"   ✓ Lazy loading detected! {final_products - initial_products} more products loaded")
            else:
                print("   No additional products loaded (no lazy loading or all already loaded)")
            
        except Exception as e:
            print(f"   Error with scrolling: {e}")
        
        # 4. Demonstrate JavaScript execution
        print("\n4. Demonstrating JavaScript execution...")
        try:
            # Get page title using JavaScript
            page_title = driver.execute_script("return document.title;")
            print(f"   Page title via JS: {page_title}")
            
            # Get page dimensions
            dimensions = driver.execute_script(
                "return {width: window.innerWidth, height: window.innerHeight};"
            )
            print(f"   Browser dimensions: {dimensions['width']}x{dimensions['height']}")
            
            # Count elements using JavaScript
            product_count = driver.execute_script(
                "return document.querySelectorAll('.css-e8at8d').length;"
            )
            print(f"   Product count via JS: {product_count}")
            
        except Exception as e:
            print(f"   Error with JavaScript execution: {e}")
        
        print("\n✓ Advanced interactions demonstration complete!")
        
    except Exception as e:
        print(f"Error in advanced interactions demo: {e}")
    
    finally:
        print("Closing browser...")
        driver.quit()

# Run the advanced interactions demo
demonstrate_advanced_interactions()
```

## 6. Data Extraction and Storage with Pandas

After scraping data with Selenium, you'll often want to store it in a structured format for analysis. Pandas is perfect for this task.

### 6.1 Converting Scraped Data to DataFrame

```python
import pandas as pd
import json
from datetime import datetime
import os

def scrape_and_store_data(max_products=15, save_formats=['csv', 'json', 'excel']):
    """
    Scrape data and store it in multiple formats using pandas
    
    Args:
        max_products (int): Maximum number of products to scrape
        save_formats (list): List of formats to save data in
    
    Returns:
        pandas.DataFrame: DataFrame containing scraped data
    """
    print("=== Scraping Data for Storage Demo ===")
    
    # First, let's scrape some data
    driver = setup_driver()
    wait = WebDriverWait(driver, 15)
    products_data = []
    
    try:
        print("Navigating to Oxylabs sandbox...")
        driver.get("https://sandbox.oxylabs.io/products")
        
        # Wait for products to load
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty")))
        wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, ".css-e8at8d")))
        
        # Get product cards
        product_cards = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
        products_to_scrape = min(len(product_cards), max_products)
        
        print(f"Scraping {products_to_scrape} products for storage demo...")
        
        for i, card in enumerate(product_cards[:products_to_scrape]):
            try:
                # Scroll into view
                driver.execute_script("arguments[0].scrollIntoView(true);", card)
                time.sleep(0.3)
                
                product_data = {
                    'product_id': i + 1,
                    'scraped_timestamp': datetime.now().isoformat(),
                }
                
                # Extract title
                try:
                    title_element = card.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
                    product_data['title'] = title_element.text.strip()
                except:
                    product_data['title'] = None
                
                # Extract price
                try:
                    price_element = card.find_element(By.CSS_SELECTOR, ".css-1xqgvqy")
                    price_text = price_element.text.strip()
                    product_data['price_text'] = price_text
                    
                    # Extract numeric price
                    import re
                    price_match = re.search(r'\$([\d,]+\.?\d*)', price_text)
                    if price_match:
                        product_data['price_numeric'] = float(price_match.group(1).replace(',', ''))
                    else:
                        product_data['price_numeric'] = None
                except:
                    product_data['price_text'] = None
                    product_data['price_numeric'] = None
                
                # Extract image URL
                try:
                    img_element = card.find_element(By.CSS_SELECTOR, "img")
                    product_data['image_url'] = img_element.get_attribute('src')
                except:
                    product_data['image_url'] = None
                
                # Check if add to cart is available
                try:
                    button = card.find_element(By.CSS_SELECTOR, "button.css-13df51w")
                    product_data['add_to_cart_available'] = button.is_enabled()
                    product_data['button_text'] = button.text.strip()
                except:
                    product_data['add_to_cart_available'] = False
                    product_data['button_text'] = None
                
                products_data.append(product_data)
                
            except Exception as e:
                print(f"Error scraping product {i+1}: {e}")
                continue
        
    except Exception as e:
        print(f"Error during scraping: {e}")
    
    finally:
        driver.quit()
    
    # Convert to DataFrame
    print(f"\nConverting {len(products_data)} products to DataFrame...")
    df = pd.DataFrame(products_data)
    
    if not df.empty:
        # Display basic info about the DataFrame
        print(f"\nDataFrame Info:")
        print(f"Shape: {df.shape}")
        print(f"Columns: {list(df.columns)}")
        print(f"\nFirst few rows:")
        print(df.head())
        
        # Data cleaning and processing
        print("\n=== Data Processing ===")
        
        # Check for missing values
        print("Missing values per column:")
        print(df.isnull().sum())
        
        # Basic statistics for numeric columns
        if 'price_numeric' in df.columns and df['price_numeric'].notna().any():
            print("\nPrice statistics:")
            print(df['price_numeric'].describe())
        
        # Save in different formats
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        
        for format_type in save_formats:
            try:
                if format_type == 'csv':
                    filename = f'oxylabs_products_{timestamp}.csv'
                    df.to_csv(filename, index=False)
                    print(f"✓ Saved as CSV: {filename}")
                
                elif format_type == 'json':
                    filename = f'oxylabs_products_{timestamp}.json'
                    df.to_json(filename, orient='records', indent=2)
                    print(f"✓ Saved as JSON: {filename}")
                
                elif format_type == 'excel':
                    filename = f'oxylabs_products_{timestamp}.xlsx'
                    df.to_excel(filename, index=False, sheet_name='Products')
                    print(f"✓ Saved as Excel: {filename}")
                
            except Exception as e:
                print(f"✗ Error saving as {format_type}: {e}")
    
    return df

# Run the scraping and storage demo
scraped_df = scrape_and_store_data(max_products=10)
```

### 6.2 Advanced Data Analysis with Pandas

```python
def analyze_scraped_data(df):
    """
    Perform advanced analysis on scraped data
    
    Args:
        df (pandas.DataFrame): DataFrame containing scraped product data
    """
    if df.empty:
        print("No data to analyze")
        return
    
    print("=== Advanced Data Analysis ===")
    
    # 1. Basic data exploration
    print("\n1. Data Overview:")
    print(f"   Total products: {len(df)}")
    print(f"   Columns: {len(df.columns)}")
    print(f"   Memory usage: {df.memory_usage(deep=True).sum()} bytes")
    
    # 2. Price analysis
    if 'price_numeric' in df.columns:
        valid_prices = df['price_numeric'].dropna()
        if not valid_prices.empty:
            print("\n2. Price Analysis:")
            print(f"   Products with valid prices: {len(valid_prices)}")
            print(f"   Average price: ${valid_prices.mean():.2f}")
            print(f"   Median price: ${valid_prices.median():.2f}")
            print(f"   Price range: ${valid_prices.min():.2f} - ${valid_prices.max():.2f}")
            print(f"   Standard deviation: ${valid_prices.std():.2f}")
            
            # Price categories
            df['price_category'] = pd.cut(df['price_numeric'], 
                                        bins=[0, 50, 100, 200, float('inf')], 
                                        labels=['Budget', 'Mid-range', 'Premium', 'Luxury'])
            
            print("\n   Price categories:")
            print(df['price_category'].value_counts())
    
    # 3. Title analysis
    if 'title' in df.columns:
        valid_titles = df['title'].dropna()
        if not valid_titles.empty:
            print("\n3. Title Analysis:")
            print(f"   Products with titles: {len(valid_titles)}")
            print(f"   Average title length: {valid_titles.str.len().mean():.1f} characters")
            
            # Most common words in titles
            all_words = ' '.join(valid_titles).lower().split()
            word_freq = pd.Series(all_words).value_counts().head(10)
            print("\n   Most common words in titles:")
            for word, count in word_freq.items():
                print(f"     {word}: {count}")
    
    # 4. Availability analysis
    if 'add_to_cart_available' in df.columns:
        print("\n4. Availability Analysis:")
        availability_counts = df['add_to_cart_available'].value_counts()
        print(f"   Available products: {availability_counts.get(True, 0)}")
        print(f"   Unavailable products: {availability_counts.get(False, 0)}")
        
        if True in availability_counts.index and False in availability_counts.index:
            availability_rate = availability_counts[True] / len(df) * 100
            print(f"   Availability rate: {availability_rate:.1f}%")
    
    # 5. Data quality assessment
    print("\n5. Data Quality Assessment:")
    completeness = (df.notna().sum() / len(df) * 100).round(1)
    print("   Completeness by column:")
    for col, pct in completeness.items():
        print(f"     {col}: {pct}%")
    
    # 6. Create summary report
    print("\n6. Summary Report:")
    summary = {
        'total_products': len(df),
        'avg_price': df['price_numeric'].mean() if 'price_numeric' in df.columns else None,
        'availability_rate': (df['add_to_cart_available'].sum() / len(df) * 100) if 'add_to_cart_available' in df.columns else None,
        'data_completeness': df.notna().all(axis=1).sum() / len(df) * 100
    }
    
    for key, value in summary.items():
        if value is not None:
            if isinstance(value, float):
                print(f"   {key.replace('_', ' ').title()}: {value:.2f}")
            else:
                print(f"   {key.replace('_', ' ').title()}: {value}")
    
    return df

# Analyze the scraped data
if 'scraped_df' in locals() and not scraped_df.empty:
    analyzed_df = analyze_scraped_data(scraped_df)
else:
    print("No scraped data available for analysis. Run the scraping function first.")
```

### 6.3 Data Visualization (Optional)

If you have matplotlib installed, you can create visualizations of your scraped data:

```python
def visualize_data(df):
    """
    Create visualizations of scraped data (requires matplotlib)
    
    Args:
        df (pandas.DataFrame): DataFrame containing scraped data
    """
    try:
        import matplotlib.pyplot as plt
        
        if df.empty:
            print("No data to visualize")
            return
        
        print("=== Data Visualization ===")
        
        # Create subplots
        fig, axes = plt.subplots(2, 2, figsize=(12, 10))
        fig.suptitle('Oxylabs Product Data Analysis', fontsize=16)
        
        # 1. Price distribution
        if 'price_numeric' in df.columns and df['price_numeric'].notna().any():
            valid_prices = df['price_numeric'].dropna()
            axes[0, 0].hist(valid_prices, bins=10, edgecolor='black', alpha=0.7)
            axes[0, 0].set_title('Price Distribution')
            axes[0, 0].set_xlabel('Price ($)')
            axes[0, 0].set_ylabel('Frequency')
        
        # 2. Availability pie chart
        if 'add_to_cart_available' in df.columns:
            availability_counts = df['add_to_cart_available'].value_counts()
            axes[0, 1].pie(availability_counts.values, 
                          labels=['Available' if x else 'Unavailable' for x in availability_counts.index],
                          autopct='%1.1f%%')
            axes[0, 1].set_title('Product Availability')
        
        # 3. Title length distribution
        if 'title' in df.columns:
            title_lengths = df['title'].dropna().str.len()
            axes[1, 0].hist(title_lengths, bins=10, edgecolor='black', alpha=0.7)
            axes[1, 0].set_title('Title Length Distribution')
            axes[1, 0].set_xlabel('Title Length (characters)')
            axes[1, 0].set_ylabel('Frequency')
        
        # 4. Price vs Availability
        if 'price_numeric' in df.columns and 'add_to_cart_available' in df.columns:
            available_prices = df[df['add_to_cart_available'] == True]['price_numeric'].dropna()
            unavailable_prices = df[df['add_to_cart_available'] == False]['price_numeric'].dropna()
            
            if not available_prices.empty:
                axes[1, 1].boxplot([available_prices, unavailable_prices], 
                                  labels=['Available', 'Unavailable'])
                axes[1, 1].set_title('Price by Availability')
                axes[1, 1].set_ylabel('Price ($)')
        
        plt.tight_layout()
        plt.show()
        
        print("✓ Visualizations created successfully!")
        
    except ImportError:
        print("Matplotlib not installed. Install with: pip install matplotlib")
    except Exception as e:
        print(f"Error creating visualizations: {e}")

# Create visualizations (if matplotlib is available)
if 'scraped_df' in locals() and not scraped_df.empty:
    visualize_data(scraped_df)
else:
    print("No data available for visualization. Run the scraping function first.")
```
