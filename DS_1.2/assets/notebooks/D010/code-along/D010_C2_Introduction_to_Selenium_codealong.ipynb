## 1. What is Selenium?

Selenium is a powerful tool for automating web browsers. It allows you to:
- **Automate web interactions**: Click buttons, fill forms, navigate pages
- **Scrape dynamic content**: Handle JavaScript-rendered content
- **Test web applications**: Automate testing workflows
- **Simulate user behavior**: Mimic real user interactions

### Why Use Selenium?
- **Dynamic Content**: Unlike BeautifulSoup, Selenium can handle JavaScript-rendered content
- **Real Browser**: Uses actual browsers (Chrome, Firefox, etc.)
- **User Simulation**: Can perform complex user interactions
- **Cross-platform**: Works on Windows, macOS, and Linux

```python
# ═══ CHECKPOINT — "Single vs Multiple Elements" ═══════════════════════════════════════════════
#
# Run this ONLY if you are behind, or a cell above threw an error and the rest
# stopped working. It rebuilds everything this section needs, so you can rejoin
# here instead of being stuck for the rest of the class.
#
# Running it when you are NOT stuck is harmless. If it leaves you stuck anyway,
# that is a bug in the template, not in your code — say so.

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
from pathlib import Path
def setup_driver(headless=False):
    """
    Configure and initialize Chrome WebDriver
    
    Args:
        headless (bool): Run browser in headless mode (no GUI)
    
    Returns:
        webdriver.Chrome: Configured Chrome WebDriver instance
    """
    chrome_options = Options()
    if headless:
        chrome_options.add_argument('--headless=new')
    chrome_options.add_argument('--disable-gpu')
    chrome_options.add_argument('--no-sandbox')
    chrome_options.add_argument('--window-size=1920,1080')
    chrome_options.add_argument('--disable-dev-shm-usage')
    service = Service(ChromeDriverManager().install())
    driver = webdriver.Chrome(service=service, options=chrome_options)
    driver.set_page_load_timeout(30)
    return driver
driver = setup_driver(headless=None)
try:
    product_cards = driver.find_elements(By.CSS_SELECTOR, '.css-e8at8d')
    print(f"Found {len(product_cards)} product cards using CSS selector '.css-e8at8d'")
    titles = driver.find_elements(By.CSS_SELECTOR, 'h4.css-7u5e79')
    print(f"Found {len(titles)} product titles using CSS selector 'h4.css-7u5e79'")
    print('\nFirst 3 product titles:')
    for i, title in enumerate(titles[:3]):
        print(f'{i + 1}. {title.text}')
except Exception as e:
    print(f'Error finding elements: {e}')
try:
    images = driver.find_elements(By.TAG_NAME, 'img')
    print(f'Found {len(images)} images on the page')
    links = driver.find_elements(By.TAG_NAME, 'a')
    print(f'Found {len(links)} links on the page')
    print('\nFirst 3 image sources:')
    for i, img in enumerate(images[:3]):
        src = img.get_attribute('src')
        alt = img.get_attribute('alt')
        print(f'{i + 1}. {src[:50]}... (alt: {alt})')
except Exception as e:
    print(f'Error finding elements: {e}')
try:
    price_elements = driver.find_elements(By.CSS_SELECTOR, "[class*='css-li4v8k']")
    print(f'Found {len(price_elements)} price elements')
    print('\nFirst 3 prices:')
    for i, price in enumerate(price_elements[:3]):
        print(f'{i + 1}. {price.text}')
except Exception as e:
    print(f'Error finding elements: {e}')
try:
    xpath_titles = driver.find_elements(By.XPATH, "//h4[contains(@class, 'css-7u5e79')]")
    print(f'Found {len(xpath_titles)} titles using XPath')
    buttons = driver.find_elements(By.XPATH, "//button[contains(@class, 'css-13df51w')]")
    print(f'Found {len(buttons)} buttons using XPath')
    print('\nFirst 3 button texts:')
    for i, button in enumerate(buttons[:3]):
        print(f'{i + 1}. {button.text}')
except Exception as e:
    print(f'Error finding elements: {e}')

print("Checkpoint restored. Ready for: Single vs Multiple Elements")
```

### 3.2 Single vs Multiple Elements

Understanding the difference between `find_element()` and `find_elements()` is crucial:

```python
print("=== Single vs Multiple Elements ===")

try:
    # find_element() - Returns the FIRST matching element
    first_title = driver.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
    print(f"First title (single): {first_title.text}")
    print(f"Type: {type(first_title)}")
    
    # find_elements() - Returns a LIST of ALL matching elements
    all_titles = driver.find_elements(By.CSS_SELECTOR, "h4.css-7u5e79")
    print(f"\nAll titles (multiple): Found {len(all_titles)} titles")
    print(f"Type: {type(all_titles)}")
    
    # Important difference in behavior when no elements found:
    try:
        # This will raise NoSuchElementException if not found
        non_existent = driver.find_element(By.CSS_SELECTOR, ".non-existent-class")
    except Exception as e:
        print(f"\nfind_element() with non-existent selector: {type(e).__name__}")
    
    # This will return an empty list if not found
    non_existent_list = driver.find_elements(By.CSS_SELECTOR, ".non-existent-class")
    print(f"find_elements() with non-existent selector: {non_existent_list} (length: {len(non_existent_list)})")
    
except Exception as e:
    print(f"Error: {e}")
```

### 3.3 Extracting Information from Elements

Once you find elements, you need to extract information from them:

```python
print("=== Extracting Information from Elements ===")

try:
    # Get the first product card for detailed extraction
    first_card = driver.find_element(By.CSS_SELECTOR, ".css-e8at8d")
    
    # Extract text content
    title_element = first_card.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
    title_text = title_element.text
    print(f"Title text: {title_text}")
    
    # Extract attributes
    link_element = first_card.find_element(By.CSS_SELECTOR, "a.css-o171kl")
    href = link_element.get_attribute("href")
    print(f"Product link: {href}")
    
    # Extract image source
    img_element = first_card.find_element(By.CSS_SELECTOR, "img")
    img_src = img_element.get_attribute("src")
    img_alt = img_element.get_attribute("alt")
    print(f"Image source: {img_src}")
    print(f"Image alt text: {img_alt}")
    
    # Extract price
    try:
        price_element = first_card.find_element(By.CSS_SELECTOR, ".css-li4v8k")
        price = price_element.text
        print(f"Price: {price}")
    except:
        print("Price not found for this product")
    
    # Extract other attributes
    print(f"\nElement tag name: {title_element.tag_name}")
    print(f"Element is displayed: {title_element.is_displayed()}")
    print(f"Element is enabled: {title_element.is_enabled()}")
    
except Exception as e:
    print(f"Error extracting information: {e}")
```

## 4. Waiting Strategies - Handling Dynamic Content

Modern websites load content dynamically with JavaScript. Simply using `time.sleep()` is unreliable and slow. Selenium provides better waiting strategies.

### Types of Waits:
1. **Implicit Wait**: Global timeout for all element finding operations
2. **Explicit Wait**: Wait for specific conditions to be met
3. **Fluent Wait**: More advanced explicit wait with custom polling

### 4.1 Implicit Waits

```python
print("=== Implicit Waits ===")

# Set implicit wait - applies to all find_element operations
driver.implicitly_wait(10)  # Wait up to 10 seconds for elements to appear

print("Implicit wait set to 10 seconds")
print("Now all find_element operations will wait up to 10 seconds before failing")

# Test implicit wait
start_time = time.time()
try:
    # This will wait up to 10 seconds for the element to appear
    element = driver.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
    end_time = time.time()
    print(f"Element found in {end_time - start_time:.2f} seconds")
except Exception as e:
    end_time = time.time()
    print(f"Element not found after {end_time - start_time:.2f} seconds: {e}")
```

### 4.2 Explicit Waits (Recommended)

Explicit waits are more flexible and reliable. They wait for specific conditions.

```python
print("=== Explicit Waits ===")

# Create a WebDriverWait object
wait = WebDriverWait(driver, 10)  # Wait up to 10 seconds

try:
    # Wait for element to be present
    print("Waiting for product grid to be present...")
    product_grid = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty"))
    )
    print("Product grid found!")
    
    # Wait for element to be visible
    print("\nWaiting for first product title to be visible...")
    first_title = wait.until(
        EC.visibility_of_element_located((By.CSS_SELECTOR, "h4.css-7u5e79"))
    )
    print(f"First title visible: {first_title.text}")
    
    # Wait for multiple elements to be present
    print("\nWaiting for all product cards to be present...")
    wait.until(
        EC.presence_of_all_elements_located((By.CSS_SELECTOR, ".css-e8at8d"))
    )
    product_cards = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
    print(f"Found {len(product_cards)} product cards")
    
except Exception as e:
    print(f"Wait condition not met: {e}")
```

### 4.3 Common Wait Conditions

Here are the most useful wait conditions:

```python
print("=== Common Wait Conditions ===")

# Create wait object
wait = WebDriverWait(driver, 10)

# 1. presence_of_element_located - Element is in DOM (may not be visible)
print("1. Testing presence_of_element_located...")
try:
    element = wait.until(EC.presence_of_element_located((By.TAG_NAME, "body")))
    print("   ✓ Body element is present in DOM")
except:
    print("   ✗ Body element not found")

# 2. visibility_of_element_located - Element is visible (has height and width > 0)
print("\n2. Testing visibility_of_element_located...")
try:
    element = wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, "h4.css-7u5e79")))
    print("   ✓ First product title is visible")
except:
    print("   ✗ First product title not visible")

# 3. element_to_be_clickable - Element is visible and enabled
print("\n3. Testing element_to_be_clickable...")
try:
    button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "button.css-13df51w")))
    print("   ✓ First button is clickable")
except:
    print("   ✗ No clickable button found")

# 4. text_to_be_present_in_element - Specific text is present in element
print("\n4. Testing text_to_be_present_in_element...")
try:
    # This might not work if we don't know the exact text
    wait_short = WebDriverWait(driver, 2)  # Shorter wait for this test
    wait_short.until(EC.text_to_be_present_in_element((By.TAG_NAME, "title"), "Oxylabs"))
    print("   ✓ 'Oxylabs' text found in title")
except:
    print("   ✗ 'Oxylabs' text not found in title (this is expected)")

print("\nWait conditions demonstration complete!")
```

## 6. Data Extraction and Storage with Pandas

After scraping data with Selenium, you'll often want to store it in a structured format for analysis. Pandas is perfect for this task.

### 6.1 Converting Scraped Data to DataFrame

```python
import pandas as pd
import json
from datetime import datetime
import os

def scrape_and_store_data(max_products=15, save_formats=['csv', 'json', 'excel']):
    """
    Scrape data and store it in multiple formats using pandas
    
    Args:
        max_products (int): Maximum number of products to scrape
        save_formats (list): List of formats to save data in
    
    Returns:
        pandas.DataFrame: DataFrame containing scraped data
    """
    print("=== Scraping Data for Storage Demo ===")
    
    # First, let's scrape some data
    driver = setup_driver()
    wait = WebDriverWait(driver, 15)
    products_data = []
    
    try:
        print("Navigating to Oxylabs sandbox...")
        driver.get("https://sandbox.oxylabs.io/products")
        
        # Wait for products to load
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, ".css-phdzty")))
        wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, ".css-e8at8d")))
        
        # Get product cards
        product_cards = driver.find_elements(By.CSS_SELECTOR, ".css-e8at8d")
        products_to_scrape = min(len(product_cards), max_products)
        
        print(f"Scraping {products_to_scrape} products for storage demo...")
        
        for i, card in enumerate(product_cards[:products_to_scrape]):
            try:
                # Scroll into view
                driver.execute_script("arguments[0].scrollIntoView(true);", card)
                time.sleep(0.3)
                
                product_data = {
                    'product_id': i + 1,
                    'scraped_timestamp': datetime.now().isoformat(),
                }
                
                # Extract title
                try:
                    title_element = card.find_element(By.CSS_SELECTOR, "h4.css-7u5e79")
                    product_data['title'] = title_element.text.strip()
                except:
                    product_data['title'] = None
                
                # Extract price
                try:
                    price_element = card.find_element(By.CSS_SELECTOR, ".css-1xqgvqy")
                    price_text = price_element.text.strip()
                    product_data['price_text'] = price_text
                    
                    # Extract numeric price
                    import re
                    price_match = re.search(r'\$([\d,]+\.?\d*)', price_text)
                    if price_match:
                        product_data['price_numeric'] = float(price_match.group(1).replace(',', ''))
                    else:
                        product_data['price_numeric'] = None
                except:
                    product_data['price_text'] = None
                    product_data['price_numeric'] = None
                
                # Extract image URL
                try:
                    img_element = card.find_element(By.CSS_SELECTOR, "img")
                    product_data['image_url'] = img_element.get_attribute('src')
                except:
                    product_data['image_url'] = None
                
                # Check if add to cart is available
                try:
                    button = card.find_element(By.CSS_SELECTOR, "button.css-13df51w")
                    product_data['add_to_cart_available'] = button.is_enabled()
                    product_data['button_text'] = button.text.strip()
                except:
                    product_data['add_to_cart_available'] = False
                    product_data['button_text'] = None
                
                products_data.append(product_data)
                
            except Exception as e:
                print(f"Error scraping product {i+1}: {e}")
                continue
        
    except Exception as e:
        print(f"Error during scraping: {e}")
    
    finally:
        driver.quit()
    
    # Convert to DataFrame
    print(f"\nConverting {len(products_data)} products to DataFrame...")
    df = pd.DataFrame(products_data)
    
    if not df.empty:
        # Display basic info about the DataFrame
        print(f"\nDataFrame Info:")
        print(f"Shape: {df.shape}")
        print(f"Columns: {list(df.columns)}")
        print(f"\nFirst few rows:")
        print(df.head())
        
        # Data cleaning and processing
        print("\n=== Data Processing ===")
        
        # Check for missing values
        print("Missing values per column:")
        print(df.isnull().sum())
        
        # Basic statistics for numeric columns
        if 'price_numeric' in df.columns and df['price_numeric'].notna().any():
            print("\nPrice statistics:")
            print(df['price_numeric'].describe())
        
        # Save in different formats
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        
        for format_type in save_formats:
            try:
                if format_type == 'csv':
                    filename = f'oxylabs_products_{timestamp}.csv'
                    df.to_csv(filename, index=False)
                    print(f"✓ Saved as CSV: {filename}")
                
                elif format_type == 'json':
                    filename = f'oxylabs_products_{timestamp}.json'
                    df.to_json(filename, orient='records', indent=2)
                    print(f"✓ Saved as JSON: {filename}")
                
                elif format_type == 'excel':
                    filename = f'oxylabs_products_{timestamp}.xlsx'
                    df.to_excel(filename, index=False, sheet_name='Products')
                    print(f"✓ Saved as Excel: {filename}")
                
            except Exception as e:
                print(f"✗ Error saving as {format_type}: {e}")
    
    return df

# Run the scraping and storage demo
scraped_df = scrape_and_store_data(max_products=10)
```
