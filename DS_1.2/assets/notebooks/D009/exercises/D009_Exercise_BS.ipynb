# Book Catalog Data Analysis - Web Scraping Project with Books to Scrape

## Project Overview
Create a web scraping application to collect book data from Books to Scrape (https://books.toscrape.com/), a sandbox website specifically designed for learning web scraping. This project will help you understand web scraping fundamentals while building a book catalog database with pricing and category analysis.

## Learning Objectives
- Build a web scraper using Python and BeautifulSoup
- Handle multiple pages and categories
- Extract structured data from HTML
- Perform data analysis on book pricing and categories
- Create visualizations of book market data
- Practice error handling and data cleaning

## Data Collection Requirements

### Book Information to Extract
1. Basic Book Details
   - Title
   - Price
   - Availability
   - Book Cover Image URL
   - Rating (Star Rating)

2. Save all the scrapped data into csv using dataframe.
