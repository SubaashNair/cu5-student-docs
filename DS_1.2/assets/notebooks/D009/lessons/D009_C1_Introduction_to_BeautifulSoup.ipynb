# Introduction to BeautifulSoup

This notebook provides a scaffolded introduction to web scraping using BeautifulSoup. We will use the Wikipedia page for Malaysia as our example.

## Ethical Web Scraping

When scraping websites, it's crucial to be respectful of their terms of service and robots.txt file. Wikimedia provides a robot policy that outlines best practices for accessing their content. Some key points include identifying your bot with a User-Agent and respecting rate limits. You can read the full policy [here](https://wikitech.wikimedia.org/wiki/Robot_policy).

## 1. Fetching the Webpage

First, we need to fetch the HTML content of the webpage. We use the `requests` library for this. We will also include a `User-Agent` in our request headers to identify our bot, and add error handling.

```python
import requests

url = 'https://en.wikipedia.org/wiki/Malaysia'
headers = {
    'User-Agent': 'MyCoolBot/0.1 (https://example.com/bot; myemail@example.com)'
}

try:
    response = requests.get(url, headers=headers)
    response.raise_for_status()  # Raise an exception for bad status codes
    html_content = response.content
except requests.exceptions.RequestException as e:
    print(f'Error fetching the webpage: {e}')
    html_content = None
```

## 2. Parsing with BeautifulSoup

Now that we have the HTML content, we can parse it using BeautifulSoup. We'll use the `html.parser`.

```python
from bs4 import BeautifulSoup

if html_content:
    soup = BeautifulSoup(html_content, 'html.parser')
```

## 3. Finding Elements by Tag

We can find the first occurrence of a tag using the `find()` method. Let's find the first paragraph (`<p>`) tag.

```python
if 'soup' in locals():
    first_paragraph = soup.find('p')
    print(first_paragraph)
```

## 4. Finding Elements by Class and ID

We can also find elements by their CSS class or ID. Let's find the infobox, which has the class `infobox`.

```python
if 'soup' in locals():
    infobox = soup.find(class_='infobox')
    print(infobox)
```

## 5. Extracting Text and Attributes

We can extract the text from an element using the `get_text()` method.

```python
if 'infobox' in locals():
    infobox_text = infobox.get_text()
    print(infobox_text)
```

We can also extract attributes from an element. Let's get the URL of the first link in the infobox.

```python
if 'infobox' in locals():
    first_link = infobox.find('a')
    if first_link:
        link_url = first_link.get('href')
        print(link_url)
```

## 6. Finding Multiple Elements

We can find all occurrences of a tag using the `find_all()` method. Let's find all the links (`<a>`) in the infobox.

```python
if 'infobox' in locals():
    all_links = infobox.find_all('a')
    for link in all_links:
        print(link.get('href'))
```

## 7. Navigating the Parse Tree

BeautifulSoup provides a way to navigate the parse tree. For example, we can find the parent of an element.

```python
if 'first_link' in locals() and first_link:
    parent_of_first_link = first_link.parent
    print(parent_of_first_link)
```

## 8. Practical Example: Scraping the Infobox

Let's put everything together to scrape the infobox from the Malaysia Wikipedia page and extract some key information.

```python
import pandas as pd

if 'soup' in locals():
    infobox = soup.find(class_='infobox')
    if infobox:
        infobox_rows = infobox.find_all('tr')

        infobox_data = []
        for row in infobox_rows:
            header = row.find('th')
            data = row.find('td')
            if header and data:
                infobox_data.append((header.get_text(strip=True), data.get_text(strip=True)))

        df = pd.DataFrame(infobox_data, columns=['Header', 'Data'])
        print(df)
```
