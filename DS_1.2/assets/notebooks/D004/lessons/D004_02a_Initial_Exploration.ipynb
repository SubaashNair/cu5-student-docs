## 1. Introduction to Exploratory Data Analysis (EDA)

### What is EDA?

**Exploratory Data Analysis (EDA)** is an approach to analyzing datasets to summarize their main characteristics, often using visual methods. It helps us:

- Understand the data structure and content
- Detect outliers and anomalies
- Discover patterns and relationships
- Test underlying assumptions
- Determine optimal factor settings

### Why is EDA Important?

EDA is crucial because:
1. **Data Quality Assessment**: Identify errors, missing values, and inconsistencies
2. **Feature Understanding**: Learn about variable distributions and relationships
3. **Model Selection**: Inform decisions about which models to use
4. **Hypothesis Generation**: Discover interesting patterns worth investigating

## 4. Data Cleaning

### 4.1 Missing Values

Missing data can significantly impact analysis. Let's identify them.

![Missing Data Heatmap Example](https://bookdown.org/mike/data_analysis/13-imputation_files/figure-html/fig-missing-data-heatmap1-1.png)
*Source: A Guide on Data Analysis - Heatmap showing missing data patterns*

```python
# Count missing values
missing_values = df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Create a summary dataframe
missing_summary = pd.DataFrame({
'Missing_Count': missing_values,
'Percentage': missing_percentage
})

# Display only columns with missing values
missing_summary = missing_summary[missing_summary['Missing_Count'] > 0].sort_values(
'Missing_Count', ascending=False
)

print("Missing Values Summary:")
print(missing_summary)

if missing_summary.empty:
    print("\n No missing values found!")
```

#### Visualize Missing Data

```python
# Visualize missing data
if not missing_summary.empty:
    plt.figure(figsize=(10, 6))
    plt.barh(missing_summary.index, missing_summary['Percentage'])
    plt.xlabel('Percentage of Missing Values')
    plt.title('Missing Values by Column')
    plt.tight_layout()
    plt.show()
else:
    print("No missing values to visualize.")
```

### 4.2 Duplicate Records

Duplicate records can skew analysis results.

```python
# Check for duplicates
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")

if duplicate_count > 0:
print(f"  Found {duplicate_count} duplicate rows!")
# Display duplicate rows
print("\nDuplicate rows:")
print(df[df.duplicated(keep=False)])
else:
print(" No duplicate rows found!")
```

###  Teaching Point: Handling Missing Data

When dealing with missing data, we have several strategies:

1. **Deletion**:
- Remove rows with missing values
- Use when: Missing data is < 5% of total

2. **Imputation**:
- **Mean/Median**: For numerical data
- **Mode**: For categorical data
- **Forward/Backward Fill**: For time-series data
- **Predictive Models**: Use ML to predict missing values

3. **Keep as Missing**:
- Some algorithms can handle missing values
- Missing might be informative itself

**Rule of Thumb**: If > 50% of a column is missing, consider dropping that column.

## 5. Univariate Analysis

Univariate analysis examines each variable individually to understand its distribution.

### 5.1 Numerical Variables Analysis

#### Understanding Histograms

![Histogram Explanation](https://www.mathsisfun.com/data/images/histogram-heights.svg)
*A histogram shows the frequency distribution of numerical data*

#### Understanding Box Plots

![Box and Whisker Plot](https://www.mathsisfun.com/data/images/box-whisker-plot.svg)
*A box plot displays the five-number summary: minimum, Q1, median, Q3, and maximum*

![Interquartile Range](https://www.mathsisfun.com/data/images/interquartile-range.svg)
*The Interquartile Range (IQR) spans from Q1 to Q3*

*Source: Math is Fun*

```python
# Select numerical columns
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()

print("Numerical columns:")
print(numerical_cols)
```

#### 5.1.1 Marriage Duration Distribution

```python
# Histogram and Box Plot for Marriage Duration
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(df['marriage_duration'].dropna(), bins=30, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Marriage Duration (years)')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Marriage Duration')
axes[0].axvline(df['marriage_duration'].mean(), color='red', linestyle='--',
label=f'Mean: {df["marriage_duration"].mean():.2f}')
axes[0].axvline(df['marriage_duration'].median(), color='green', linestyle='--',
label=f'Median: {df["marriage_duration"].median():.2f}')
axes[0].legend()

# Box Plot
axes[1].boxplot(df['marriage_duration'].dropna(), vert=True)
axes[1].set_ylabel('Marriage Duration (years)')
axes[1].set_title('Box Plot of Marriage Duration')
axes[1].set_xticklabels(['Marriage Duration'])

plt.tight_layout()
plt.show()

# Statistics
print(f"Mean: {df['marriage_duration'].mean():.2f} years")
print(f"Median: {df['marriage_duration'].median():.2f} years")
print(f"Std Dev: {df['marriage_duration'].std():.2f} years")
```

#### 5.1.2 Income Distribution (Men vs Women)

```python
# Compare income distributions
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Men's Income - Histogram
axes[0, 0].hist(df['income_man'].dropna(), bins=50, edgecolor='black', alpha=0.7, color='blue')
axes[0, 0].set_xlabel('Income (Man)')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title('Distribution of Men\'s Income')
axes[0, 0].axvline(df['income_man'].mean(), color='red', linestyle='--',
label=f'Mean: {df["income_man"].mean():.2f}')
axes[0, 0].legend()

# Men's Income - Box Plot
axes[0, 1].boxplot(df['income_man'].dropna(), vert=True)
axes[0, 1].set_ylabel('Income (Man)')
axes[0, 1].set_title('Box Plot of Men\'s Income')

# Women's Income - Histogram
axes[1, 0].hist(df['income_woman'].dropna(), bins=50, edgecolor='black', alpha=0.7, color='pink')
axes[1, 0].set_xlabel('Income (Woman)')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_title('Distribution of Women\'s Income')
axes[1, 0].axvline(df['income_woman'].mean(), color='red', linestyle='--',
label=f'Mean: {df["income_woman"].mean():.2f}')
axes[1, 0].legend()

# Women's Income - Box Plot
axes[1, 1].boxplot(df['income_woman'].dropna(), vert=True)
axes[1, 1].set_ylabel('Income (Woman)')
axes[1, 1].set_title('Box Plot of Women\'s Income')

plt.tight_layout()
plt.show()

# Summary statistics
print("\n Income Comparison:")
print(f"Men - Mean: {df['income_man'].mean():.2f}, Median: {df['income_man'].median():.2f}")
print(f"Women - Mean: {df['income_woman'].mean():.2f}, Median: {df['income_woman'].median():.2f}")
```

#### 5.1.3 Number of Kids Distribution

```python
# Distribution of number of kids
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Count plot
kids_counts = df['num_kids'].value_counts().sort_index()
axes[0].bar(kids_counts.index, kids_counts.values, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Number of Kids')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Number of Children')
axes[0].grid(axis='y', alpha=0.3)

# Pie chart
axes[1].pie(kids_counts.values, labels=kids_counts.index, autopct='%1.1f%%', startangle=90)
axes[1].set_title('Proportion of Number of Children')

plt.tight_layout()
plt.show()

print("\nNumber of Kids Statistics:")
print(df['num_kids'].value_counts().sort_index())
```

### Quick Knowledge Check

Before moving to bivariate analysis, test your understanding:

1. **What does a right-skewed distribution tell us about the data?**
   - More values cluster at the lower end with a tail extending right

2. **When should you use median instead of mean?**
   - When data has outliers or is skewed

3. **What does a box plot's "whisker" represent?**
   - Typically 1.5 x IQR from the quartiles (or min/max if closer)

4. **If the mean is greater than the median, what does that suggest?**
   - The distribution is right-skewed (positively skewed)

## 6. Bivariate and Multivariate Analysis

Bivariate analysis examines relationships between two variables.

### 6.1 Correlation Analysis

#### Understanding Correlation

![Correlation Examples](https://www.mathsisfun.com/data/images/correlation-examples.svg)
*Correlation measures the strength and direction of relationships between variables*

**Correlation Coefficient (r)**:
- **r = 1**: Perfect positive correlation
- **r = 0**: No correlation
- **r = -1**: Perfect negative correlation

```python
# Calculate correlation matrix
correlation_matrix = df[numerical_cols].corr()

# Create heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
square=True, linewidths=1, fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features', fontsize=16, pad=20)
plt.tight_layout()
plt.show()
```

### 6.2 Relationship Between Income (Men vs Women)

```python
# Scatter plot: Income Man vs Income Woman
plt.figure(figsize=(10, 6))
plt.scatter(df['income_man'], df['income_woman'], alpha=0.5, edgecolor='black', linewidth=0.5)
plt.xlabel('Income (Man)', fontsize=12)
plt.ylabel('Income (Woman)', fontsize=12)
plt.title('Relationship Between Men\'s and Women\'s Income', fontsize=14)

# Add diagonal line for reference
max_income = max(df['income_man'].max(), df['income_woman'].max())
plt.plot([0, max_income], [0, max_income], 'r--', label='Equal Income Line', linewidth=2)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Calculate correlation
income_corr = df['income_man'].corr(df['income_woman'])
print(f"\nCorrelation between Men's and Women's Income: {income_corr:.3f}")
```

### 6.3 Marriage Duration vs Number of Kids

```python
# Box plot: Marriage Duration by Number of Kids
plt.figure(figsize=(12, 6))

# Prepare data - remove NaN values
data_for_plot = df[['marriage_duration', 'num_kids']].dropna()

# Create box plot
sns.boxplot(data=data_for_plot, x='num_kids', y='marriage_duration', palette='Set2')
plt.xlabel('Number of Kids', fontsize=12)
plt.ylabel('Marriage Duration (years)', fontsize=12)
plt.title('Marriage Duration by Number of Children', fontsize=14)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Summary statistics by number of kids
print("\nAverage Marriage Duration by Number of Kids:")
print(df.groupby('num_kids')['marriage_duration'].agg(['mean', 'median', 'count']))
```

### 6.4 Education Level vs Income

```python
# Box plots: Income by Education Level
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Men's Income by Education
sns.boxplot(data=df, x='education_man', y='income_man', ax=axes[0], palette='Blues')
axes[0].set_xlabel('Education Level', fontsize=12)
axes[0].set_ylabel('Income', fontsize=12)
axes[0].set_title('Men\'s Income by Education Level', fontsize=14)
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(axis='y', alpha=0.3)

# Women's Income by Education
sns.boxplot(data=df, x='education_woman', y='income_woman', ax=axes[1], palette='Reds')
axes[1].set_xlabel('Education Level', fontsize=12)
axes[1].set_ylabel('Income', fontsize=12)
axes[1].set_title('Women\'s Income by Education Level', fontsize=14)
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# Summary statistics
print("\nMean Income by Education Level - Men:")
print(df.groupby('education_man')['income_man'].mean().sort_values(ascending=False))

print("\nMean Income by Education Level - Women:")
print(df.groupby('education_woman')['income_woman'].mean().sort_values(ascending=False))
```

## 7. Time-Based Analysis

Let's explore trends over time.

```python
# Convert date columns to datetime
df['divorce_date'] = pd.to_datetime(df['divorce_date'])
df['marriage_date'] = pd.to_datetime(df['marriage_date'])

# Extract year from divorce date
df['divorce_year'] = df['divorce_date'].dt.year

# Count divorces by year
divorces_by_year = df['divorce_year'].value_counts().sort_index()

# Plot divorces over time
plt.figure(figsize=(14, 6))
plt.plot(divorces_by_year.index, divorces_by_year.values, marker='o', linewidth=2, markersize=8)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Number of Divorces', fontsize=12)
plt.title('Divorce Trends Over Time', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nDivorces by Year:")
print(divorces_by_year)
```
