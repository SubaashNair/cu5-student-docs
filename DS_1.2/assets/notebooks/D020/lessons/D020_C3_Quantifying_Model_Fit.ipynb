# Quantifying Model Fit

## 1. Introduction: Predicting is not enough

We have learned how to fit a linear regression model and make predictions. But simply drawing a line through data points doesn't mean it's a *good* line.

**Goodness of Fit** refers to how well the model describes the observed data. In this notebook, we will explore key metrics to quantify this fit:
1.  **Coefficient of Determination ($R^2$)**
2.  **Residual Standard Error (RSE)**
3.  **Root Mean Squared Error (RMSE)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

sns.set_theme(style="whitegrid")
```

```python
# Load data and fit model (Baseline)
file_path = "../../00-data/Introduction to Regression Fish.csv"
df = pd.read_csv(file_path)

X = df['length_cm']
y = df['mass_g']
X_const = sm.add_constant(X)
model = sm.OLS(y, X_const).fit()
print(model.summary())
```

## 2. Coefficient of Determination ($R^2$)

The **$R^2$** value is the most common metric for regression. It represents the **proportion of the variance for the dependent variable that's explained by an independent variable**.

### **Formula**
$$ R^2 = 1 - \frac{RSS}{TSS} $$

Where:
*   **RSS (Residual Sum of Squares):** $\sum (y_i - \hat{y}_i)^2$. The variation that is left unexplained by the model.
*   **TSS (Total Sum of Squares):** $\sum (y_i - \bar{y})^2$. The total variation in the data (variance).

### **Interpretation**
*   **$R^2 = 1$**: Perfect fit. All tracking points lie on the line. (0 residuals).
*   **$R^2 = 0$**: The model explains none of the variability (basically fitting the mean line).
*   **Example**: An $R^2$ of **0.87** means **87% of the variation in Mass can be explained by Length**.

![R Squared](https://upload.wikimedia.org/wikipedia/commons/thumb/8/86/Coefficient_of_Determination.svg/1200px-Coefficient_of_Determination.svg.png)
*(Image source: Wikimedia Commons)*

```python
# Extracting R-squared directly
r_squared = model.rsquared
print(f"R-squared: {r_squared:.4f}")
```

## 3. Residual Standard Error (RSE)

While $R^2$ is a proportion (unitless), **RSE** gives us a measure of error in the **units of the dependent variable** (grams).

It estimates the standard deviation of the error term $\epsilon$. Roughly speaking, it is the average amount that the response will deviate from the true regression line.

### **Formula**
$$ RSE = \sqrt{\frac{RSS}{n - 2}} $$

*   $RSS$: Residual Sum of Squares.
*   $n$: Number of observations.
*   We divide by $n-2$ because 2 degrees of freedom were used to estimate $\beta_0$ and $\beta_1$.

### **Interpretation**
*   If RSE = 145, it means any prediction of mass is typically off by about **145 grams**.
*   Lower RSE is better.

```python
# Calculating RSE Manually
residuals = model.resid
RSS = (residuals ** 2).sum()
n = len(df)
RSE = np.sqrt(RSS / (n - 2))

print(f"Residual Sum of Squares (RSS): {RSS:.2f}")
print(f"Residual Standard Error (RSE): {RSE:.2f} grams")
```

## 4. Root Mean Squared Error (RMSE)

**RMSE** is very similar to RSE and is the standard metric used in Machine Learning. 

### **Formula**
$$ RMSE = \sqrt{\frac{\sum (y_i - \hat{y}_i)^2}{n}} = \sqrt{\frac{RSS}{n}} $$

The difference is that it divides by $n$ instead of $n-2$. For large datasets, RSE and RMSE are almost identical.

### **Why use RMSE?**
*   It penalizes large errors heavily (due to squaring).
*   Standard metric for comparing different ML models.

```python
# Calculating RMSE
from statsmodels.tools.eval_measures import rmse

y_pred = model.predict(X_const)
rmse_val = rmse(y, y_pred)

print(f"Root Mean Squared Error (RMSE): {rmse_val:.2f} grams")
```

## 5. Visualizing the Error

A good model should have residuals that are randomly scattered around zero. Any pattern (U-shape, funnel) suggests a bad fit.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=y_pred, y=residuals)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel('Predicted Mass')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```
