## The one idea to hold on to

You almost never have the population. Every number you report is a **sample estimate that
would have come out differently with different rows**. The bootstrap is how you measure
that "differently" for any statistic at all — including ones with no formula.

# How this is used in data science and machine learning

This is the most directly transferable notebook in the module. Sampling is not a survey
topic — it is what a train/test split *is*, and the bootstrap is how you put error bars on
a model.

### 1. A train/test split is a random sample, and inherits every problem in this notebook
`train_test_split` draws exactly the simple random sample you drew above. So:
- **Convenience sampling is what you do accidentally.** Splitting time-ordered data at
  random lets the model see the future — the same bias you saw with `df.head(10)`, just
  harder to spot.
- **Stratified sampling is the fix for rare classes.** `train_test_split(..., stratify=y)`
  preserves the class ratio in both halves. Without it, a 2%-positive target can produce a
  test set with almost no positives, and a meaningless score.
- **Cluster sampling is `GroupKFold`.** When rows share a group — several visits by one
  patient, several sessions by one user — splitting *within* the group leaks. You must
  sample whole groups, exactly as you sampled whole coffee varieties above.

### 2. Weighted sampling is how imbalanced targets get handled
The `weights=` argument you used is the same mechanism as `class_weight='balanced'`, as
oversampling a minority class, and as propensity weighting in causal work.

### 3. The bootstrap is how you know whether a model actually improved
This is the highest-value technique in the module.

A test-set accuracy of `0.86` is **one draw from a distribution**. Model B scoring `0.88`
may be genuinely better, or may be the same model on a luckier test set. Resample the test
set with replacement 1,000 times, score both models on each resample, and you get a
distribution of the *difference*. If that distribution comfortably excludes zero, the
improvement is real.

```python
# the pattern, applied to a model metric instead of a column mean
diffs = []
for _ in range(1000):
    idx = rng.choice(len(y_test), len(y_test), replace=True)
    diffs.append(score(y_test[idx], pred_b[idx]) - score(y_test[idx], pred_a[idx]))
lower, upper = np.percentile(diffs, [2.5, 97.5])   # 95% CI on the improvement
```

It works for **any** metric — AUC, F1, RMSE, revenue per user — including ones with no
standard-error formula. That is why it is worth more than any single test in Lesson 03.

### 4. Bootstrapping is literally the "B" in bagging
A Random Forest fits each tree on a bootstrap resample of the training rows, in the same
way you resampled rows above. "Bagging" is short for **b**ootstrap **agg**regating, and
out-of-bag error is scored on the rows a given resample happened to miss.

### 5. Standard error is why you should distrust a single number
The formula `SE = s / sqrt(n)` says accuracy improves with the **square root** of sample
size. To halve your error you need four times the data. That arithmetic is behind every
"is it worth collecting more data?" conversation you will have.

### 6. Confidence intervals turn a number into a decision
"Model A: 0.86" is a number. "Model A: 0.86, 95% CI [0.83, 0.89]; Model B: 0.88,
95% CI [0.85, 0.91]" is a decision — and it tells you honestly that these two models have
not yet been distinguished.
