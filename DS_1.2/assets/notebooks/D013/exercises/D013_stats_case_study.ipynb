# Case Study: Food Consumption & Statistics

## Introduction
In this case study, we will apply statistical concepts learned in previous chapters (Summary Statistics, Random Numbers & Probability, Distributions & CLT) to a real-world dataset: **`food_consumption.csv`**.

This dataset contains information about food consumption and CO2 emissions for various countries and food categories.
