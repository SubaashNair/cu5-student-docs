## Problem Description

You have been hired to build a **Library Management System** for a local library. The system needs to:

1. **Manage Books**: Store book information (title, author, ISBN, genre, copies available)
2. **Manage Members**: Store member information (ID, name, email, membership type)
3. **Track Borrowing**: Record which books are borrowed by which members
4. **Calculate Fines**: Calculate late fees for overdue books
5. **Generate Reports**: Create reports in various formats (text, CSV, JSON)
6. **Data Persistence**: Save and load data from files

### Data Structures

1. **Books Database** (Dictionary):
   - Key: ISBN (string)
   - Value: Dictionary with keys: `title`, `author`, `genre`, `total_copies`, `available_copies`, `borrowers` (list of member IDs)

2. **Members Database** (Dictionary):
   - Key: Member ID (string)
   - Value: Dictionary with keys: `name`, `email`, `membership_type`, `borrowed_books` (set of ISBNs), `fine_balance` (float)

3. **Transaction History** (List of Tuples):
   - Each tuple: `(member_id, isbn, action, date)`
   - Actions: `'borrow'`, `'return'`

4. **Genres** (Set):
   - Collection of all unique book genres in the library

### Required Features

Your program should provide a menu with the following options:

1. **Add Book**: Add a new book to the catalog
2. **Register Member**: Register a new library member
3. **Borrow Book**: Allow a member to borrow a book
4. **Return Book**: Process a book return
5. **Search Books**: Search books by title, author, or genre
6. **Calculate Fines**: Calculate and update fines for overdue books
7. **View Statistics**: Display library statistics
8. **Generate Reports**: Create reports in text, CSV, or JSON format
9. **Save Data**: Save all data to files (JSON format)
10. **Load Data**: Load data from files
11. **Exit**: Save data and exit

```python
# TODO: Implement add_book function
def add_book(books, genres):
    """
    Add a new book to the library catalog.
    
    Parameters:
    - books: Dictionary of books
    - genres: Set of genres
    
    Returns:
    - True if book added successfully, False otherwise
    """
    # Your code here
    # 1. Get ISBN from user (use .strip() and .upper() string methods)
    isbn = input('Enter ISBN').strip().upper()
    # 2. Check if ISBN already exists (use 'in' operator)
    if isbn in books:
        print(f'Error: Book with {isbn} already exists.')
        return False
    # 3. Get book details (title, author, genre, total_copies)
    title = input('Enter book title: ').strip()
    author = input('Enter book author: ').strip()
    genre = input('Enter book genre: ').strip()
    # 4. Validate total_copies is a positive integer (use int() and conditional)
    try:
        total_copies = int(input('Enter total copies'))
        if total_copies <=0:
            print('Total copies must be positive integer')
            return False
    except ValueError:
        print('Total copies must be a number')
        return False
    # 5. Add book to dictionary
    books[isbn] = {
        'title':title,
        'author':author,
        'genre':genre,
        'total_copies':total_copies,
        'available_copies':total_copies,
        'borrowers':[]
    }
    # 6. Add genre to genres set
    genres.add(genre)
    # 7. Return True/False
    print(f"Book {title} added successfully")
    return True
    # pass
```
