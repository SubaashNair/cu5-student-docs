```python
from __future__ import annotations

from pathlib import Path
from pypdf import PdfReader
import requests

DATA_DIR = Path("case_study_pdf_data")
DATA_DIR.mkdir(exist_ok=True)

main_pdf_url = "https://stacks.cdc.gov/view/cdc/152724/cdc_152724_DS1.pdf"
extension_pdf_url = "https://www.w3.org/WAI/WCAG20/Techniques/working-examples/PDF20/table.pdf"

main_pdf_path = DATA_DIR / "brfss_user_guide.pdf"
extension_pdf_path = DATA_DIR / "w3c_table_example.pdf"

def safe_extract_text(page) -> str:
    return page.extract_text() or ""
```

## Step 1 - Download the PDF files

Run the next cell to download the main case study PDF. The extension PDF is optional.


```python
downloads = [
    (main_pdf_url, main_pdf_path),
    (extension_pdf_url, extension_pdf_path),
]

for url, path in downloads:
    if path.exists():
        print(f"Already available: {path}")
        continue

    print(f"Downloading: {url}")
    response = requests.get(url, timeout=20)
    response.raise_for_status()
    path.write_bytes(response.content)
    print(f"Saved to: {path} ({len(response.content):,} bytes)")
```
