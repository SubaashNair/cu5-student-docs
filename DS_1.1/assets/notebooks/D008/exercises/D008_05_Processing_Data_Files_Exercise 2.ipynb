# Exercise 5: JSON File Processing Exercise
## Working with Stock Market Data

This exercise will teach you how to read, process, and analyze JSON data in Python.

## Part 1: Reading JSON Data

First, let's load the stock data from the JSON file.

```python
import json

# Read the JSON file
with open('../../00-data/stocks-data.json', 'r') as file:
    stocks = json.load(file)

# Display the first stock entry
print("First stock entry:")
print(json.dumps(stocks[0], indent=2))
print(f"\nTotal number of stocks: {len(stocks)}")
```

## Part 2: Basic Data Extraction

**Exercise 1:** Extract all company names from the dataset.

```python
# Your code here
company_names = [stock['company'] for stock in stocks]

print(company_names)
```

**Exercise 2:** Extract all stock symbols.

```python
# Your code here
symbols = [stock['symbol'] for stock in stocks]

print(symbols)
```

## Part 3: Filtering and Searching

**Exercise 3:** Find all companies whose names contain the word "Inc."

```python
# Your code here
inc_companies = [stock['company'] for stock in stocks if 'Inc' in stock['company']]

print(inc_companies)
```

**Exercise 4:** Find stocks where the initial_price was greater than $50.

```python
# Your code here
expensive_stocks = [stock for stock in stocks if stock['initial_price']>50]
expensive_stocks
```

## Part 4: Data Analysis

**Exercise 5:** Calculate the average initial price of all stocks.

```python
# Your code here
average_initial_price = sum(stock['initial_price'] for stock in stocks)/ len(stocks)

print(f"Average initial price: ${average_initial_price:.2f}")
```

**Exercise 6:** Calculate the average price in 2007.

```python
# Your code here
average_2007_price = sum(stock['price_2007'] for stock in stocks) / len(stocks)

print(f"Average 2007 price: ${average_2007_price:.2f}")
```

**Exercise 7:** Find the stock with the highest price in 2007.

```python
# Your code here
highest_2007 = max(stocks,key=lambda stock: stock['price_2007'])

print(f"Highest stock in 2007: {highest_2007['company']} at ${highest_2007['price_2007']}")
```

## Part 5: Price Change Analysis

**Exercise 8:** Calculate which stocks gained value from initial price to 2007.

```python
# Your code here
gainers = [stock for stock in stocks if stock['price_2007']>stock['initial_price']]

print(f"Number of stocks that gained: {len(gainers)}")
for stock in gainers:
    gain = stock['price_2007'] - stock['initial_price']
    print(f"{stock['company']}: ${gain:.2f}")
```

**Exercise 9:** Calculate the percentage change from initial price to 2007 for each stock.

```python
# Your code here
for stock in stocks:
    percentage_change = ((stock['price_2007'] - stock['initial_price']) / stock['initial_price'])*100
    print(f"{stock['company']}: {percentage_change:.2f}%")
```

## Part 6: Creating New JSON Data

**Exercise 10:** Create a new list of dictionaries containing only company name, symbol, and 2007 price. Then save it to a new JSON file.

```python
# Your code here
simplified_data = [{
    'company':stock['company'],
    'symbol':stock['symbol'],
    'price_2007':stock['price_2007']
}
                  for stock in stocks]

# Save to new file
with open('stocks-2007-summary.json', 'w') as file:
    json.dump(simplified_data, file, indent=2)

print("Data saved to stocks-2007-summary.json")
print(json.dumps(simplified_data[0], indent=2))
```

## Part 7: Advanced Analysis

**Exercise 11:** Find the stock with the best percentage gain from initial price to 2007.

```python
# Your code here
best_performer = max(stocks,key=lambda s: (s['price_2007']-s['initial_price'])/s['initial_price'])

print(f"Best performer: {best_performer}")
```

**Exercise 12:** Create a dictionary grouping stocks by their performance (gained vs lost value).

```python
# Your code here
performance_groups = {
    'gainers': [],
    'losers': []
}

for s in stocks:
    if s['price_2007']>s['initial_price']:
        performance_groups['gainers'].append(s['company'])
    else:
        performance_groups['losers'].append(s['company'])


print(f"Gainers: {len(performance_groups['gainers'])}")
print(f"Losers: {len(performance_groups['losers'])}")
```

## Bonus Challenge

**Exercise 13:** Write a function that takes a company name and returns all its information in a formatted string.

```python
def get_stock_info(company_name):
    # Your code here
    for s in stocks:
        if company_name.lower() in s['company'].lower():
            pct_change = ((s['price_2007']-s['initial_price']) /s['initial_price'])*100
            return (
                f"Company: {s['company']}\n"
                f"Symbol: {s['symbol']}\n"
                f"Initial Price: ${s['initial_price']:.2f}\n"
                f"Price in 2007: ${s['price_2007']:.2f}\n"
                f"Change: {pct_change:+.2f}"
            )
    return f"No stock found for company '{company_name}'"

# Test your function
print(get_stock_info("Microsoft"))
print(get_stock_info("Apple"))
print(get_stock_info("Unknown"))
```
