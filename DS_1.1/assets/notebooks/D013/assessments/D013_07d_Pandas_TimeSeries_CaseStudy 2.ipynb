### Solution 1: Basic Time Selection

```python
# =============================================================================
# SOLUTION 1: Basic Time Selection
# =============================================================================

print("=== SOLUTION 1 ===")

# a) Trips on March 15, 2023
# Partial string indexing selects all hours on that date
march_15 = df.loc['2023-03-15']
total_trips_march_15 = march_15['trips'].sum()
print(f"\na) Total trips on March 15, 2023: {total_trips_march_15:,}")

# b) Total revenue for June 2023
# 'YYYY-MM' format selects all hours in that month
june_2023 = df.loc['2023-06']
june_revenue = june_2023['revenue'].sum()
print(f"\nb) Total revenue for June 2023: ${june_revenue:,.2f}")

# c) Q1 2024 average hourly trips
# Date range slicing with inclusive endpoints
q1_2024 = df.loc['2024-01-01':'2024-03-31']
q1_avg_trips = q1_2024['trips'].mean()
print(f"\nc) Q1 2024 average hourly trips: {q1_avg_trips:.1f}")
```

### Solution 2: Hourly Demand Patterns

```python
# =============================================================================
# SOLUTION 2: Hourly Demand Patterns
# =============================================================================

print("=== SOLUTION 2 ===")

# a) Average trips per hour
hourly_avg = df.groupby('hour')['trips'].mean()
print("\na) Average trips by hour:")
print(hourly_avg.round(1))

# b) Peak and off-peak hours
peak_hour = hourly_avg.idxmax()
offpeak_hour = hourly_avg.idxmin()
print(f"\nb) Peak hour: {peak_hour}:00 ({hourly_avg[peak_hour]:.1f} avg trips)")
print(f"   Off-peak hour: {offpeak_hour}:00 ({hourly_avg[offpeak_hour]:.1f} avg trips)")

# c) Bar chart
fig, ax = plt.subplots(figsize=(12, 5))
colors = ['coral' if h in [7, 8, 17, 18] else 'steelblue' for h in range(24)]
ax.bar(range(24), hourly_avg.values, color=colors, alpha=0.7)
ax.set_xlabel('Hour of Day')
ax.set_ylabel('Average Trips')
ax.set_title('Average Hourly Trips (Rush Hours Highlighted)', fontweight='bold')
ax.set_xticks(range(24))
plt.tight_layout()
plt.show()

# d) Weekday vs Weekend pattern
weekday_hourly = df[~df['is_weekend']].groupby('hour')['trips'].mean()
weekend_hourly = df[df['is_weekend']].groupby('hour')['trips'].mean()

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(range(24), weekday_hourly.values, 'b-o', label='Weekday', linewidth=2)
ax.plot(range(24), weekend_hourly.values, 'r-s', label='Weekend', linewidth=2)
ax.set_xlabel('Hour of Day')
ax.set_ylabel('Average Trips')
ax.set_title('Hourly Demand: Weekday vs Weekend', fontweight='bold')
ax.set_xticks(range(24))
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Solution 3: Resampling and Aggregation

```python
# =============================================================================
# SOLUTION 3: Resampling and Aggregation
# =============================================================================

print("=== SOLUTION 3 ===")

# a) Daily totals
daily_totals = df[['trips', 'revenue']].resample('D').sum()
print("\na) Daily totals (first 10 days):")
print(daily_totals.head(10).round(2))

# b) Weekly averages
weekly_avg = df[['trips', 'revenue', 'active_drivers']].resample('W').mean()
print("\nb) Weekly averages (first 8 weeks):")
print(weekly_avg.head(8).round(2))

# c) Monthly statistics with multiple aggregations
monthly_stats = df.resample('ME').agg({
    'trips': ['sum', 'max'],
    'revenue': 'sum',
    'avg_fare': 'mean'
})
# Flatten column names
monthly_stats.columns = ['total_trips', 'max_hourly_trips', 'total_revenue', 'avg_fare']
print("\nc) Monthly statistics:")
print(monthly_stats.round(2))

# d) Month-over-month growth rate
monthly_trips = df['trips'].resample('ME').sum()
mom_growth = monthly_trips.pct_change() * 100
print("\nd) Month-over-Month growth rate:")
print(mom_growth.round(2))

highest_growth_month = mom_growth.idxmax()
print(f"\n   Highest growth: {highest_growth_month.strftime('%B %Y')} ({mom_growth[highest_growth_month]:.1f}%)")
```

### Solution 4: Rolling Statistics

```python
# =============================================================================
# SOLUTION 4: Rolling Statistics
# =============================================================================

print("=== SOLUTION 4 ===")

# a) 24-hour rolling average of trips
df_rolling = df.copy()
df_rolling['trips_24h_ma'] = df['trips'].rolling(window=24).mean()
print("\na) 24-hour rolling average (sample):")
print(df_rolling[['trips', 'trips_24h_ma']].iloc[20:30].round(1))

# b) 7-day rolling average of daily revenue
daily_revenue = df['revenue'].resample('D').sum()
revenue_7d_ma = daily_revenue.rolling(window=7).mean()
print("\nb) 7-day rolling average of daily revenue (sample):")
print(revenue_7d_ma.head(14).round(2))

# c) 30-day rolling standard deviation
daily_trips = df['trips'].resample('D').sum()
trips_30d_std = daily_trips.rolling(window=30).std()
print("\nc) 30-day rolling std of daily trips (sample):")
print(trips_30d_std.tail(10).round(1))

# d) Visualization with moving averages
trips_7d_ma = daily_trips.rolling(window=7).mean()
trips_30d_ma = daily_trips.rolling(window=30).mean()

fig, ax = plt.subplots(figsize=(14, 6))
ax.plot(daily_trips.index, daily_trips.values, linewidth=0.5, alpha=0.5, color='gray', label='Daily')
ax.plot(trips_7d_ma.index, trips_7d_ma.values, linewidth=1.5, color='steelblue', label='7-day MA')
ax.plot(trips_30d_ma.index, trips_30d_ma.values, linewidth=2, color='coral', label='30-day MA')
ax.set_xlabel('Date')
ax.set_ylabel('Daily Trips')
ax.set_title('Daily Trips with Moving Averages', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Solution 5: Seasonal Analysis

```python
# =============================================================================
# SOLUTION 5: Seasonal Analysis
# =============================================================================

print("=== SOLUTION 5 ===")

# a) Average daily trips by month
daily_trips = df['trips'].resample('D').sum()
monthly_pattern = daily_trips.groupby(daily_trips.index.month).mean()
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly_pattern.index = month_names
print("\na) Average daily trips by month:")
print(monthly_pattern.round(0))

# b) Busiest and slowest months
busiest = monthly_pattern.idxmax()
slowest = monthly_pattern.idxmin()
print(f"\nb) Busiest month: {busiest} ({monthly_pattern[busiest]:,.0f} avg daily trips)")
print(f"   Slowest month: {slowest} ({monthly_pattern[slowest]:,.0f} avg daily trips)")

# c) Seasonal index
overall_avg = monthly_pattern.mean()
seasonal_index = (monthly_pattern / overall_avg).round(3)
print("\nc) Seasonal index (1.0 = average):")
print(seasonal_index)

# d) Q1 vs Q3 comparison
q1_data = daily_trips[daily_trips.index.quarter == 1]
q3_data = daily_trips[daily_trips.index.quarter == 3]
q1_avg = q1_data.mean()
q3_avg = q3_data.mean()
print(f"\nd) Q1 (Winter) avg daily trips: {q1_avg:,.0f}")
print(f"   Q3 (Summer) avg daily trips: {q3_avg:,.0f}")
print(f"   Summer is {((q3_avg - q1_avg) / q1_avg * 100):.1f}% higher than Winter")
```

### Solution 6: Weather Impact Analysis

```python
# =============================================================================
# SOLUTION 6: Weather Impact Analysis
# =============================================================================

print("=== SOLUTION 6 ===")

# a) Average trips: rainy vs non-rainy
weather_impact = df.groupby('is_raining')['trips'].mean()
no_rain_trips = weather_impact[False]
rain_trips = weather_impact[True]
print(f"\na) Average hourly trips:")
print(f"   No rain: {no_rain_trips:.1f}")
print(f"   Rain: {rain_trips:.1f}")

# b) Percentage decrease during rain
pct_decrease = (no_rain_trips - rain_trips) / no_rain_trips * 100
print(f"\nb) Rain reduces trips by {pct_decrease:.1f}%")

# c) Fare comparison
fare_by_weather = df.groupby('is_raining')['avg_fare'].mean()
print(f"\nc) Average fare:")
print(f"   No rain: ${fare_by_weather[False]:.2f}")
print(f"   Rain: ${fare_by_weather[True]:.2f}")
print(f"   Difference: {((fare_by_weather[True] - fare_by_weather[False]) / fare_by_weather[False] * 100):.1f}%")

# d) Hour with biggest rain impact
hourly_no_rain = df[~df['is_raining']].groupby('hour')['trips'].mean()
hourly_rain = df[df['is_raining']].groupby('hour')['trips'].mean()
hourly_impact = (hourly_no_rain - hourly_rain) / hourly_no_rain * 100
worst_hour = hourly_impact.idxmax()
print(f"\nd) Hour most affected by rain: {worst_hour}:00 ({hourly_impact[worst_hour]:.1f}% decrease)")
```

### Solution 7: Year-over-Year Comparison

```python
# =============================================================================
# SOLUTION 7: Year-over-Year Comparison
# =============================================================================

print("=== SOLUTION 7 ===")

# a) Total trips and revenue by year (Jan-Jun only)
# Filter to Jan-Jun for fair comparison
df_janjun_2023 = df.loc['2023-01':'2023-06']
df_janjun_2024 = df.loc['2024-01':'2024-06']

totals_2023 = df_janjun_2023[['trips', 'revenue']].sum()
totals_2024 = df_janjun_2024[['trips', 'revenue']].sum()

print("\na) Jan-Jun totals:")
print(f"   2023 - Trips: {totals_2023['trips']:,.0f}, Revenue: ${totals_2023['revenue']:,.2f}")
print(f"   2024 - Trips: {totals_2024['trips']:,.0f}, Revenue: ${totals_2024['revenue']:,.2f}")

# b) YoY growth rate
trips_growth = (totals_2024['trips'] - totals_2023['trips']) / totals_2023['trips'] * 100
revenue_growth = (totals_2024['revenue'] - totals_2023['revenue']) / totals_2023['revenue'] * 100
print(f"\nb) Year-over-Year growth:")
print(f"   Trips: +{trips_growth:.1f}%")
print(f"   Revenue: +{revenue_growth:.1f}%")

# c) Monthly comparison visualization
monthly_2023 = df_janjun_2023['trips'].resample('ME').sum()
monthly_2024 = df_janjun_2024['trips'].resample('ME').sum()

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
x = np.arange(len(months))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
bars1 = ax.bar(x - width/2, monthly_2023.values, width, label='2023', color='steelblue', alpha=0.7)
bars2 = ax.bar(x + width/2, monthly_2024.values, width, label='2024', color='coral', alpha=0.7)
ax.set_ylabel('Total Trips')
ax.set_title('Monthly Trips: 2023 vs 2024', fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(months)
ax.legend()
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()

# d) Month with highest improvement
monthly_growth = (monthly_2024.values - monthly_2023.values) / monthly_2023.values * 100
monthly_growth_series = pd.Series(monthly_growth, index=months)
best_improvement = monthly_growth_series.idxmax()
print(f"\nd) Month with highest improvement: {best_improvement} (+{monthly_growth_series[best_improvement]:.1f}%)")
```

### Solution 8: Driver Efficiency Analysis (Bonus)

```python
# =============================================================================
# SOLUTION 8: Driver Efficiency Analysis (BONUS)
# =============================================================================

print("=== SOLUTION 8 (BONUS) ===")

# a) Trips per driver
df_eff = df.copy()
df_eff['trips_per_driver'] = df_eff['trips'] / df_eff['active_drivers']
avg_efficiency = df_eff['trips_per_driver'].mean()
print(f"\na) Average trips per driver per hour: {avg_efficiency:.2f}")

# b) Hour with highest efficiency
hourly_efficiency = df_eff.groupby('hour')['trips_per_driver'].mean()
best_hour = hourly_efficiency.idxmax()
print(f"\nb) Most efficient hour: {best_hour}:00 ({hourly_efficiency[best_hour]:.2f} trips/driver)")

# c) Weekly trend in efficiency
weekly_trips = df_eff['trips'].resample('W').sum()
weekly_drivers = df_eff['active_drivers'].resample('W').sum()
weekly_efficiency = weekly_trips / weekly_drivers
print("\nc) Weekly efficiency trend (last 10 weeks):")
print(weekly_efficiency.tail(10).round(3))

# d) Is efficiency improving?
first_month_eff = df_eff.loc['2023-01']['trips_per_driver'].mean()
last_month_eff = df_eff.loc['2024-06']['trips_per_driver'].mean()
eff_change = (last_month_eff - first_month_eff) / first_month_eff * 100
print(f"\nd) Efficiency change over time:")
print(f"   Jan 2023: {first_month_eff:.2f} trips/driver/hour")
print(f"   Jun 2024: {last_month_eff:.2f} trips/driver/hour")
print(f"   Change: {eff_change:+.1f}%")
print(f"   Efficiency is {'improving' if eff_change > 0 else 'declining'}!")
```
