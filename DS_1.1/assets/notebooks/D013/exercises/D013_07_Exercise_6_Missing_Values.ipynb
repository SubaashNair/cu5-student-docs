The dataset you provided, `wind_data.csv`, corresponds to the **Ireland wind data from 1961 to 1978**, which contains **daily average wind speeds (in knots)** recorded at **12 synoptic meteorological stations** across the Republic of Ireland .

This dataset is well known in statistical literature and was used in the work of **Haslett and Raftery (1989)** for time series and spatial modeling of wind data . The columns after the date (`Yr`, `Mo`, `Dy`) represent the following 12 locations (stations), in order:

- **RPT** – Roche’s Point  
- **VAL** – Valentia Observatory  
- **ROS** – Rosslare  
- **KIL** – Kilkeel  
- **SHA** – Shannon Airport  
- **BIR** – Birr  
- **DUB** – Dublin Airport  
- **CLA** – Claremorris  
- **MUL** – Mullingar  
- **CLO** – Clones  
- **BEL** – Belmullet  
- **MAL** – Malin Head  

These are official meteorological stations operated by **Met Éireann**, Ireland’s national meteorological service .

The wind speeds are given in **knots**, and missing values (represented as `NaN` or blank in your file) indicate days where measurements were not recorded or were invalid.

This dataset is often used for:
- Time series analysis  
- Spatial statistics  
- Missing data imputation methods  
- Climate and meteorological studies

### Step 4. Year 2061? Do we really have data from this year? Create a function to fix it and apply it.

```python
# year=dt.year
```

```python
def fix_century(dt):
    if dt.year >= 2000:
        return dt.replace(year=dt.year-100)
    return dt
```

```python
df['Yr_Mo_Dy'] = df['Yr_Mo_Dy'].apply(fix_century)
```
