## Problem Description
You are managing a small store and need a system to track your inventory. 
Each item in your inventory has a **Name**, **Price**, and **Quantity**.

**Requirements:**
1.  **Data Structure**: Store the inventory as a **list of tuples** (e.g., `[("Apple", 0.5, 100), ("Milk", 2.5, 50)]`).
2.  **Functions**: Create the following functions:
    - `add_item(inventory, name, price, quantity)`: Adds a new item to the list.
    - `update_quantity(inventory, name, new_quantity)`: Updates the quantity of an existing item.
    - `view_inventory(inventory)`: Prints all items in a formatted table and calculates the **total value** of the inventory (Price * Quantity for all items).
3.  **Main Loop**: Create a menu-driven loop that allows the user to choose an action (Add, Update, View, Exit).


### Step 1: Initialize and Helper Functions
Define your list (start with some dummy data) and the functions.
- `add_item`: Appends a tuple to the list.
- `update_quantity`: Loops through the list to find the item and updates it. *Hint: Tuple are immutable, so you might need to replace the tuple or use a list of lists if you prefer mutable items, but try with tuples for this exercise.*
- `view_inventory`: Loops through the list, printing details and summing up the total value.

```python
# TODO: Write your functions here
def add_item(inventory, name, price, quantity):
    item = (name,price,quantity)
    inventory.append(item)
    print(f"{name} added.")
def update_qty(inventory,name,new_quantity):
    found = False
    for i,item in enumerate(inventory):
        if item[0] == name:
            # creating new tuple using old name and price but with new quantity value
            inventory[i] = [item[0],item[1],new_quantity]
            found = True
            break
    if not found:
        print(f"Item {name} not found")
```
