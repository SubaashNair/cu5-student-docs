# 5. Return Values

## Getting Results from Functions

So far, our functions have printed output. But often, you want a function to **return** a result that you can use later.

**Syntax:**
```python
def function_name():
    # Calculate something
    return result
```

````python
# Function that returns a value
def add(a, b):
    result = a + b
    return result

# Store the returned value
answer = add(5, 3)
print("5 + 3 =", answer)
````

````python
# You can use the return value directly
print("10 + 20 =", add(10, 20))
````

````python
# You can use returned values in calculations
total = add(5, 3) + add(2, 7)
print("(5+3) + (2+7) =", total)
````

### return vs print

| `print` | `return` |
|---------|----------|
| Displays output to screen | Gives value back to caller |
| Function returns `None` | Function returns a value |
| Can't use result in calculations | Can store and use result |

````python
# Difference between print and return
def add_print(a, b):
    print(a + b)  # Just prints, returns None

def add_return(a, b):
    return a + b  # Returns the value

result1 = add_print(5, 3)  # Prints 8
print("result1 is:", result1)  # None!

result2 = add_return(5, 3)
print("result2 is:", result2)  # 8
````

# 6. Default Parameters

## Optional Arguments

You can give parameters **default values**. If the caller doesn't provide a value, the default is used.

**Syntax:**
```python
def function_name(param1, param2=default_value):
    # param2 will use default_value if not provided
```

````python
def greet(name, greeting="Hello"):
    print(greeting + ",", name)

# Using default greeting
greet("Alice")

# Providing custom greeting
greet("Bob", "Good morning")
greet("Charlie", "Hi")
````

````python
# Multiple default parameters
def describe_pet(name, animal="dog", age=1):
    print(name, "is a", age, "year old", animal)

describe_pet("Max")  # Uses all defaults
describe_pet("Whiskers", "cat")  # Override animal
describe_pet("Buddy", "dog", 5)  # Override both
````

````python
# Using keyword arguments
# You can specify which parameter you're setting by name
describe_pet("Goldie", animal="fish")
describe_pet("Rex", age=3)  # animal uses default
````

# 8. Appendix: Exception Handling

## (Optional/Advanced)

Sometimes code can fail due to unexpected situations (like dividing by zero). **Exception handling** lets your program handle errors gracefully instead of crashing.

> **Note**: This is an advanced topic. You can skip this section and come back to it later.

### The Problem

````python
# This would crash the program!
# Uncomment to see the error:
# result = 10 / 0
````

### The Solution: try-except

````python
# Handle the error gracefully
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
    result = None

print("Program continues...")
print("Result:", result)
````

````python
# Safe division function
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None

print(safe_divide(10, 2))   # Works
print(safe_divide(10, 0))   # Handled gracefully
````

### Common Exceptions

| Exception | When it happens |
|-----------|----------------|
| `ZeroDivisionError` | Dividing by zero |
| `TypeError` | Wrong type (e.g., adding string + number) |
| `ValueError` | Wrong value (e.g., int("hello")) |
| `IndexError` | List index out of range |
| `KeyError` | Dictionary key not found |

````python
# Handling multiple error types
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Please provide numbers only!")
        return None

print(safe_divide(10, 2))        # Works
print(safe_divide(10, 0))        # ZeroDivisionError
print(safe_divide("10", 2))      # TypeError
````
