## Multiple Conditions

Use `&` for AND, `|` for OR. **Important:** Wrap each condition in parentheses!

````python
# Engineers earning more than $75,000
senior_engineers = df[(df['Department'] == 'Engineering') & (df['Salary'] > 75000)]
print(senior_engineers)
````

````python
# Sales OR Marketing
sales_marketing = df[(df['Department'] == 'Sales') | (df['Department'] == 'Marketing')]
print(sales_marketing)
````

# 7. GroupBy Operations

**GroupBy** follows the "split-apply-combine" pattern:
1. **Split** data into groups
2. **Apply** a function to each group
3. **Combine** results

````python
print(df)
````

````python
# Average salary by department
df.groupby('Department')['Salary'].mean()
````

````python
# Multiple aggregations
df.groupby('Department')['Salary'].agg(['mean', 'min', 'max', 'count'])
````

````python
# Multiple columns
df.groupby('Department').agg({
    'Salary': 'mean',
    'Age': 'mean',
    'Years': 'sum'
})
````

````python
# Size of each group
df.groupby('Department').size()
````
