# 1. Why Loops?

## The Problem

Without loops, repetitive tasks require repetitive code:

````python
# Without a loop - printing 1 to 5
print(1)
print(2)
print(3)
print(4)
print(5)

# Imagine doing this for 1 to 1000!
````

## The Solution: Loops!

````python
# With a loop - much better!
for number in [1, 2, 3, 4, 5]:
    print(number)
````

````python
# Even better - works for any range
for number in range(1, 6):
    print(number)
````

Python has two types of loops:
- **while loop**: Repeats while a condition is True
- **for loop**: Repeats for each item in a sequence

Let's learn both!

# 2. The while Loop

## Repeat While True

A `while` loop repeats as long as a condition is `True`.

**Syntax:**
```python
while condition:
    # This code repeats while condition is True
    do_something()
```

Think of it as: "Keep doing this **while** this condition is true."

### Simple Counter Example

````python
# Count from 1 to 5
count = 1

while count <= 5:
    print(count)
    count = count + 1  # Don't forget to update the counter!

print("Done counting!")
````

**What happened?**
1. `count` starts at 1
2. Is `count <= 5`? Yes (1 <= 5), so print 1, then count becomes 2
3. Is `count <= 5`? Yes (2 <= 5), so print 2, then count becomes 3
4. ... continues until...
5. Is `count <= 5`? No (6 <= 5 is False), so stop

### Flowchart: while Loop

```
        ┌─────────────┐
        │   Start     │
        └──────┬──────┘
               │
               ▼
        ┌─────────────┐
   ┌───▶│ Condition   │───── False ─────┐
   │    │   True?     │                 │
   │    └──────┬──────┘                 │
   │           │ True                   │
   │           ▼                        │
   │    ┌─────────────┐                 │
   │    │ Run code    │                 │
   │    │ inside loop │                 │
   │    └──────┬──────┘                 │
   │           │                        │
   └───────────┘                        │
                                        ▼
                                 ┌─────────────┐
                                 │  Continue   │
                                 │  program    │
                                 └─────────────┘
```

### Countdown Example

````python
# Countdown from 5 to 1
countdown = 5

while countdown > 0:
    print(countdown)
    countdown = countdown - 1

print("Blast off!")
````

### Adding Numbers Example

````python
# Add numbers from 1 to 10
total = 0
number = 1

while number <= 10:
    total = total + number
    number = number + 1

print("Sum of 1 to 10 is:", total)
````

### DANGER: Infinite Loops!

If the condition never becomes `False`, the loop runs forever! This is called an **infinite loop**.

**Common mistake:** Forgetting to update the counter.

````python
# DON'T RUN THIS - it would run forever!
# count = 1
# while count <= 5:
#     print(count)
#     # Oops! Forgot: count = count + 1
#     # count is always 1, so count <= 5 is always True

print("The code above would run forever if uncommented!")
````

# 3. The for Loop

## Repeat for Each Item

A `for` loop repeats once for each item in a sequence (like a list or string).

**Syntax:**
```python
for item in sequence:
    # This code runs for each item
    do_something(item)
```

Think of it as: "**For each** item in this collection, do something."

### Looping Through a List

````python
# Print each fruit in the list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

````python
# You can use any variable name
fruits = ["apple", "banana", "cherry"]

for x in fruits:
    print(x)
    
# But descriptive names are better!
````

````python
# Calculate total price
prices = [10, 20, 30, 40]
total = 0

for price in prices:
    total = total + price

print("Total:", total)
````

### Looping Through a String

````python
# Print each character in a string
word = "Python"

for letter in word:
    print(letter)
````

````python
# Count vowels in a word
word = "programming"
vowel_count = 0

for letter in word:
    if letter in "aeiou":
        vowel_count = vowel_count + 1

print("Number of vowels:", vowel_count)
````

# 4. The range() Function

## Generate Number Sequences

The `range()` function creates a sequence of numbers - perfect for loops!

**Three ways to use range:**

| Syntax | Meaning | Example |
|--------|---------|--------|
| `range(stop)` | 0 to stop-1 | `range(5)` → 0,1,2,3,4 |
| `range(start, stop)` | start to stop-1 | `range(1,5)` → 1,2,3,4 |
| `range(start, stop, step)` | start to stop-1 by step | `range(0,10,2)` → 0,2,4,6,8 |

### range(stop)

````python
# Print 0 to 4 (5 numbers total)
for i in range(5):
    print(i)
````

### range(start, stop)

````python
# Print 1 to 5
for i in range(1, 6):  # Note: 6, not 5!
    print(i)
````

````python
# Print 10 to 15
for i in range(10, 16):
    print(i)
````

### range(start, stop, step)

````python
# Even numbers from 0 to 10
for i in range(0, 11, 2):
    print(i)
````

````python
# Count by 5s
for i in range(0, 26, 5):
    print(i)
````

````python
# Countdown using negative step
for i in range(5, 0, -1):
    print(i)
print("Blast off!")
````

### Practical Example: Sum of Numbers

````python
# Sum of 1 to 100
total = 0

for number in range(1, 101):
    total = total + number

print("Sum of 1 to 100:", total)
````

# 5. break and continue

## Control Loop Flow

Sometimes you need more control over your loops:
- **break**: Exit the loop immediately
- **continue**: Skip to the next iteration

### The break Statement

`break` exits the loop completely, even if the condition is still True.

````python
# Stop when we find "banana"
fruits = ["apple", "banana", "cherry", "date"]

for fruit in fruits:
    if fruit == "banana":
        print("Found banana! Stopping.")
        break
    print(fruit)
````

````python
# Find the first number divisible by 7
for number in range(1, 100):
    if number % 7 == 0:
        print("First number divisible by 7:", number)
        break
````

### The continue Statement

`continue` skips the rest of the current iteration and moves to the next one.

````python
# Skip banana, print everything else
fruits = ["apple", "banana", "cherry", "date"]

for fruit in fruits:
    if fruit == "banana":
        continue  # Skip banana
    print(fruit)
````

````python
# Print only odd numbers
for number in range(1, 11):
    if number % 2 == 0:
        continue  # Skip even numbers
    print(number)
````

### Comparison: break vs continue

```
break:     [1] → [2] → [STOP]  (exit loop)
continue:  [1] → [SKIP] → [3] → [4]  (skip one, continue)
```
