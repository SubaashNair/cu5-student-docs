# Module 07a: Pandas Fundamentals

## Why Pandas?

Imagine you have a spreadsheet with 10,000 rows of customer data. You need to:
- Find all customers from a specific city
- Calculate the average purchase amount
- Group sales by month

With regular Python lists and dictionaries, this would require many lines of complex code. **Pandas** makes these tasks simple and fast!

Pandas is:
- Built on top of NumPy (so it's fast)
- Designed for **tabular data** (like spreadsheets or SQL tables)
- The most important tool for data analysis in Python

# 3. Reading Data

In real-world data science, you'll usually load data from files rather than typing it in.

## Reading CSV Files

CSV (Comma-Separated Values) is the most common format for tabular data.

````python
# Read a CSV file
# df = pd.read_csv('filename.csv')

# Read from a URL
url = 'https://raw.githubusercontent.com/datasets/population/master/data/population.csv'
df_population = pd.read_csv(url)

print("Shape:", df_population.shape)
print("\nFirst 5 rows:")
print(df_population.head())
````

### Common read_csv() Parameters

| Parameter | Description | Example |
|-----------|-------------|--------|
| `sep` | Delimiter | `sep=';'` for semicolon |
| `header` | Row number for column names | `header=0` (default) |
| `index_col` | Column to use as index | `index_col='id'` |
| `usecols` | Columns to read | `usecols=['name', 'age']` |
| `nrows` | Number of rows to read | `nrows=100` |

````python
# Read only specific columns and first 100 rows
df_small = pd.read_csv(url, 
                       usecols=['Country Name', 'Year', 'Value'],
                       nrows=100)
print(df_small.head())
````

# 4. Writing Data

Save your DataFrames to various formats.

````python
# Sample DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 90, 78]
})

# Save to CSV
df.to_csv('output.csv', index=False)  # index=False prevents saving row numbers
print("Saved to output.csv")

# Save to JSON
df.to_json('output.json', orient='records', indent=2)
print("Saved to output.json")

# Verify by reading back
print("\nRead back from CSV:")
print(pd.read_csv('output.csv'))
````
