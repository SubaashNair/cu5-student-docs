### Bonus (after D004) - Exercise 6.3: Sentence Statistics
For the sentence below, calculate and display:
1. Total characters (including spaces)
2. Number of words
3. Number of vowels (a, e, i, o, u)
4. Average word length

````python
sentence = "Python is a powerful programming language for data science"
# Your code here
vowels = "aeiouAEIOU"

# counter
num_vowels = 0
# checking for each alphabet in the string
for char in sentence:
    # checking the alphabet with the vowels 
    # returns True if matches
    if char in vowels:
        # if true increase the counter
        num_vowels +=1
        
print(num_vowels)
````
