# 3. String Indexing

Each character in a string has a position called an **index**. Python uses **zero-based indexing** - the first character is at index 0.

```
String:   H  e  l  l  o
Index:    0  1  2  3  4
Negative: -5 -4 -3 -2 -1
```

````python
word = "Hello"

# Positive indexing (from the start)
print("First character:", word[0])   # H
print("Second character:", word[1])  # e
print("Last character:", word[4])    # o
````

````python
# Negative indexing (from the end)
print("Last character:", word[-1])    # o
print("Second to last:", word[-2])    # l
print("First character:", word[-5])   # H
````

# 4. String Slicing

**Slicing** lets you extract a portion of a string.

**Syntax:** `string[start:end]` - gets characters from `start` up to (but not including) `end`

````python
text = "Hello, World!"

# Basic slicing
print(text[0:5])   # Hello (characters 0, 1, 2, 3, 4)
print(text[7:12])  # World (characters 7, 8, 9, 10, 11)
````

## Omitting Start or End

````python
text = "Hello, World!"

# Omit start - defaults to 0
print(text[:5])    # Hello (same as text[0:5])

# Omit end - goes to the end
print(text[7:])    # World!

# Omit both - copies the whole string
print(text[:])     # Hello, World!
````

## Negative Slicing

````python
text = "Hello, World!"

print(text[-6:])    # World! (last 6 characters)
print(text[:-1])    # Hello, World (everything except last character)
print(text[-6:-1])  # World (from -6 up to but not including -1)
````

## Step (Stride)

**Syntax:** `string[start:end:step]` - gets every `step`th character

````python
text = "Hello, World!"

print(text[::2])    # Hlo ol! (every 2nd character)
print(text[::3])    # Hl r! (every 3rd character)
print(text[::-1])   # !dlroW ,olleH (reversed!)
````

### Quick Reference

| Slice | Description | Example with "Hello" |
|-------|-------------|---------------------|
| `s[0]` | First character | `H` |
| `s[-1]` | Last character | `o` |
| `s[1:4]` | Characters 1 to 3 | `ell` |
| `s[:3]` | First 3 characters | `Hel` |
| `s[2:]` | From character 2 to end | `llo` |
| `s[:-2]` | All except last 2 | `Hel` |
| `s[::-1]` | Reversed | `olleH` |

# 5. String Methods

Strings have many built-in **methods** (functions that belong to strings). Call them using dot notation: `string.method()`

## Case Conversion

````python
text = "Hello, World!"

print(text.upper())      # HELLO, WORLD!
print(text.lower())      # hello, world!
print(text.title())      # Hello, World!
print(text.capitalize()) # Hello, world! (only first letter)
````

## Finding and Counting

````python
text = "Hello, World! Hello, Python!"

# Find position of substring (returns -1 if not found)
print(text.find("World"))    # 7 (index where "World" starts)
print(text.find("Java"))     # -1 (not found)

# Count occurrences
print(text.count("Hello"))   # 2
print(text.count("o"))       # 4
````

## Checking String Content

````python
# Checking start and end
filename = "report.pdf"

print(filename.startswith("report"))  # True
print(filename.endswith(".pdf"))      # True
print(filename.endswith(".txt"))      # False
````

````python
# Checking character types
print("123".isdigit())      # True - all digits
print("abc".isalpha())      # True - all letters
print("abc123".isalnum())   # True - all letters or digits
print("   ".isspace())      # True - all whitespace
````

### Common String Methods Reference

| Method | Description | Example |
|--------|-------------|--------|
| `upper()` | Convert to uppercase | `"hi".upper()` → `"HI"` |

| `count(sub)` | Count occurrences | `"hello".count("l")` → `2` |
| `startswith(s)` | Check if starts with | `"hello".startswith("he")` → `True` |
| `endswith(s)` | Check if ends with | `"hello".endswith("lo")` → `True` |

# 6. String Formatting

There are several ways to insert variables into strings.

## Method 1: Concatenation (Old Way)

````python
name = "Alice"
age = 25

# Concatenation - requires str() for numbers
message = "My name is " + name + " and I am " + str(age) + " years old."
print(message)
````

## Method 2: % Formatting (Old Way)

````python
name = "Alice"
age = 25

# %s for strings, %d for integers, %f for floats
message = "My name is %s and I am %d years old." % (name, age)
print(message)

# Control decimal places with %.2f
price = 19.99
print("Price: $%.2f" % price)
````

## Method 3: .format() Method

````python
name = "Alice"
age = 25

# Use {} as placeholders
message = "My name is {} and I am {} years old.".format(name, age)
print(message)

# With position numbers
message = "My name is {0} and I am {1}. {0} loves Python!".format(name, age)
print(message)
````

## Method 4: f-Strings (Best Method!)

f-Strings (formatted string literals) are the modern, preferred way. Add `f` before the string and put variables in `{}`.

````python
name = "Alice"
age = 25

# f-string - cleanest syntax!
message = f"My name is {name} and I am {age} years old."
print(message)
````

````python
# You can put expressions inside {}
a = 5
b = 3
print(f"{a} + {b} = {a + b}")
print(f"{a} * {b} = {a * b}")
````

````python
# Formatting numbers
price = 49.99
quantity = 3
total = price * quantity

print(f"Price: ${price:.2f}")           # 2 decimal places
print(f"Total: ${total:.2f}")           # 2 decimal places
print(f"Quantity: {quantity:03d}")       # Pad with zeros
````
