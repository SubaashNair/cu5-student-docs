# Dictionaries

A **dictionary** is an unordered collection of key-value pairs. Think of it like a real dictionary where you look up a word (the key) to find its definition (the value).

Dictionaries are:
- **Mutable**: You can add, modify, and remove items
- **Unordered**: Items don't have a fixed position (though Python 3.7+ maintains insertion order)
- **Indexed by keys**: You access values using their keys, not numeric indices
- **Fast**: Looking up values by key is very efficient

## Creating Dictionaries

Dictionaries are created using curly braces `{}` with key-value pairs separated by colons `:`.

```python
# Creating an empty dictionary
empty_dict = {}
print(type(empty_dict))

# Creating a dictionary with initial values
student = {
    'name': 'Alice',
    'age': 25,
    'course': 'Data Science',
    'gpa': 3.8
}
print(student)
```

```python
# Alternative: Using the dict() constructor
person = dict(name='Bob', age=30, city='Kuala Lumpur')
print(person)

# From a list of tuples
pairs = [('a', 1), ('b', 2), ('c', 3)]
letter_numbers = dict(pairs)
print(letter_numbers)
```

## Accessing Dictionary Values

You access values using their keys in square brackets `[]` or the `.get()` method.

```python
student = {'name': 'Alice', 'age': 25, 'course': 'Data Science'}

# Using square brackets
print(student['name'])
print(student['age'])

# Using .get() method - safer, returns None if key doesn't exist
print(student.get('course'))
print(student.get('grade'))  # Returns None, no error
print(student.get('grade', 'Not assigned'))  # Returns default value
```

## Modifying Dictionaries

Dictionaries are mutable - you can add, update, and remove items.

```python
student = {'name': 'Alice', 'age': 25}

# Adding a new key-value pair
student['course'] = 'Data Science'
print(student)

# Updating an existing value
student['age'] = 26
print(student)

# Adding multiple items with .update()
student.update({'gpa': 3.8, 'semester': 3})
print(student)
```

```python
# Removing items
student = {'name': 'Alice', 'age': 25, 'course': 'Data Science', 'gpa': 3.8}

# Using del
del student['gpa']
print(student)

# Using .pop() - returns the removed value
removed_value = student.pop('age')

# Note: f-strings (f"...") let you insert variables directly into strings
# The variable goes inside curly braces: f"Value is {variable}"
print(f"Removed: {removed_value}")
print(student)

# Using .clear() to remove all items
student.clear()
print(student)  # Empty dictionary
```

## Dictionary Methods

Here are the most commonly used dictionary methods:

| Method | Description |
|--------|-------------|
| `dict.keys()` | Returns all keys |
| `dict.values()` | Returns all values |
| `dict.items()` | Returns all key-value pairs as tuples |
| `dict.get(key, default)` | Returns value for key, or default if not found |
| `dict.pop(key)` | Removes and returns value for key |
| `dict.update(other)` | Updates dict with key-value pairs from other |
| `dict.clear()` | Removes all items |

```python
student = {'name': 'Alice', 'age': 25, 'course': 'Data Science'}

# Getting keys, values, and items
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))
```

## Iterating Over Dictionaries

You can loop through dictionaries in several ways.

```python
student = {'name': 'Alice', 'age': 25, 'course': 'Data Science'}

# Iterate over keys (default)
print("Keys:")
for key in student:
    print(f"  {key}")

# Iterate over values
print("\nValues:")
for value in student.values():
    print(f"  {value}")

# Iterate over key-value pairs
print("\nKey-Value Pairs:")
for key, value in student.items():
    print(f"  {key}: {value}")
```

## Checking for Keys

Use the `in` operator to check if a key exists in a dictionary.

```python
student = {'name': 'Alice', 'age': 25, 'course': 'Data Science'}

# Check if key exists
print('name' in student)     # True
print('grade' in student)    # False
print('Alice' in student)    # False - checks keys, not values!

# Check if value exists
print('Alice' in student.values())  # True
```

## Nested Dictionaries

Dictionaries can contain other dictionaries as values, creating nested structures. This is very common when working with JSON data.

```python
# A nested dictionary representing students in a class
classroom = {
    'student1': {
        'name': 'Alice',
        'age': 25,
        'grades': {'math': 90, 'science': 85, 'english': 88}
    },
    'student2': {
        'name': 'Bob',
        'age': 23,
        'grades': {'math': 78, 'science': 92, 'english': 80}
    }
}

# Accessing nested values
print(classroom['student1']['name'])
print(classroom['student1']['grades']['math'])

# Modifying nested values
classroom['student2']['grades']['math'] = 82
print(classroom['student2']['grades'])
```

## Dictionary Comprehensions

Similar to list comprehensions, you can create dictionaries using comprehension syntax.

```python
# Create a dictionary of squares
squares = {x: x**2 for x in range(1, 6)}
print(squares)

# Create a dictionary from two lists
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
name_age = {name: age for name, age in zip(names, ages)}
print(name_age)

# Filter with conditions
numbers = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
even_numbers = {k: v for k, v in numbers.items() if v % 2 == 0}
print(even_numbers)
```

# Sets

A **set** is an unordered collection of **unique** elements. Sets are useful when you need to:
- Remove duplicates from a collection
- Test membership efficiently
- Perform mathematical set operations (union, intersection, etc.)

Sets are:
- **Mutable**: You can add and remove elements
- **Unordered**: Elements have no position
- **Unique**: No duplicate elements allowed
- **Unhashable elements not allowed**: Sets can only contain immutable types (no lists or dictionaries)

## Creating Sets

```python
# Creating a set with curly braces
fruits = {'apple', 'banana', 'cherry'}
print(fruits)
print(type(fruits))

# Note: {} creates an empty dict, not a set!
empty_set = set()  # Use set() for empty set
print(type(empty_set))

# Creating a set from a list (removes duplicates!)
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
unique_numbers = set(numbers)
print(unique_numbers)
```

```python
# Sets automatically remove duplicates
duplicates = {'apple', 'banana', 'apple', 'cherry', 'banana'}
print(duplicates)  # Only unique values kept
```

## Modifying Sets

```python
fruits = {'apple', 'banana', 'cherry'}

# Adding elements
fruits.add('orange')
print(fruits)

# Adding multiple elements
fruits.update(['mango', 'grape'])
print(fruits)

# Removing elements
fruits.remove('banana')  # Raises error if not found
print(fruits)

fruits.discard('kiwi')  # No error if not found
print(fruits)

# Pop removes and returns an arbitrary element
popped = fruits.pop()
print(f"Popped: {popped}")
print(fruits)
```

## Set Operations

Sets support mathematical set operations:

| Operation | Method | Operator | Description |
|-----------|--------|----------|-------------|
| Union | `a.union(b)` | `a \| b` | All elements from both sets |
| Intersection | `a.intersection(b)` | `a & b` | Elements common to both sets |
| Difference | `a.difference(b)` | `a - b` | Elements in a but not in b |
| Symmetric Difference | `a.symmetric_difference(b)` | `a ^ b` | Elements in either set, but not both |

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# Union - all elements from both sets
print("Union:", set_a | set_b)
print("Union:", set_a.union(set_b))

# Intersection - elements in both sets
print("Intersection:", set_a & set_b)
print("Intersection:", set_a.intersection(set_b))

# Difference - elements in a but not in b
print("Difference (a-b):", set_a - set_b)
print("Difference (b-a):", set_b - set_a)

# Symmetric difference - elements in either but not both
print("Symmetric Difference:", set_a ^ set_b)
```

## Set Comparisons

```python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}
set_c = {1, 2, 3}

# Subset - all elements of a are in b
print(f"Is {set_a} a subset of {set_b}?", set_a.issubset(set_b))  # True
print(f"Is {set_a} a subset of {set_b}?", set_a <= set_b)  # True

# Superset - b contains all elements of a
print(f"Is {set_b} a superset of {set_a}?", set_b.issuperset(set_a))  # True

# Equality
print(f"Is {set_a} equal to {set_c}?", set_a == set_c)  # True

# Disjoint - no common elements
set_d = {6, 7, 8}
print(f"Are {set_a} and {set_d} disjoint?", set_a.isdisjoint(set_d))  # True
```

## Practical Use Cases for Sets

```python
# Use Case 1: Removing duplicates from a list
names = ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'David']
unique_names = list(set(names))
print("Unique names:", unique_names)
```

```python
# Use Case 2: Finding common elements
python_students = {'Alice', 'Bob', 'Charlie', 'David'}
java_students = {'Charlie', 'David', 'Eve', 'Frank'}

# Students taking both courses
both = python_students & java_students
print("Taking both courses:", both)

# Students taking only Python
only_python = python_students - java_students
print("Only Python:", only_python)

# All students
all_students = python_students | java_students
print("All students:", all_students)
```

```python
# Use Case 3: Fast membership testing
# Sets are much faster than lists for checking membership

valid_codes = {'A001', 'A002', 'B001', 'B002', 'C001'}

user_code = 'B001'
if user_code in valid_codes:
    print(f"{user_code} is valid!")
else:
    print(f"{user_code} is not valid!")
```

# Comparing Data Structures

Let's summarize when to use each data structure:

| Data Structure | Ordered? | Mutable? | Duplicates? | Use Case |
|----------------|----------|----------|-------------|----------|
| **List** | Yes | Yes | Yes | General-purpose ordered collection |
| **Tuple** | Yes | No | Yes | Fixed data, dictionary keys |
| **Dictionary** | No* | Yes | Keys: No, Values: Yes | Key-value lookups, JSON data |
| **Set** | No | Yes | No | Unique elements, set operations |

*Python 3.7+ maintains insertion order for dictionaries

```python
# Quick comparison
my_list = [1, 2, 3, 3]      # Ordered, allows duplicates
my_tuple = (1, 2, 3, 3)     # Ordered, allows duplicates, immutable
my_dict = {'a': 1, 'b': 2}  # Key-value pairs
my_set = {1, 2, 3, 3}       # Unique elements only

print("List:", my_list)
print("Tuple:", my_tuple)
print("Dict:", my_dict)
print("Set:", my_set)  # Notice: only one 3!
```
