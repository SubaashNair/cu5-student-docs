## Problem Description
You have been hired to build a basic Course Enrollment System for a university. The system needs to manage two main types of data: **Courses** and **Students**.

### Data Structures
1.  **Courses Database**: A dictionary where the **Keys** are Course Codes (e.g., "CS101") and the **Values** are dictionaries containing:
    - `title` (String)
    - `credits` (Integer)
    - `prerequisites` (Set of Course Codes)
    
    *Example*: 
    ```python
    courses = {
        "CS101": {"title": "Intro to CS", "credits": 3, "prerequisites": set()},
        "CS102": {"title": "Data Structures", "credits": 4, "prerequisites": {"CS101"}}
    }
    ```

2.  **Students Database**: A dictionary where the **Keys** are Student IDs (e.g., "S001") and the **Values** are dictionaries containing:
    - `name` (String)
    - `enrolled_courses` (Set of Course Codes)

### Required Features
Your program should provide a menu with the following options:
1.  **Add Course**: Allow the admin to add a new course. Ask for Code, Title, Credits, and Prerequisites (comma-separated). **Validation**: Check if the course code already exists.
2.  **Register Student**: Register a new student. Ask for Name and generate/ask for an ID. **Validation**: Check if ID exists.
3.  **Enroll Student**: Enroll a student in a specific course. 
    - **Checks required**:
        - Does the Student ID exist?
        - Does the Course Code exist?
        - Is the student *already* enrolled in this course?
        - Has the student completed all **prerequisites**? (Assume a student has "completed" a prerequisite if it is in their `enrolled_courses` set for this simple simulation, or you can add a `completed_courses` set if you want to be extra fancy. *For this exercise, checking `enrolled_courses` is sufficient*).
4.  **Drop Course**: Remove a course from a student's enrollment list.
5.  **Student Report**: Print a list of all students, their enrolled courses, and total credits they are taking.

### Step 2: Helper Functions
Create functions for each main feature. This makes your code cleaner and easier to debug.
- `add_course(course_code, title, credits, prerequisites)`
- `register_student(student_id, name)`
- `enroll_student(student_id, course_code)` -> *This is the logic-heavy one!*
- `drop_course(student_id, course_code)`

````python
courses = {
    "CS101": {"title": "Intro to Programming", "credits": 3, "prerequisites": set()},
    "CS102": {"title": "Data Structures", "credits": 4, "prerequisites": {"CS101"}},
    "MATH101": {"title": "Calculus I", "credits": 4, "prerequisites": set()},
    "CS201": {"title": "Algorithms", "credits": 3, "prerequisites": {"CS102", "MATH101"}}
}

students = {
    "S001": {"name": "Alice", "enrolled_courses": {"CS101"}},
    "S002": {"name": "Bob", "enrolled_courses": set()}
}
````

````python
# TODO: Write your functions here
def add_course():
    # todo
    code = input('Enter your course code: ').upper()
    if code in courses:
        print('Course is available')
        return
    title = input('Enter course title')
    credits = int(input('Enter credits'))
    return

    prereq_input = input('Enter your prerequisites: (comma seperated and empty if none)')
    if prereq_input.strip():
        prereqs = {p.strip().upper() for p in prereq_input.split(',')}
    else:
        prereqs = set()

    courses[code] = {
        'title':title,
        'credits':credits,
        'prerequisites':prereqs
    }
    print(f'Course {code} added!')

def register_student():
    sid = input('Enter student ID: ').upper()
    if sid in students:
        print('Student already exist')
        return
    name = input('Enter student name: ')
    students[sid] = {
        'name':name,
        'enrolled_courses':set(),
    }
    print(f"Student {name} ({sid}) registered")

def enrolled_student():
    sid = input('Enter student ID').upper()
    if sid not in students:
        print('Student not found')
        return
    code  = input('Enter course code to enroll').upper()
    if code not in courses:
        print('Couse not found')
        return

    # getting the refernece of students enrolled in courses set
    student_courses = students[sid]['enrolled_courses']
    print(student_courses)
    if code in student_courses:
        print('Student already enrolled in course')
        return

    required_pre = courses[code]['prerequisites']
    print(required_pre)
    
    if required_pre.issubset(student_courses):
        # {CS101}.issusbet({CS101,MATH101}) true
       # {CS101,MATH101}.issubset( {CS101}) false
        student_courses.add(code)
    else:
        missing = required_pre - student_courses
        print(f'Enrollment failed: Missing prerequisites: {missing}')

def drop_course():
    sid = input('Enter student ID').upper()
    if sid not in students:
        print('Error student not found')
        return
    code = input('Enter course to drop').upper()
    student_courses = students[sid]['enrolled_courses']
    if code in student_courses:
        student_courses.remove(code)
        print(f'dropped {code} for student {sid}')
    else:
        print('student not enrolled in this course')

def generate_report():
    print(f"{'ID':<10} {'Name':<15} {'Credits':<8} {'Courses'}")
    print('-'*50)
    for sid,data in students.items():
        name = data['name']
        enrolled = data['enrolled_courses']

        total_credits = 0
        course_list = []
        for c in enrolled:
            total_credits+=courses[c]['credits']
            course_list.append(c)
        course_s = ', '.join(sorted(course_list))
        print(f"{sid:<10} {name:<15} {total_credits:<8} {course_s}")
    print('\n')
````
