# Part 0: Setup and Data Generation

Run this cell to import NumPy and generate the dataset.

````python
import numpy as np

# Set seed for reproducibility - DO NOT CHANGE
np.random.seed(42)

# City information
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
num_cities = len(cities)
num_days = 30

# Generate temperature data (Celsius)
# Each city has a base temperature + daily variation + trend
base_temps = np.array([15, 25, 10, 28, 35])  # Base temperatures for each city
daily_variation = np.random.randn(num_cities, num_days) * 5  # Random daily changes
seasonal_trend = np.sin(np.linspace(0, np.pi, num_days)) * 3  # Slight warming trend

# Temperature matrix: rows = cities, columns = days
temperatures = base_temps.reshape(-1, 1) + daily_variation + seasonal_trend
temperatures = np.round(temperatures, 1)

# Generate precipitation data (mm)
precipitation = np.random.exponential(scale=5, size=(num_cities, num_days))
precipitation = np.round(precipitation, 1)
# Some days have no rain
precipitation[precipitation < 2] = 0

# Generate wind speed data (km/h)
wind_speed = np.random.uniform(5, 40, size=(num_cities, num_days))
wind_speed = np.round(wind_speed, 1)

# Day labels
days = np.arange(1, num_days + 1)

print("Dataset generated successfully!")
print(f"\nCities: {cities}")
print(f"Number of days: {num_days}")
print(f"\nTemperature matrix shape: {temperatures.shape}")
print(f"Precipitation matrix shape: {precipitation.shape}")
print(f"Wind speed matrix shape: {wind_speed.shape}")
````
