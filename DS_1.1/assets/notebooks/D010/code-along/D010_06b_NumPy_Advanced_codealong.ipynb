## Broadcasting Rules

NumPy compares shapes element-by-element, starting from the **trailing** (rightmost) dimensions:

1. **Rule 1**: If arrays have different numbers of dimensions, pad the smaller shape with 1s on the left
2. **Rule 2**: Arrays with size 1 in a dimension are "stretched" to match the other array
3. **Rule 3**: If sizes don't match and neither is 1, broadcasting fails

| Array 1 Shape | Array 2 Shape | Result Shape | Works? |
|--------------|---------------|--------------|--------|
| (3, 4) | (4,) | (3, 4) | Yes |
| (3, 4) | (3, 1) | (3, 4) | Yes |
| (3, 4) | (3,) | Error | No |
| (5, 3, 4) | (3, 4) | (5, 3, 4) | Yes |
| (5, 3, 4) | (1, 4) | (5, 3, 4) | Yes |

## Other Linear Algebra Operations

````python
A = np.array([[1, 2],
              [3, 4]])

# Transpose
print("Transpose:")
print(A.T)

# Determinant
print("\nDeterminant:", np.linalg.det(A))

# Inverse
print("\nInverse:")
print(np.linalg.inv(A))

# Verify: A @ A_inv = Identity
print("\nA @ A_inv (should be identity):")
print(np.round(A @ np.linalg.inv(A), 10))
````

````python
# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
````

## Solving Linear Equations

Solve systems of linear equations: Ax = b

````python
# System of equations:
# 2x + 3y = 8
# 4x + 1y = 10

A = np.array([[2, 3],
              [4, 1]])
b = np.array([8, 10])

# Solve for x
x = np.linalg.solve(A, b)
print("Solution (x, y):", x)

# Verify
print("Verification A @ x:", A @ x)
````
