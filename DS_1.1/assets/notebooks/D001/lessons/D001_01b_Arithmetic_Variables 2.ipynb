# 1. Arithmetic Operations

Python can work as a powerful calculator! Here are the basic arithmetic operators:

| Symbol | Operation | Example | Result |
|--------|-----------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division | `5 / 2` | `2.5` |
| `//` | Floor Division | `5 // 2` | `2` |
| `%` | Modulo (remainder) | `5 % 2` | `1` |
| `**` | Exponentiation | `5 ** 2` | `25` |

## Basic Operations

```python
5 + 3, 5+3
```

```python
# Addition
print("5 + 3 =", 5 + 3)
```

```python
# Subtraction
print("10 - 4 =", 10 - 4)
```

```python
# Multiplication
print("6 * 7 =", 6 * 7)
```

```python
# Division (always returns a decimal)
print("15 / 4 =", 15 / 4)
```

## Floor Division and Modulo

**Floor division** (`//`) divides and rounds down to the nearest whole number.

**Modulo** (`%`) gives the remainder after division.

```python
# Floor division - rounds down
print("15 // 4 =", 15 // 4)  # 15 divided by 4 is 3.75, rounded down = 3
```

```python
# Modulo - gives the remainder
print("15 % 4 =", 15 % 4)  # 15 divided by 4 is 3 remainder 3
```

**Real-world use of modulo:**
- Check if a number is even: `number % 2 == 0`
- Convert minutes to hours and minutes
- Cycle through values (like days of the week)

```python
# Example: Convert 135 minutes to hours and minutes
total_minutes = 135
hours = total_minutes // 60
minutes = total_minutes % 60
print(f"{total_minutes} minutes = {hours} hours and {minutes} minutes")
```

```python
type(f"{total_minutes}")a
```

## Exponentiation (Powers)

```python
# ** raises a number to a power
print("2 ** 3 =", 2 ** 3)   # 2 to the power of 3 = 2 * 2 * 2 = 8
print("5 ** 2 =", 5 ** 2)   # 5 squared = 25
print("9 ** 0.5 =", 9 ** 0.5)  # Square root of 9 = 3
```

## Order of Operations (PEMDAS)

Python follows standard math order of operations:

1. **P**arentheses
2. **E**xponents
3. **M**ultiplication and **D**ivision (left to right)
4. **A**ddition and **S**ubtraction (left to right)

```python
# Without parentheses - multiplication happens first
print("2 + 3 * 4 =", 2 + 3 * 4)  # 3*4=12, then 2+12=14
```

```python
# With parentheses - addition happens first
print("(2 + 3) * 4 =", (2 + 3) * 4)  # 2+3=5, then 5*4=20
```

# 2. Numbers: Integers vs Floats

Python has two main types of numbers:

| Type | Description | Examples |
|------|-------------|----------|
| `int` | Whole numbers (integers) | `1`, `42`, `-7`, `0` |
| `float` | Decimal numbers | `3.14`, `2.0`, `-0.5` |

## The type() Function

Use `type()` to check what kind of data you have:

```python
# Check the type of different values
print(type(42))      # int - whole number
print(type(3.14))    # float - decimal number
print(type("hello")) # str - string (text)
```

## Integer vs Float Division

**Important:** Division (`/`) always returns a float, even if the result is a whole number!

```python
# Division always returns a float
print(10 / 2)       # Result: 5.0 (float, not int)
print(type(10 / 2)) # float
```

```python
# Floor division returns an int (when both operands are int)
print(10 // 2)       # Result: 5 (int)
print(type(10 // 2)) # int
```

## Mixing Integers and Floats

When you mix int and float in an operation, the result is always a float:

```python
print(5 + 2.0)      # 7.0 (float)
print(3 * 1.5)      # 4.5 (float)
print(type(5 + 2.0)) # float
```

A **variable** is a name that refers to a value. Think of it like a labeled box where you can store data.

## Using Variables in Calculations

```python
# Calculate the area of a rectangle
width = 10
height = 5
area = width * height

print("Width:", width)
print("Height:", height)
print("Area:", area)
```

## Variables Can Change

That's why they're called "variables" - their values can vary!

```python
score = 10
print("Starting score:", score)

score = score + 5  # Add 5 to the current score
print("After bonus:", score)

score = score * 2  # Double the score
print("After doubling:", score)
```

## Shorthand Operators

Python has shortcuts for updating variables:

```python
x = 10

x += 5   # Same as: x = x + 5
print("After += 5:", x)

x -= 3   # Same as: x = x - 3
print("After -= 3:", x)

x *= 2   # Same as: x = x * 2
print("After *= 2:", x)

x /= 4   # Same as: x = x / 4
print("After /= 4:", x)
```

## Multiple Assignment

You can assign multiple variables at once:

```python
# Assign same value to multiple variables
a = b = c = 0
print(a, b, c)
```

```python
# Assign different values in one line
x, y, z = 1, 2, 3
print(x, y, z)
```

```python
# Swap two variables (Python trick!)
a = 10
b = 20
print("Before swap: a =", a, ", b =", b)

a, b = b, a  # Swap!
print("After swap: a =", a, ", b =", b)
```

# 4. Variable Naming Rules

## Rules (Must Follow)

1. **Must start with a letter or underscore** (`_`)
2. **Can only contain letters, numbers, and underscores**
3. **Case-sensitive** (`Age`, `age`, and `AGE` are different variables)
4. **Cannot be a Python keyword** (like `print`, `if`, `for`, etc.)

```python
# Valid variable names
name = "Alice"
age2 = 25
_private = "secret"
user_name = "bob123"

# Invalid variable names (uncomment to see errors)
# 2nd_place = "silver"   # Can't start with number
# my-name = "Alice"      # Can't use hyphens
# my name = "Alice"      # Can't use spaces
# print = "Hello"        # Can't use Python keywords
```

## Best Practices (Should Follow)

| Practice | Good | Bad |
|----------|------|-----|
| Use descriptive names | `student_age` | `x` |
| Use lowercase with underscores | `first_name` | `firstName` |
| Avoid single letters (except loops) | `counter` | `c` |
| Avoid confusing characters | `total_1` | `total_l` (l looks like 1) |

# 5. Type Conversion

Sometimes you need to convert one type to another.

## Converting Between Numbers

```python
# int() converts to integer (truncates decimals)
print(int(3.7))    # 3 (not rounded, just truncated)
print(int(3.2))    # 3
print(int(-2.9))   # -2
```

```python
# float() converts to float
print(float(5))    # 5.0
print(float("3.14"))  # 3.14 (from string)
```

```python
# round() for proper rounding
print(round(3.7))    # 4
print(round(3.2))    # 3
print(round(3.14159, 2))  # 3.14 (2 decimal places)
```

## Converting to/from Strings

```python
# str() converts to string
age = 25
message = "I am " + str(age) + " years old"
print(message)
```

```python
# int() and float() can convert strings to numbers
number_string = "42"
number = int(number_string)
print(number + 8)  # Now we can do math!
```

## Common Error: Type Mismatch

You can't directly combine strings and numbers:

```python
age = 25

# This causes an error:
# print("I am " + age + " years old")  # TypeError!

# Solution 1: Convert to string
print("I am " + str(age) + " years old")

# Solution 2: Use comma in print (adds spaces automatically)
print("I am", age, "years old")

# Solution 3: Use f-string (best method!)
print(f"I am {age} years old")
```
