# 1. What is a Program?

## The Big Picture

Most machines are designed to do **one thing**. A fan blows air. A toaster makes toast. A calculator does math.

A **computer** is different. It's designed to run **programs** - sets of instructions that tell it what to do.

> "A computer does two things: **perform calculations** and **remember the results**."
> 
> *- John V. Guttag*

With just these two abilities, a computer can:
- Play games
- Browse the internet
- Analyze data
- Create art
- And much more!

## What is Programming?

**Programming** is writing instructions for a computer to follow.

These instructions must be **very precise** because computers don't understand context or make assumptions like humans do.

For example, if you tell a human "make me a sandwich," they understand what you mean. But a computer needs step-by-step instructions:

1. Get two slices of bread
2. Get peanut butter from the cabinet
3. Open the peanut butter jar
4. Pick up a knife
5. Scoop peanut butter with the knife
6. Spread peanut butter on one slice of bread
7. ... and so on

This level of precision is what makes programming challenging - but also powerful!

# 2. What is Python?

## Why Python?

There are many programming languages (JavaScript, Java, C++, R, etc.). We use **Python** because:

| Reason | Explanation |
|--------|-------------|
| **Easy to learn** | Python reads almost like English |
| **Widely used** | Used by Google, Netflix, Instagram, NASA |
| **Great for data science** | Powerful libraries like Pandas, NumPy, scikit-learn |
| **Large community** | Lots of help available online |

## How Python Works

Computers only understand **binary** (1s and 0s). Python is a **high-level language** - it lets you write in something close to English, and then translates it to binary for the computer.

```
You write:     print("Hello")
                    |
                    v
Python translates to binary
                    |
                    v
Computer executes:  01001000 01100101 01101100 01101100 01101111
```

The **Python interpreter** is the program that does this translation.

# 3. Your First Program

## The print() Function

The `print()` function displays text on the screen. It's one of the most basic - and most useful - Python commands.

**Syntax:**
```python
print("Your message here")
```

Let's try it!

````python
# Your first Python program!
print("Hello, World!")
````

**Congratulations!** You just wrote your first Python program!

## Understanding the Code

Let's break down `print("Hello, World!")`:

| Part | Meaning |
|------|--------|
| `print` | The name of the function (built into Python) |
| `()` | Parentheses tell Python to "run" or "call" the function |
| `"Hello, World!"` | The **argument** - what we want to print |
| `""` | Quotation marks indicate this is text (a **string**) |

## Printing Multiple Lines

You can use multiple `print()` statements to display multiple lines:

````python
print("Line 1")
print("Line 2")
print("Line 3")
````

## Single vs Double Quotes

Python accepts both single quotes `'` and double quotes `"` for strings:

````python
print("Hello")  # Double quotes
print('Hello')  # Single quotes - same result!
````

````python
# 'It's name"
````

**When to use which?**

Use double quotes when your text contains a single quote (apostrophe):

````python
print("I'm learning Python!")  # Works!
# print('I'm learning Python!')  # This would cause an error
````

Use single quotes when your text contains double quotes:

````python
print('She said "Hello!"')  # Works!
````

## Printing Numbers

You can also print numbers - without quotes:

````python
print(42)       # An integer
print(3.14)     # A decimal number
print("42")     # This is a string "42", not the number 42
````

````python
42, "42", 4.2
````

````python
type(42), type("42"), type(4.2)
````

## Common Errors

Here are some mistakes beginners often make:

````python
# ERROR 1: Missing quotation marks
# print(Hello)  # This will cause an error!
hello='hello'
print(hello)
# ERROR 2: Mismatched quotes
# print("Hello')  # This will cause an error!

# ERROR 3: Forgetting parentheses (Python 3)
# print "Hello"  # This will cause an error!

# CORRECT:
print("Hello")
````

# 4. Comments

**Comments** are notes in your code that Python ignores. They help you (and others) understand what your code does.

Use the `#` symbol to create a comment:

````python
# This is a comment - Python ignores this line
print("Hello")  # You can also add comments after code

# Comments are useful for:
# - Explaining what your code does
# - Leaving notes for yourself
# - Temporarily disabling code
# This is a comment
# print('hi')
````

## Multi-line Comments

For longer comments, use triple quotes:

````python
"""
This is a multi-line comment.
It can span multiple lines.
Useful for longer explanations.
"""

print("The comment above is ignored")
````
