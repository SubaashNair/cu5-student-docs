# 1. Boolean Values

## What is a Boolean?

A **Boolean** is a data type that can only have one of two values:
- `True` (yes, correct, on)
- `False` (no, incorrect, off)

The name comes from mathematician George Boole, who developed Boolean logic.

> **Important**: In Python, `True` and `False` must be capitalized!

````python
# Boolean values in Python
print(True)
print(False)
````

````python
# We can store Boolean values in variables
is_raining = True
is_sunny = False

print("Is it raining?", is_raining)
print("Is it sunny?", is_sunny)
````

````python
# Check the type of a Boolean
print(type(True))
print(type(False))
````

# 2. Comparison Operators

## Comparing Values

**Comparison operators** let us compare two values. The result is always a Boolean (`True` or `False`).

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal | `5 >= 5` | `True` |
| `<=` | Less than or equal | `5 <= 3` | `False` |

> **Note**: `==` (double equals) is for comparison. `=` (single equals) is for assignment.

### Equal To (==)

````python
# Is 5 equal to 5?
print(5 == 5)
````

````python
# Is 5 equal to 3?
print(5 == 3)
````

````python
# Works with strings too!
print("hello" == "hello")
print("hello" == "Hello")  # Case matters!
````

### Not Equal To (!=)

````python
# Is 5 not equal to 3?
print(5 != 3)
````

````python
# Is 5 not equal to 5?
print(5 != 5)
````

### Greater Than (>) and Less Than (<)

````python
# Is 10 greater than 5?
print(10 > 5)
````

````python
# Is 3 less than 7?
print(3 < 7)
````

````python
# Is 5 greater than 5?
print(5 > 5)  # False - they are equal, not greater
````

### Greater Than or Equal (>=) and Less Than or Equal (<=)

````python
# Is 5 greater than or equal to 5?
print(5 >= 5)  # True - they are equal
````

````python
# Is 3 less than or equal to 5?
print(3 <= 5)
````

### Using Variables in Comparisons

````python
age = 18
voting_age = 18

# Can this person vote?
can_vote = age >= voting_age
print("Can vote:", can_vote)
````

````python
temperature = 35

# Is it hot? (above 30 degrees)
is_hot = temperature > 30
print("Is it hot?", is_hot)
````

# 3. Logical Operators

## Combining Conditions

Sometimes you need to check multiple conditions at once. **Logical operators** let you combine Boolean values:

| Operator | Meaning | Example |
|----------|---------|--------|
| `and` | Both must be True | `True and False` → `False` |
| `or` | At least one must be True | `True or False` → `True` |
| `not` | Reverses the value | `not True` → `False` |

### The `and` Operator

Both conditions must be `True` for the result to be `True`.

````python
print(True and True)    # Both true -> True
print(True and False)   # One false -> False
print(False and False)  # Both false -> False
````

````python
# Real example: Can you drive?
has_license = True
has_car = True

can_drive = has_license and has_car
print("Can drive:", can_drive)
````

````python
# Another example: Is the number in range 1-10?
number = 7

in_range = number >= 1 and number <= 10
print("Is", number, "between 1 and 10?", in_range)
````

### The `or` Operator

At least one condition must be `True` for the result to be `True`.

````python
print(True or True)    # At least one true -> True
print(True or False)   # At least one true -> True
print(False or False)  # None true -> False
````

````python
# Real example: Is it a weekend?
day = "Saturday"

is_weekend = day == "Saturday" or day == "Sunday"
print("Is it a weekend?", is_weekend)
````

### The `not` Operator

Reverses a Boolean value.

````python
print(not True)   # Opposite of True -> False
print(not False)  # Opposite of False -> True
````

````python
# Real example: Is the store open?
is_closed = True

is_open = not is_closed
print("Is the store open?", is_open)
````

# 4. The if Statement

## Making Decisions

The `if` statement runs code only when a condition is `True`.

**Syntax:**
```python
if condition:
    # This code runs only if condition is True
    do_something()
```

> **Important**: The code inside the `if` must be **indented** (usually 4 spaces)!

````python
# Simple if statement
temperature = 35

if temperature > 30:
    print("It's hot today!")
````

````python
# Nothing happens if condition is False
temperature = 20

if temperature > 30:
    print("It's hot today!")

print("Program continues here regardless")
````

````python
# Multiple lines inside an if
age = 21

if age >= 18:
    print("You are an adult.")
    print("You can vote.")
    print("You have adult responsibilities.")
````

```
               ▼
        ┌─────────────┐

        └──────┬──────┘                 │
               │ True                   │
               ▼                        │
        ┌─────────────┐                 │
        │ Run code    │                 │
        │ inside if   │                 │
        └──────┬──────┘                 │
               │                        │
               ▼◄───────────────────────┘
        ┌─────────────┐
        │  Continue   │
        │  program    │
        └─────────────┘
```

# 5. if-else

## Two Choices

Sometimes you want to do one thing if a condition is True, and something else if it's False.

**Syntax:**
```python
if condition:
    # Runs if condition is True
else:
    # Runs if condition is False
```

````python
age = 16

if age >= 18:
    print("You can vote.")
else:
    print("You cannot vote yet.")
````

````python
# Try changing the age
age = 21

if age >= 18:
    print("You can vote.")
else:
    print("You cannot vote yet.")
````

````python
# Even or odd number
number = 7

if number % 2 == 0:  # % gives the remainder
    print(number, "is even")
else:
    print(number, "is odd")
````

# 6. if-elif-else

## Multiple Choices

When you have more than two possibilities, use `elif` (short for "else if").

**Syntax:**
```python
if condition1:
    # Runs if condition1 is True
elif condition2:
    # Runs if condition1 is False and condition2 is True
elif condition3:
    # Runs if condition1 and condition2 are False, condition3 is True
else:
    # Runs if all conditions are False
```

> **Note**: Only ONE block of code runs - Python stops checking after finding a True condition.

````python
# Grade classification
score = 75

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
````

````python
# Traffic light
light = "yellow"

if light == "red":
    print("Stop!")
elif light == "yellow":
    print("Slow down...")
elif light == "green":
    print("Go!")
else:
    print("Invalid light color")
````

````python
# Temperature advice
temp = 25

if temp < 0:
    print("Freezing! Stay inside.")
elif temp < 10:
    print("Cold. Wear a jacket.")
elif temp < 20:
    print("Cool. A light sweater is fine.")
elif temp < 30:
    print("Nice weather!")
else:
    print("Hot! Stay hydrated.")
````

### Order Matters!

Conditions are checked from top to bottom. The first True condition wins.

````python
# WRONG order - score of 95 would print "Grade: C" first!
score = 95

# This is WRONG:
if score >= 70:
    print("WRONG: Grade C")  # This matches first!
elif score >= 80:
    print("WRONG: Grade B")
elif score >= 90:
    print("WRONG: Grade A")  # Never reached
````

````python
# CORRECT order - check highest first
score = 95

if score >= 90:
    print("CORRECT: Grade A")
elif score >= 80:
    print("CORRECT: Grade B")
elif score >= 70:
    print("CORRECT: Grade C")
````
