# Long Short-Term Memory (LSTM) and Gated Recurrent Units (GRU)

## The Vanishing Gradient Problem
As we saw with `SimpleRNN`, training on long sequences is difficult because gradients can become extremely small (vanish) or extremely large (explode) as they are backpropagated through time. This makes it hard for the network to learn long-term dependencies.

## 1. LSTM (Long Short-Term Memory)
LSTMs were designed to solve the vanishing gradient problem. They introduce a **cell state** that runs through the network with only minor linear interactions, allowing information to flow unchanged.

An LSTM unit has three gates:
1. **Forget Gate**: Decides what information to throw away from the cell state.
2. **Input Gate**: Decides which new information to store in the cell state.
3. **Output Gate**: Decides what to output based on the cell state.

## 2. GRU (Gated Recurrent Unit)
GRUs are a simplified variation of LSTMs. They combine the forget and input gates into a single **update gate** and merge the cell state and hidden state. They often achieve similar performance to LSTMs but are faster to compute.

## Project: Sentiment Analysis on IMDB Reviews
We will compare LSTM and GRU models on a binary classification task: predicting whether a movie review is positive or negative.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, GRU, Dense, Dropout

tf.random.set_seed(42)
```

## 3. Data Loading and Preprocessing
We will use the top 10,000 most frequent words and pad sequences to a length of 200 words.

```python
VOCAB_SIZE = 10000
MAX_LEN = 200

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=VOCAB_SIZE)

print(f"Training samples: {len(X_train)}")
print(f"Test samples: {len(X_test)}")

# Padding sequences
X_train = pad_sequences(X_train, maxlen=MAX_LEN)
X_test = pad_sequences(X_test, maxlen=MAX_LEN)

print(f"Padded shape: {X_train.shape}")
```

## 4. Building the LSTM Model
A standard text classification architecture:
1. **Embedding Layer**: Converts integer indices (words) into dense vectors of fixed size.
2. **LSTM Layer**: Processes the sequence.
3. **Dense Layer**: Valid for classification (Sigmoid activation).

```python
embedding_dim = 32

model_lstm = Sequential([
    Embedding(input_dim=VOCAB_SIZE, output_dim=embedding_dim, input_length=MAX_LEN),
    LSTM(64),
    Dense(1, activation='sigmoid')
])

model_lstm.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_lstm.summary()
```

## 5. Building the GRU Model
Exactly the same architecture, but swapping LSTM for GRU.

```python
model_gru = Sequential([
    Embedding(input_dim=VOCAB_SIZE, output_dim=embedding_dim, input_length=MAX_LEN),
    GRU(64),
    Dense(1, activation='sigmoid')
])

model_gru.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_gru.summary()
```

## 6. Training Comparison

```python
print("Training LSTM...")
history_lstm = model_lstm.fit(X_train, y_train, epochs=3, batch_size=64, validation_split=0.2, verbose=1)

print("\nTraining GRU...")
history_gru = model_gru.fit(X_train, y_train, epochs=3, batch_size=64, validation_split=0.2, verbose=1)
```

```python
# Plotting comparison
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history_lstm.history['accuracy'], label='LSTM Train')
plt.plot(history_lstm.history['val_accuracy'], label='LSTM Val')
plt.title('LSTM Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_gru.history['accuracy'], label='GRU Train')
plt.plot(history_gru.history['val_accuracy'], label='GRU Val')
plt.title('GRU Accuracy')
plt.legend()

plt.show()
```

## Conclusion
You should observe that both models perform well on this dataset. GRUs often train slightly faster and perform comparably on smaller datasets, while LSTMs may be better for very long sequences with complex dependencies.
