## 1. Beyond Sequential: The Functional API

So far, we've used the `Sequential` API. It's simple but limited: it allows only **one input** and **one output**, and layers must be stacked linearly.

For complex architectures (like ResNet, Inception, or multi-modal models), we use the **Functional API**.

### Concept: Model as a Graph
In the Functional API, you manipulate the data tensors directly. You treat layers as functions that take a tensor and return a tensor.

![Sequential vs Functional](https://external-content.duckduckgo.com/iu/?u=https%3A%2F%2Fmedia.geeksforgeeks.org%2Fwp-content%2Fuploads%2F20250626134023098301%2FSequential-vs-Functional-API-in-Keras.webp&f=1&nofb=1&ipt=16151e191815bc0bf1e3e1fd87a8055cb0211419a8b5f2fd28af4876a225f86b)

*Sequential (Linear) vs Functional (Branching/Graph)*

### Syntax Comparison

````python
from tensorflow.keras import layers, models, Input

# --- 1. Sequential Approach (Old) ---
seq_model = models.Sequential([
    Input(shape=(10,)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])

# --- 2. Functional Approach (New) ---
# A. Define Input Node
inputs = Input(shape=(10,))

# B. Pass input through layers (Layer is a function)
x = layers.Dense(32, activation='relu')(inputs)

# C. Define Output Node
outputs = layers.Dense(1)(x)

# D. Create Model by specifying Start and End points
func_model = models.Model(inputs=inputs, outputs=outputs)

func_model.summary()
````

## 4. Callbacks: Monitoring Training

Callbacks are objects that perform actions at various stages of training (e.g., at the end of every epoch).

### Essential Callbacks
1.  **ModelCheckpoint**: Saves the model *only if* it improves (e.g., Validation Loss decreases).
2.  **EarlyStopping**: Stops training if the model stops improving, preventing overfitting and wasting time.


![Callbacks Flow](https://external-content.duckduckgo.com/iu/?u=https%3A%2F%2Fassets.zilliz.com%2FFigure_9_Early_stopping_strategy_8bfda98553.png&f=1&nofb=1&ipt=6b9345737421e10ccc4ea310cd95e9365a9964ed6e0b3237f6b49e221b3ed647)

### Implementation
