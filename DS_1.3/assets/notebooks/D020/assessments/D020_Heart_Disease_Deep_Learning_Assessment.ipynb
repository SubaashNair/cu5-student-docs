# Deep Learning Case Study: Heart Disease Prediction

## Business Context

Heart disease is one of the leading causes of death worldwide. Early detection can save lives. In this case study, you will build a deep learning model to predict the presence of heart disease based on medical indicators.

**Your Mission**: As a data scientist at a healthcare analytics company, you need to build a production-ready deep learning model that can predict heart disease with high accuracy.

## Dataset

The **Heart Disease UCI dataset** contains 303 patient records with 14 attributes:

**Features**:
- `age`: Age in years
- `sex`: Sex (1 = male, 0 = female)
- `cp`: Chest pain type (0-3)
- `trestbps`: Resting blood pressure (mm Hg)
- `chol`: Serum cholesterol (mg/dl)
- `fbs`: Fasting blood sugar > 120 mg/dl (1 = true, 0 = false)
- `restecg`: Resting ECG results (0-2)
- `thalach`: Maximum heart rate achieved
- `exang`: Exercise induced angina (1 = yes, 0 = no)
- `oldpeak`: ST depression induced by exercise
- `slope`: Slope of peak exercise ST segment (0-2)
- `ca`: Number of major vessels colored by fluoroscopy (0-3)
- `thal`: Thalassemia (1 = normal, 2 = fixed defect, 3 = reversible defect)

**Target**:
- `target`: Heart disease presence (1 = disease, 0 = no disease)
