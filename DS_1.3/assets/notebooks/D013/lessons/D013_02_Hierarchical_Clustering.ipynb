# C2: Visualizing Hierarchies

## 1. Introduction

In unsupervised learning, understanding the relationships between data points is key. Two powerful techniques for visualizing these relationships are **Hierarchical Clustering** and **t-SNE**.

### Hierarchical Clustering
Unlike K-Means, which requires pre-specifying the number of clusters, hierarchical clustering builds a tree of clusters. It can be:
1.  **Agglomerative (Bottom-Up)**: Start with each point as a cluster and merge the closest pairs.
2.  **Divisive (Top-Down)**: Start with all points in one cluster and split them recursively.

We visualize the result using a **Dendrogram**.

![Dendrogram Concept](https://upload.wikimedia.org/wikipedia/commons/a/ad/Hierarchical_clustering_simple_diagram.svg)
*(Source: Wikimedia Commons - Dendrogram showing how clusters merge at different distances)*

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.manifold import TSNE

# Set plot style
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 6)
```

## 2. Hierarchical Clustering with Eurovision Data

We will look at how countries vote in the Eurovision Song Contest. Countries with similar cultures or geographies often vote similarly. Can we find these "voting blocs"?

We use `eurovision_2016.csv`. It contains 'From country', 'To country', and 'Televote Points'.

```python
# Load the dataset
df_euro = pd.read_csv('../../00-data/eurovision_2016.csv')

print("Original Data:")
print(df_euro.head())

# We need a matrix where rows are 'From country', columns are 'To country', and values are 'Televote Points'
# Pivot the data
votes_pivot = df_euro.pivot(index='From country', columns='To country', values='Televote Points')

# Fill missing values with 0 (if a country didn't give points, we assume 0)
votes_pivot = votes_pivot.fillna(0)

print("\nPivoted Voting Matrix:")
print(votes_pivot.head())
```

### Creating the Dendrogram
We will sample a subset of countries to make the dendrogram readable, or use the full set if manageable (~40 countries).

```python
# Let's use a sample to ensure readability, or all if we can fit them.
# We'll take a sample of 20 countries for better visualization clarity in this example.
sample_countries = votes_pivot.sample(n=30, random_state=42) 

# Calculate the linkage: using 'complete' linkage (max distance between clusters)
mergings = linkage(sample_countries.values, method='complete')

# Plot the dendrogram
plt.figure(figsize=(15, 8))
dendrogram(
    mergings,
    labels=sample_countries.index.values,
    leaf_rotation=90,
    leaf_font_size=10
)
plt.title('Hierarchical Clustering of Eurovision 2016 Votes')
plt.ylabel('Distance')
plt.show()
```

**Interpretation**: You should see countries grouping together. For example, Baltic states or Balkan states often appear in the same branches.

## 3. t-SNE for 2D Visualization

Hierarchical clustering gives a tree, but what if we want a 2D map? **t-SNE** (t-Distributed Stochastic Neighbor Embedding) is fantastic for this.

Steps:
1.  Create a `TSNE` model.
2.  `fit_transform` the data.
3.  Plot the resulting x and y coordinates.

*Note: t-SNE has a learning rate concept. It's often trial and error, but usually between 50-200. Here we use defaults or tweak slightly.*

```python
# Create a TSNE model with 2 components (2D)
# learning_rate=100 is a good starting point
model = TSNE(learning_rate=100, random_state=42, perplexity=10)

# Fit and transform the samples
transformed = model.fit_transform(sample_countries.values)

# Select the x and y coordinates of the embedded points
xs = transformed[:, 0]
ys = transformed[:, 1]

# Scatter plot
plt.figure(figsize=(12, 8))
plt.scatter(xs, ys)

# Annotate the points with country names
for x, y, country in zip(xs, ys, sample_countries.index):
    plt.annotate(country, (x, y), fontsize=9, alpha=0.75)

plt.title('t-SNE Map of Eurovision Voting')
plt.xlabel('t-SNE feature 1')
plt.ylabel('t-SNE feature 2')
plt.show()
```
