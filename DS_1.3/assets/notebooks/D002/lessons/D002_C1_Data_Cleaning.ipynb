## 2. Handling Missing Values

Missing data can reduce the power of a model or lead to biased estimates. We have several strategies to handle them:

### Strategies for Missing Data

| Strategy | When to Use | Pros | Cons |
|----------|-------------|------|------|
| **Drop Rows** | Few missing values, random missingness | Simple, preserves column structure | Loses data |
| **Drop Columns** | >50% missing, column not important | Reduces dimensionality | Loses information |
| **Impute (Mean/Median)** | Numerical data, random missingness | Preserves data volume | Can reduce variance |
| **Impute (Mode)** | Categorical data | Preserves data volume | May introduce bias |
| **Impute (Constant)** | Business logic requires specific value | Transparent | May not be realistic |

![Missing Data](https://miro.medium.com/v2/resize:fit:1400/1*9lCF8z5F8eF9yPzPq_A-qQ.png)

```python
# Check for missing values in each column
missing_counts = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)

missing_df = pd.DataFrame({'Missing Count': missing_counts, 'Missing %': missing_pct})
print("Missing Values Summary:")
print(missing_df[missing_df['Missing Count'] > 0])
```

### Strategy 1: Dropping Columns

If a column has too many missing values (e.g., >50% or completely empty), it might be better to drop it.

```python
# Drop 'lat' and 'lon' as they appear to be completely empty
df_clean = df.drop(columns=['lat', 'lon'])
print(f"Columns remaining: {df_clean.shape[1]}")
df_clean.head(2)
```

### Strategy 2: Imputation with Mode (for Categorical)

For categorical columns like `Difficulty`, we can fill missing values with the **mode** (most frequent value) or a placeholder like "Unknown".

```python
# Look at the raw categories BEFORE deciding anything
print("Raw value counts:")
print(df_clean['Difficulty'].value_counts(dropna=False))

# 'Easy' and 'Easy ' are the same difficulty typed two ways. Until the
# whitespace is stripped they are two categories, and every count, every mode
# and every one-hot column built from this comparison is wrong.
df_clean['Difficulty'] = df_clean['Difficulty'].str.strip()

print("\nAfter stripping whitespace:")
print(df_clean['Difficulty'].value_counts(dropna=False))
print("\nMode of Difficulty:", df_clean['Difficulty'].mode()[0])
```

```python
# Fill missing Difficulty with 'Unknown'
df_clean['Difficulty'] = df_clean['Difficulty'].fillna('Unknown')

# Verify changes
print("Value counts after imputation:")
print(df_clean['Difficulty'].value_counts())
```

## 3. Outlier Detection and Handling

**Outliers** are data points that differ significantly from other observations. They can arise from:
- Measurement errors
- Data entry errors
- Natural variation

### Why Handle Outliers?

| Impact | Description |
|--------|-------------|
| **Skew Statistics** | Mean, standard deviation become misleading |
| **Affect Models** | Linear models and distance-based algorithms are sensitive |
| **Mask Patterns** | Can hide true relationships in data |

### Common Detection Methods

1. **Visual Methods**: Box plots, scatter plots
2. **IQR Method**: Values outside 1.5 * IQR from Q1/Q3
3. **Z-Score Method**: Values more than 3 standard deviations from mean

### 3.1 Visual Detection with Box Plots

Box plots show the distribution of data and highlight potential outliers as points beyond the whiskers.

```python
# Let's use the wine dataset for outlier demonstration
wine_df = pd.read_csv('../../00-data/wine_types.csv')

# Select numerical columns for outlier analysis
numerical_cols = ['Alcohol', 'Malic acid', 'Magnesium', 'Proline']

# Create box plots
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, col in enumerate(numerical_cols):
    axes[i].boxplot(wine_df[col].dropna())
    axes[i].set_title(col)
    axes[i].set_ylabel('Value')
plt.suptitle('Box Plots for Outlier Detection', fontsize=14)
plt.tight_layout()
plt.show()
```

### 3.2 IQR Method (Interquartile Range)

The IQR method defines outliers as values that fall below Q1 - 1.5*IQR or above Q3 + 1.5*IQR.

$$IQR = Q3 - Q1$$
$$Lower\ Bound = Q1 - 1.5 \times IQR$$
$$Upper\ Bound = Q3 + 1.5 \times IQR$$

```python
def detect_outliers_iqr(data, column):
    """Detect outliers using IQR method."""
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    
    print(f"Column: {column}")
    print(f"  Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
    print(f"  Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
    print(f"  Number of Outliers: {len(outliers)}")
    
    return outliers, lower_bound, upper_bound

# Detect outliers in Magnesium
outliers, lb, ub = detect_outliers_iqr(wine_df, 'Magnesium')
```

### 3.3 Z-Score Method

The Z-score measures how many standard deviations a value is from the mean. Values with |Z| > 3 are typically considered outliers.

$$Z = \frac{x - \mu}{\sigma}$$

```python
from scipy import stats

def detect_outliers_zscore(data, column, threshold=3):
    """Detect outliers using the Z-score method.

    NOTE ON THE INDEX. stats.zscore(data[column].dropna()) returns positions in
    the DROPPED array, which stop matching the DataFrame the moment any row is
    missing. Keeping the index and reattaching it is what makes the result
    usable: what comes back below are real DataFrame labels.

    NOTE ON ddof. stats.zscore uses ddof=0 by default and pandas .std() uses
    ddof=1, so printing one beside the other compares two different numbers.
    Both are computed at ddof=0 here.
    """
    series = data[column].dropna()
    z_scores = np.abs(stats.zscore(series))
    outlier_index = series.index[z_scores > threshold]

    print(f"Column: {column}")
    print(f"  Mean: {series.mean():.2f}, Std: {series.std(ddof=0):.2f}")
    print(f"  Threshold: |Z| > {threshold}")
    print(f"  Number of Outliers: {len(outlier_index)}")

    return outlier_index

# Detect outliers in Magnesium using Z-score
outlier_idx = detect_outliers_zscore(wine_df, 'Magnesium')

# Because the index is real, the rows can be looked at directly
wine_df.loc[outlier_idx, ['Magnesium']]
```

### 3.4 Handling Outliers

Once detected, we can handle outliers in several ways:

| Method | Description | When to Use |
|--------|-------------|-------------|
| **Remove** | Delete outlier rows | When outliers are errors |
| **Cap/Clip** | Replace with boundary values | When you want to keep data but limit extremes |
| **Transform** | Apply log or other transformations | When data is naturally skewed |
| **Keep** | Leave as is | When outliers are valid data points |

```python
# Example: Capping outliers at the IQR bounds
def cap_outliers(data, column):
    """Cap outliers at IQR bounds."""
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    data_capped = data.copy()
    data_capped[column] = data_capped[column].clip(lower=lower_bound, upper=upper_bound)
    
    return data_capped

# Before capping
print("Before capping:")
print(f"  Max Magnesium: {wine_df['Magnesium'].max()}")

# After capping
wine_capped = cap_outliers(wine_df, 'Magnesium')
print("\nAfter capping:")
print(f"  Max Magnesium: {wine_capped['Magnesium'].max():.2f}")
```

## 4. Data Cleaning and Formatting

Real-world data is messy. Common issues include:
- Mixed text and numbers in a column
- Inconsistent formatting (capitalization, whitespace)
- Wrong data types

### Example: Cleaning the Length Column

In our hiking dataset, the `Length` column has values like "0.8 miles" which need to be converted to numbers.

```python
# Inspect unique values in Length
print("Unique values in Length:")
print(df_clean['Length'].unique())
```

```python
# Function to clean length
def clean_length(val):
    """Convert length string to numeric value."""
    if pd.isna(val):
        return np.nan
    val = str(val).lower()
    if 'various' in val:
        return np.nan  # Treat 'Various' as missing
    # Extract the number (remove 'miles', 'mile')
    val = val.replace('miles', '').replace('mile', '').strip()
    try:
        return float(val)
    except ValueError:
        return np.nan

# Apply the cleaning function
df_clean['Length_Numeric'] = df_clean['Length'].apply(clean_length)

# Check the result
print("Before and After cleaning:")
df_clean[['Length', 'Length_Numeric']].head()
```

```python
# Fill remaining NaNs with the median (more robust to outliers than mean)
median_length = df_clean['Length_Numeric'].median()
df_clean['Length_Numeric'] = df_clean['Length_Numeric'].fillna(median_length)

print(f"Filled missing lengths with median: {median_length:.2f} miles")
```

Ensuring correct data types is essential for:

## 5b. Duplicate Rows

Duplicates arrive when an export runs twice, when a join fans out, or when the same
record is entered by two people. They inflate every count and shift every average,
and nothing about the DataFrame looks wrong.

| Call | What it answers |
|---|---|
| `df.duplicated()` | boolean per row: is this an exact copy of an earlier row? |
| `df.duplicated().sum()` | how many |
| `df.duplicated(subset=['id'])` | copies by key, even when other columns differ |
| `df.drop_duplicates()` | keep the first of each group |

Check before you drop. `drop_duplicates()` on a table where repeats are legitimate
(one row per visit, not per patient) silently deletes real data.

```python
# Exact duplicate rows
print(f"Exact duplicate rows: {df.duplicated().sum()}")

# Duplicates by key alone, which is the more common case
print(f"Repeated Prop_ID values: {df.duplicated(subset=['Prop_ID']).sum()}")

# What dropping them would change
print(f"\nRows before: {len(df)}")
print(f"Rows after drop_duplicates(): {len(df.drop_duplicates())}")
```
