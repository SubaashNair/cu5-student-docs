## 2. Understanding Tensors

A **Tensor** is a generalization of scalars, vectors, and matrices to higher dimensions. It is simply a collection of numbers arranged in a specific shape.

### Terminology:
*   **Scalar (Rank 0):** A single number. Shape: `()`
*   **Vector (Rank 1):** An array of numbers. Shape: `(D0,)`
*   **Matrix (Rank 2):** A 2D array of numbers. Shape: `(D0, D1)`
*   **3D Tensor (Rank 3):** A cube of numbers. Shape: `(D0, D1, D2)`

## 3. Constants in TensorFlow

A **Constant** is a type of tensor whose value **cannot be changed** once it is created. They are **immutable**.

Constants are useful for storing fixed values like hyperparameters, input data, or configuration settings that should remain static during execution.

### Creating Constants
We use `tf.constant()` to create constants. You can also use convenience functions like `tf.zeros()`, `tf.ones()`, and `tf.fill()`.

````python
# Define a scalar constant (Rank 0)
c_scalar = tf.constant(5)
print("Scalar:", c_scalar)
print("Shape:", c_scalar.shape)

# Define a vector constant (Rank 1)
c_vector = tf.constant([1, 2, 3])
print("\nVector:", c_vector)
print("Shape:", c_vector.shape)

# Define a matrix constant (Rank 2)
c_matrix = tf.constant([[1, 2], [3, 4]])
print("\nMatrix:", c_matrix)
print("Shape:", c_matrix.shape)
````

### Convenience Functions
Creating tensors of all zeros, ones, or a specific value.

````python
# Tensor of zeros
zeros = tf.zeros(shape=[2, 3])  # 2 rows, 3 columns
print("Zeros:\n", zeros.numpy())

# Tensor of ones
ones = tf.ones(shape=[3, 1])
print("\nOnes:\n", ones.numpy())

# Tensor filled with a specific value
filled = tf.fill(dims=[2, 2], value=7)
print("\nFilled with 7:\n", filled.numpy())
````

## 4. Variables in TensorFlow

A **Variable** is a special type of tensor that is **mutable**, meaning its value **can be changed** after creation.

Variables are crucial in Deep Learning because they hold the **weights and biases** of a neural network. These parameters must update during training to minimize the error (loss).

### Key Differences: Constants vs. Variables

| Feature | `tf.constant` | `tf.Variable` |
| :--- | :--- | :--- |
| **Mutability** | Immutable (cannot change) | Mutable (can change) |
| **Usage** | Fixed data, Hyperparameters | Trainable parameters (Weights, Biases) |
| **Creation** | `tf.constant()` | `tf.Variable()` |
| **Optimizable** | No | Yes (via Gradient Descent) |

````python
# Create a Variable
var_1 = tf.Variable([1.0, 2.0, 3.0], dtype=tf.float32)

print("Original Variable:", var_1.numpy())

# Update the variable using assign()
# Note: The new value must have the same shape and dtype
var_1.assign([10.0, 20.0, 30.0])

print("Updated Variable:", var_1.numpy())
````

### Variable Operations
You can also perform operations like `assign_add` (+=) and `assign_sub` (-=).

````python
v = tf.Variable(10)

# Add 5 to the variable
v.assign_add(5) 
print("After assign_add(5):", v.numpy())  # Should be 15

# Subtract 3 from the variable
v.assign_sub(3)
print("After assign_sub(3):", v.numpy())  # Should be 12
````

## 5. Basic Operations

TensorFlow supports standard mathematical operations. You can use element-wise operations like addition and multiplication, as well as matrix operations.

### Element-wise Operations
Operations happen element by element. Tensors must usually have compatible shapes.

````python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

# Addition
add_res = tf.add(a, b)
print("Addition:", add_res.numpy())
# Alternatively: a + b

# Multiplication (Element-wise)
mul_res = tf.multiply(a, b)
print("Element-wise Multiplication:", mul_res.numpy())
# Alternatively: a * b
````

### Matrix Multiplication
To multiply matrices (dot product), use `tf.matmul()` or the `@` operator. 
**Rule:** The number of columns in the first matrix must equal the number of rows in the second matrix.

````python
mat_a = tf.constant([[1, 2], 
                     [3, 4]])  # Shape (2, 2)

mat_b = tf.constant([[5, 6], 
                     [7, 8]])  # Shape (2, 2)

# Matrix Multiplication
matmul_res = tf.matmul(mat_a, mat_b)
print("Matrix Multiplication Result:\n", matmul_res.numpy())
````

### Reduction (Summing dimensions)
`tf.reduce_sum()` is used to sum elements across dimensions. This is commonly used in loss functions.

````python
t = tf.constant([[1, 2, 3], 
                 [4, 5, 6]])

print("Original Tensor:\n", t.numpy())

# Sum all elements
total_sum = tf.reduce_sum(t)
print("Total Sum:", total_sum.numpy())

# Sum along columns (axis 0) -> Collapses rows
col_sum = tf.reduce_sum(t, axis=0)
print("Column Sum (axis=0):", col_sum.numpy())

# Sum along rows (axis 1) -> Collapses columns
row_sum = tf.reduce_sum(t, axis=1)
print("Row Sum (axis=1):", row_sum.numpy())
````

## 6. Advanced Operations

### Reshaping
Reshaping is critical in Deep Learning, for example, when flattening an image (2D) into a vector (1D) for a dense layer.

````python
# Create a 1D tensor with 6 elements
flat = tf.constant([1, 2, 3, 4, 5, 6])

# Reshape to 2x3
reshaped_2x3 = tf.reshape(flat, [2, 3])
print("Reshaped to 2x3:\n", reshaped_2x3.numpy())

# Reshape to 3x2
reshaped_3x2 = tf.reshape(flat, [3, 2])
print("Reshaped to 3x2:\n", reshaped_3x2.numpy())
````

### Gradients (Automatic Differentiation)
TensorFlow can automatically compute the gradient (slope) of operations, which is the engine behind backpropagation in training neural networks. We use `tf.GradientTape()` to record operations for automatic differentiation.

````python
x = tf.Variable(3.0)

# We want to find the gradient of y = x^2 with respect to x
# At x = 3, y = 9. The derivative dy/dx = 2x. So at x=3, gradient should be 6.

with tf.GradientTape() as tape:
    y = x ** 2

# Calculate gradient
dy_dx = tape.gradient(y, x)

print(f"y = x^2 where x = {x.numpy()}")
print(f"dy/dx = {dy_dx.numpy()}")
````
