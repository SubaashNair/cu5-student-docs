## 2. Understanding Tensors

A **Tensor** is a generalization of scalars, vectors, and matrices to higher dimensions. It is simply a collection of numbers arranged in a specific shape.

### Terminology:
*   **Scalar (Rank 0):** A single number. Shape: `()`
*   **Vector (Rank 1):** An array of numbers. Shape: `(D0,)`
*   **Matrix (Rank 2):** A 2D array of numbers. Shape: `(D0, D1)`
*   **3D Tensor (Rank 3):** A cube of numbers. Shape: `(D0, D1, D2)`

## 4. Variables in TensorFlow

A **Variable** is a special type of tensor that is **mutable**, meaning its value **can be changed** after creation.

Variables are crucial in Deep Learning because they hold the **weights and biases** of a neural network. These parameters must update during training to minimize the error (loss).

### Key Differences: Constants vs. Variables

| Feature | `tf.constant` | `tf.Variable` |
| :--- | :--- | :--- |
| **Mutability** | Immutable (cannot change) | Mutable (can change) |
| **Usage** | Fixed data, Hyperparameters | Trainable parameters (Weights, Biases) |
| **Creation** | `tf.constant()` | `tf.Variable()` |
| **Optimizable** | No | Yes (via Gradient Descent) |

### Gradients (Automatic Differentiation)
TensorFlow can automatically compute the gradient (slope) of operations, which is the engine behind backpropagation in training neural networks. We use `tf.GradientTape()` to record operations for automatic differentiation.

````python
x = tf.Variable(3.0)

# We want to find the gradient of y = x^2 with respect to x
# At x = 3, y = 9. The derivative dy/dx = 2x. So at x=3, gradient should be 6.

with tf.GradientTape() as tape:
    y = x ** 2

# Calculate gradient
dy_dx = tape.gradient(y, x)

print(f"y = x^2 where x = {x.numpy()}")
print(f"dy/dx = {dy_dx.numpy()}")
````
